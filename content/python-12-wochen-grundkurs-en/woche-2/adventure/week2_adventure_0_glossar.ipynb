# 📖 Glossary – 💎 Week 2 – Data Types and Variables: The Four Magic Elements

| Term | Meaning | Example |
|------|---------|--------|
| `f-String` | Modern text formatting with variables directly in the text | `f"Hello {name}"` |
| `Float` | Decimal number | `3.14` |
| `type()` | Check the data type of a variable | `type(42)` → `<class 'int'>` |
| `len()` | Length of a text or list | `len("Hello")` → `5` |
| `int()` | Convert to whole number | `int("42")` → `42` |
| `float()` | Convert to decimal number | `float("3")` → `3.0` |
| `str()` | Convert to text | `str(42)` → `"42"` |
| `bool()` | Convert to boolean value | `bool(0)` → `False` |
| `.upper()` | Convert text to uppercase | `"hello".upper()` → `"HELLO"` |
| `.lower()` | Convert text to lowercase | `"HELLO".lower()` → `"hello"` |
| `.strip()` | Remove spaces at the edges | `" hi ".strip()` → `"hi"` |
| `.split()` | Split text at a character | `"a,b".split(",")` → `["a","b"]` |
| `.replace()` | Replace part of text with another | `"Hello".replace("H","h")` → `"hello"` |

### From Week 1

| Term | Meaning | Example |
|------|---------|--------|
| `print()` | Outputs text | `print("Hello")` |
| **Variable** | Container with a name | `name = "Aria"` |
| `str()` | Converts number to text | `str(42)` → `"42"` |
| `input()` | Reads user input | `name = input("Name? ")` |

```python
name = "Aria"
level = 5

# f-String – variable directly in text
print(f"Hero: {name}, Level: {level}")

# type() – check type
print(type(name))    # <class 'str'>
print(type(level))   # <class 'int'>

# len() – length
print(len(name))     # 4

# String methods
print(name.upper())  # ARIA
print(name.lower())  # aria
```
