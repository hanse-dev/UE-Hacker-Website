## 📚 Chapter 1: The Spell Import – Borrowing Foreign Magic

**What it is:** A module is a collection of predefined spells (functions) that other mages have already written. With `import` you borrow them.

**The structure of an import:**
```python
# Import the entire module
import random

# Import with a nickname
import random as r

# Import only specific spells
from string import ascii_letters
```

**Step-by-step explanation:**
1. **`import`** - Magic word for loading
2. **`random`** - Name of the library
3. **`as`** - Optional nickname
4. **`from ... import ...`** - Imports only individual spells

**Why the Archivist cares about this:**
- 🎯 No need to reinvent spells
- 🔍 Use tested, reliable incantations
- ⚡ Your code stays organised
- 🗝️ Access to specialised magic you would otherwise lack

````python
# Example 1: Importing modules
import random
import time

print(f"Random number: {random.randint(1, 10)}")

# Import with a nickname
import random as r
print(f"Again (via nickname r): {r.randint(1, 10)}")

# Import only specific spells
from string import ascii_letters
print(f"First 5 letters of the rune alphabet: {ascii_letters[:5]}")
````

````python
# Example 2: Comparing different import methods
import random
print("Method 1 - import random:")
print(f"  random.randint(1, 6): {random.randint(1, 6)}")

import random as r
print("\nMethod 2 - import random as r:")
print(f"  r.randint(1, 6): {r.randint(1, 6)}")

from random import randint
print("\nMethod 3 - from random import randint:")
print(f"  randint(1, 6): {randint(1, 6)}")
````

````python
# Example 3: Using several modules together
import random
import time

print("=== The Archivist tests your first invocation ===")
waiting_time = random.uniform(0.5, 1.5)
print(f"A seal spell begins to activate ({waiting_time:.1f} seconds)...")
time.sleep(waiting_time)
print("The first seal lights up!")
````

## 🎲 Chapter 2: The random Module – Oracle of Chance

**The most important oracle spells:**

**Generating numbers:**
- `random.randint(a, b)` - Random whole number from a to b
- `random.uniform(a, b)` - Random decimal number from a to b
- `random.random()` - Random number between 0 and 1

**Choosing from collections:**
- `random.choice(list)` - One random element
- `random.sample(list, k)` - k random, distinct elements
- `random.shuffle(list)` - Shuffle a list (changes the list itself!)

**Reproducible magic:**
- `random.seed(number)` - Makes "random" results repeatable (handy for testing)

````python
# Example 1: Consulting the dice oracle
import random

print("=== The Dice Oracle ===")
print(f"Roll 1: {random.randint(1, 6)}")
print(f"Roll 2: {random.randint(1, 6)}")
print(f"Decimal 0-1: {random.random():.3f}")
print(f"Decimal 10-20: {random.uniform(10, 20):.2f}")

# Reproducible with a seed
random.seed(42)
print(f"\nWith seed 42: {random.randint(1, 100)}")
random.seed(42)
print(f"With seed 42 (again): {random.randint(1, 100)} (same result!)")
````

````python
# Example 2: Shelves, scrolls and shuffling
import random

shelves = ["Shelf of the Elements", "Shelf of Beasts", "Shelf of Stars", "Shelf of Shadows", "Shelf of Heroes"]
print(f"All shelves: {shelves}")

# One random shelf
chosen_shelf = random.choice(shelves)
print(f"Your shelf: {chosen_shelf}")

# Draw three distinct scrolls (no repeats!)
drawn_scrolls = random.sample(shelves, 3)
print(f"Drawn scrolls: {drawn_scrolls}")

# Shuffle the order
order = shelves.copy()
random.shuffle(order)
print(f"Shuffled reading order: {order}")
````

## 🔤 Chapter 3: The string Module – Rune Forge

**What it is:** The `string` module gives you ready-made character sets, which you can use to forge rune keys (passwords).

**The most important character sets:**
- `string.ascii_letters` - All letters a-z and A-Z
- `string.ascii_lowercase` - Only lowercase letters
- `string.ascii_uppercase` - Only uppercase letters
- `string.digits` - All digits 0-9

**Combined with `random.choice()` you can forge random rune keys from them!**

````python
# Example: Forging a rune key
import random
import string

alphabet = string.ascii_letters + string.digits
print(f"Rune alphabet ({len(alphabet)} characters): {alphabet}")

# Draw a single rune
print(f"\nA random rune: {random.choice(alphabet)}")

# Forge a whole key (8 runes)
key = "".join(random.choice(alphabet) for _ in range(8))
print(f"Your rune key: {key}")
````

## ⏳ Chapter 4: Time Magic – time and datetime

**What it is:** Some doors in the archive only open if you wait – or only at certain times of day.

**The most important time spells:**
- `time.sleep(seconds)` - Pause the program for a while
- `time.time()` - Retrieve the current timestamp
- `from datetime import datetime` and `datetime.now()` - Current date and time
- `datetime.now().hour` - Just the current hour

````python
# Example: A waiting ritual and the current hour
import time
from datetime import datetime

print("A ritual begins to take effect...")
time.sleep(1)
print("The ritual is complete!")

now = datetime.now()
print(f"\nCurrent time: {now.strftime('%H:%M:%S')}")

if 6 <= now.hour < 18:
    print("It is day – the Sun Gates of the archive stand open.")
else:
    print("It is night – only the Moon Gates of the archive are open.")
````

## 🧮 Chapter 5: The math Module – The Calculation Chamber

**What it is:** The **Calculation Chamber** holds the archive's most powerful number spells – constants and functions for everything beyond simple arithmetic.

**The most important Calculation Chamber tools:**

**Constants:**
- `math.pi` - The number Pi (3.14159...)
- `math.e` - Euler's number (2.71828...)
- `math.inf` - Infinity

**Basic tools:**
- `math.sqrt(x)` - Square root
- `math.pow(x, y)` - x to the power of y
- `math.ceil(x)` - Round up
- `math.floor(x)` - Round down
- `math.fabs(x)` - Absolute value

````python
# 🔍 Exercises with math constants
import math

print("=== Mathematical Constants ===")
print(f"Pi (π): {math.pi}")
print(f"Euler's number (e): {math.e}")
print(f"Infinity: {math.inf}")
print(f"Tau (2π): {math.tau}")

# Practical application: circle calculations
radius = 10
circumference = 2 * math.pi * radius
area = math.pi * radius ** 2
print(f"\nCircle with r={radius}:")
print(f"  Circumference: {circumference:.2f}")
print(f"  Area: {area:.2f}")
````

````python
# Example 2: Basic mathematical tools
import math

print("=== Mathematical Tools ===")

# Roots and powers
print(f"sqrt(16): {math.sqrt(16)}")
print(f"pow(2, 3): {math.pow(2, 3)}")
print(f"2**3: {2**3}")

# Rounding
number = 3.7
print(f"\nNumber: {number}")
print(f"round({number}): {round(number)}")
print(f"math.ceil({number}): {math.ceil(number)}")
print(f"math.floor({number}): {math.floor(number)}")

# Absolute value
print(f"\nmath.fabs(-5.5): {math.fabs(-5.5)}")
print(f"abs(-5.5): {abs(-5.5)}")
````

````python
# Example 3: Advanced tools
import math

print("=== Advanced Tools ===")

# Trigonometric functions (in radians!)
angle_deg = 45
angle_rad = math.radians(angle_deg)
print(f"{angle_deg}° = {angle_rad:.3f} rad")
print(f"sin({angle_deg}°): {math.sin(angle_rad):.3f}")
print(f"cos({angle_deg}°): {math.cos(angle_rad):.3f}")
print(f"tan({angle_deg}°): {math.tan(angle_rad):.3f}")

# Logarithms
print(f"\nlog(10): {math.log(10):.3f}")
print(f"log10(100): {math.log10(100)}")
print(f"log2(8): {math.log2(8)}")

# Special tools
print(f"\nfactorial(5): {math.factorial(5)}")
print(f"gcd(24, 36): {math.gcd(24, 36)}")
print(f"lcm(12, 15): {math.lcm(12, 15)}")
````
