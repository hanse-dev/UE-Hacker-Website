# 🐛 Debug-Quest – 🏰 Woche 10 – OOP Grundlagen: Die magischen Blaupausen!

Ein junger Magier hat versucht, seine ersten Klassen zu erstellen, aber etwas läuft schief! Finde die Fehler:

### 🐛 Bug #1

Was ist falsch? Finde und behebe den Fehler!

```python
class schwertkämpfer:
    def __init__(self, name):
        self.name = name

kämpfer = schwertkämpfer()
print(kämpfer.name)
```

### 🐛 Bug #2

Was ist falsch? Finde und behebe den Fehler!

```python
class Magier:
    def __init(name, mana):
        name = name
        mana = mana

merlin = Magier("Merlin", 150)
print(merlin.name)
```

### 🐛 Bug #3

Was ist falsch? Finde und behebe den Fehler!

```python
class Drache:
    def __init__(self, name):
        self.name = name
    
    def vorstellen():
        print(f"Ich bin {self.name}!")

drache = Drache("Smaug")
drache.vorstellen()
```
