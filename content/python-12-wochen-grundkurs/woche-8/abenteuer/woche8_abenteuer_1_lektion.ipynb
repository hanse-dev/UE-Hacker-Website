## Datenstruktur 1: Eigenschaften mit Namen

**Was es ist:** Ein Dictionary ist wie ein **Steckbrief**: Zu jeder **Eigenschaft** (Schlüssel) gehört ein **Wert**. Du erstellst ihn mit geschweiften Klammern.

**Der Aufbau eines Steckbriefs:**
```python
# Leeren Steckbrief anlegen
leerer_steckbrief = {}

# Steckbrief mit Eigenschaften füllen
held = {"name": "Aria", "klasse": "Magierin", "level": 15}
```

**Schritt-für-Schritt-Erklärung:**
1. **`{}`** – Geschweifte Klammern definieren einen Steckbrief
2. **`:`** – Doppelpunkt trennt Eigenschaftsname (Schlüssel) vom Wert
3. **`,`** – Kommas trennen die Einträge
4. **`[key]`** – Mit dem Schlüssel greifst du auf den Wert zu

**Warum so nützlich:**
- 🎯 Eigenschaften mit Namen statt Position (wie bei Listen)
- 🔍 Schneller Zugriff über den Schlüssel
- ⚡ Flexibel erweiterbar und änderbar
- 🎲 Verschiedene Werttypen pro Eigenschaft möglich

````python
# Beispiel 1: Einfache Steckbriefe erstellen
# Leerer Steckbrief
leerer_steckbrief = {}
print(f"Leerer Steckbrief: {leerer_steckbrief}")

# Helden-Steckbrief
held = {
    "name": "Aria",
    "klasse": "Magierin",
    "level": 15,
    "lebenspunkte": 120
}
print(f"Helden-Steckbrief: {held}")

# Monster-Steckbrief
monster = {
    "name": "Drache",
    "typ": "Feuer",
    "staerke": 50,
    "beute": True
}
print(f"Monster-Steckbrief: {monster}")
````

````python
# Beispiel 2: Auf Steckbriefe zugreifen
held = {
    "name": "Aria",
    "klasse": "Magierin",
    "level": 15,
    "lebenspunkte": 120
}

print("=== Helden-Steckbrief auslesen ===")
print(f"Helden-Name: {held['name']}")
print(f"Klasse: {held['klasse']}")
print(f"Level: {held['level']}")
print(f"Lebenspunkte: {held['lebenspunkte']}")

# Mit get() Methode (sicherer)
print(f"\nMit get(): {held.get('name')}")
print(f"Nicht existierend: {held.get('mana', 'Unbekannt')}")
````

````python
# Beispiel 3: Steckbriefe ändern und erweitern
held = {
    "name": "Aria",
    "klasse": "Magierin",
    "level": 15
}

print(f"Original: {held}")

# Wert ändern
held["level"] = 16
print(f"Nach Level-Up: {held}")

# Neue Eigenschaft hinzufügen
held["mana"] = 80
print(f"Nach Mana hinzufügen: {held}")

# Eintrag entfernen
entfernt = held.pop("klasse")
print(f"Entfernt: {entfernt}")
print(f"Nach entfernen: {held}")
````

## Datenstruktur 1b

Um alle Einträge eines Dictionaries zu verarbeiten, gibt es drei Methoden:

| Methode | Liefert | Beispiel |
|---------|---------|----------|
| `.keys()` | alle Schlüssel | `for k in held.keys()` |
| `.values()` | alle Werte | `for v in held.values()` |
| `.items()` | Schlüssel + Wert als Paar | `for k, v in held.items()` |

`.items()` ist am nützlichsten – du bekommst beides auf einmal.

````python
held = {
    "name": "Aria",
    "klasse": "Magierin",
    "level": 15,
    "lebenspunkte": 120
}

# .keys() – nur Schlüssel
print("Eigenschaften:", list(held.keys()))

# .values() – nur Werte
print("Werte:", list(held.values()))

# .items() – Schlüssel + Wert
print("\n=== Helden-Steckbrief ===")
for eigenschaft, wert in held.items():
    print(f"  {eigenschaft}: {wert}")
````

## Datenstruktur 2: Die magische Kapsel

**Die wichtigsten Artefakt-Eigenschaften:**

**Erstellung:**
- `artefakt = (1, 2, 3)` - Mit runden Klammern
- `artefakt = 1, 2, 3` - Ohne Klammern
- `artefakt = (1,)` - Ein Element mit Komma

**Unveränderlichkeit:**
- Einmal erstellt, können Artefakte nicht geändert werden
- Perfekt für ewige Schätze
- Magisch geschützt vor Veränderung

````python
# 🔍 Übungen mit Artefakten

# Beispiel 1: Artefakte erstellen
koordinaten = (100, 200)
print(f"Koordinaten: {koordinaten}")

# Ohne Klammern
position = 10, 20
print(f"Position: {position}")

# Ein Element
einzel = (42,)
print(f"Einzelnes Artefakt: {einzel}")

# Zugriff auf Elemente
print(f"\nX-Koordinate: {koordinaten[0]}")
print(f"Y-Koordinate: {koordinaten[1]}")
````

````python
# Beispiel 2: Artefakte sind unveränderlich
zauber_status = ("Aktiv", "Stabil", "Mächtig")
print(f"Zauber-Status: {zauber_status}")

# Zugriff funktioniert
print(f"Status: {zauber_status[0]}")

# Ändern funktioniert NICHT!
try:
    zauber_status[0] = "Inaktiv"
except TypeError as e:
    print(f"Fehler: {e}")

# Aber Artefakte in Listen können geändert werden
zauber = [
    ("Feuerball", "Aktiv"),
    ("Heilung", "Stabil"),
    ("Schild", "Mächtig")
]
zauber[1] = ("Heilung", "Inaktiv")
print(f"\nZauber nach Änderung: {zauber}")
````

````python
# Beispiel 3: Artefakt unpacking und Methoden
held_daten = ("Aria", "Magierin", 15, 120)

# Artefakt unpacking
name, klasse, level, lebenspunkte = held_daten
print("=== Artefakt Unpacking ===")
print(f"Name: {name}")
print(f"Klasse: {klasse}")
print(f"Level: {level}")
print(f"Lebenspunkte: {lebenspunkte}")

# Artefakt-Methoden
print(f"\nAnzahl Elemente: {len(held_daten)}")
print(f"Index von 'Magierin': {held_daten.index('Magierin')}")
print(f"Anzahl 'Magierin': {held_daten.count('Magierin')}")
````

## ✨ Verschachtelte Strukturen – Steckbriefe im Steckbrief

**Die mächtigsten Kombinationen:**

**Steckbrief in Steckbrief:**
- `person = {"name": "Alex", "ausruestung": {"waffe": "X", "ruestung": "Y"}}`

**Liste von Steckbriefen:**
- `helden = [{"name": "A", "klasse": "M"}, {"name": "B", "klasse": "K"}]`

**Artefakt (Tupel) als Steckbrief-Schlüssel:**
- `positionen = {(10, 20): "Punkt A", (30, 40): "Punkt B"}`

````python
# Beispiel 1: Steckbrief in Steckbrief
dungeon = {
    "name": "Verdammnis-Festung",
    "location": {
        "region": "Schattenlande",
        "koordinaten": (100, 200),
        "reich": "Pyralia"
    },
    "gefaehrten": {
        "boss": "Lich-König",
        "monster": "Skelette",
        "fallen": "Magisch"
    }
}

print("=== Verschachtelter Steckbrief ===")
print(f"Dungeon: {dungeon['name']}")
print(f"Region: {dungeon['location']['region']}")
print(f"Koordinaten: {dungeon['location']['koordinaten']}")
print(f"Boss: {dungeon['gefaehrten']['boss']}")
````

````python
# Beispiel 2: Liste von Steckbriefen
helden = [
    {
        "name": "Aria",
        "klasse": "Magierin",
        "level": 15,
        "erfahrung": 2500
    },
    {
        "name": "Thorin",
        "klasse": "Krieger",
        "level": 18,
        "erfahrung": 3200
    },
    {
        "name": "Luna",
        "klasse": "Schurkin",
        "level": 12,
        "erfahrung": 1800
    }
]

print("=== Helden-Liste ===")
for held in helden:
    print(f"{held['name']} - {held['klasse']} (Level {held['level']})")

# Nach Level filtern
starke = [h for h in helden if h['level'] > 14]
print(f"\nStarke Helden: {[h['name'] for h in starke]}")
````

````python
# Beispiel 3: Komplexe Strukturen
# Artefakt (Tupel) als Steckbrief-Schlüssel
schatz_karte = {
    (100, 200): "Goldtruhe",
    (300, 400): "Magie-Truhe",
    (500, 600): "Artefakt-Truhe"
}

print("=== Schatz-Karte (Tupel als Schlüssel) ===")
for position, schatz in schatz_karte.items():
    print(f"{position}: {schatz}")

# Komplexes Questsystem
quests = {
    "aktiv": [
        {
            "id": 1,
            "ziel": "Drache besiegen",
            "belohnung": ("Gold", 1000, "XP", 500),
            "status": "In Bearbeitung"
        }
    ],
    "abgeschlossen": [
        {
            "id": 2,
            "ziel": "Schatz finden",
            "belohnung": ("Silber", 500, "XP", 250),
            "status": "Erfolg"
        }
    ]
}

print(f"\nAktive Quests: {len(quests['aktiv'])}")
print(f"Abgeschlossene Quests: {len(quests['abgeschlossen'])}")
````
