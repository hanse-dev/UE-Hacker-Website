### 🐛 Bug #2

```python
# 🐛 Bug #2: Falscher Konstruktor

class Cyborg:
    def __init(name, power_level):
        name = name
        power_level = power_level

terminator = Cyborg("T-800", 9000)
print(terminator.name)
```

### 🐛 Bug #3

**Was passiert:** Methoden in Klassen brauchen immer `self` als ersten Parameter. Ohne `self` weiß die Methode nicht, zu welchem Objekt sie gehört.

```python
# 🐛 Bug #3: Methode ohne self

class Roboter:
    def __init__(self, bezeichnung):
        self.bezeichnung = bezeichnung
    
    def bewegen():
        print(f"{self.bezeichnung} bewegt sich! 🤖")

roboter = Roboter("R2-D2")
roboter.bewegen()
```
