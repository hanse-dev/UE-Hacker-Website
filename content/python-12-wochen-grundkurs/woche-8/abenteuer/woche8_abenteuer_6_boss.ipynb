### ⭐⭐⭐⭐☆ Boss-Quest 1: Der Schatz-Steckbrief

**Belohnung:** 500 XP + Schatzkarte

Das Gildenarchiv braucht Steckbriefe für alle gefundenen Schätze!

**Schritt 1 – Steckbrief anlegen:**
Erstelle ein Dictionary für einen Schatz mit mindestens 4 Eigenschaften (z.B. Name, Wert, Ort, Seltenheit). Gib ihn übersichtlich aus.

**Schritt 2 – Schätze sammeln:**
Lege 3 verschiedene Schatz-Steckbriefe an und speichere sie in einer Liste.

**Schritt 3 – Archiv durchsuchen:**
Schreibe Code, der alle Schätze mit einem Wert über 500 Gold findet und ausgibt. Zeige am Ende den Gesamtwert aller Schätze.

**Bonus:** Sortiere die Schätze nach Wert (höchster zuerst).

### ⭐⭐⭐⭐⭐ Boss-Quest 2: Der Dungeon-Manager

**Belohnung:** 600 XP + Dungeon-Meister-Siegel

Der Dungeon lebt – jede Ebene hat eigene Monster und Schätze!

**Schritt 1 – Dungeon aufbauen:**
Erstelle ein Dictionary mit 3 Ebenen. Jede Ebene ist selbst ein Dictionary mit Monster, Schatz und Schwierigkeit.

**Schritt 2 – Ebenen erkunden:**
Gib jede Ebene mit allen Details aus. Zeige an welche Ebene am gefährlichsten ist.

**Schritt 3 – Held schicken:**
Erstelle einen Held-Steckbrief (Dictionary) und prüfe für jede Ebene: Ist der Held stark genug (Level ≥ Schwierigkeit)? Gib aus, welche Ebenen er betreten kann.

**Bonus:** Lass den Held nach dem Sieg über jede Ebene aufsteigen (Level + 1).

### ⭐⭐⭐⭐☆ Boss-Quest 3: Das Helden-Archiv

**Belohnung:** 400 XP + Archiv-Schlüssel

Die Gilde führt ein Register aller aktiven Abenteurer!

**Schritt 1 – Helden anlegen:**
Erstelle 4 Helden als Dictionaries (Name, Klasse, Level, Spezialfähigkeit) und speichere sie in einer Liste.

**Schritt 2 – Suche:**
Schreibe Code, der alle Helden einer bestimmten Klasse findet. Teste die Suche mit mindestens 2 Klassen.

**Schritt 3 – Rangliste:**
Finde den Helden mit dem höchsten Level und gib die Rangliste (sortiert nach Level) aus.

**Bonus:** Füge jedem Helden ein Tupel mit seinen letzten 3 Missionen hinzu.
