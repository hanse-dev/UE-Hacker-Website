## Lesson 1: String Gallop

**What it is:** The f-String is the modern way to connect text and variables into beautiful messages.

**Important:** Always put an `f` before the quotation marks!

```python
# Example 1: Simple f-Strings
horse_name = "Thunder"
age = 8
print(f"Hello {horse_name}!")
print(f"You are {age} years old!")

# Example 2: With calculations
feed_morning = 3
feed_evening = 2
print(f"Daily feed: {feed_morning + feed_evening} kg")

# Example 3: Multiple variables
rider = "Sarah"
horse = "Luna"
discipline = "Dressage"
print(f"{rider} and {horse} are training {discipline}!")
```

## Lesson 2: The Four Hoofbeat Types

**What it is:** Every piece of information at the riding ranch has a type – just like a horse shows one of four gaits. You recognise the gait by its rhythm, and you recognise the type with `type()`:

| Hoofbeat Type | Gait | Term | Example | Why the gait fits |
|----------------|------|------|---------|--------------------|
| 🚶 Walk | calm, steady | `str` (text) | `"Thunder"` | Walk is the gait for talking/communicating |
| 🐎 Trot | firmly timed, countable | `int` (whole number) | `7` | Trot has a clear, countable two-beat rhythm |
| 🏇 Canter | flowing, never quite the same | `float` (decimal number) | `4.25` | Canter is a flowing three-beat, never quite exact |
| 🦘 Jump | either cleared or not | `bool` (truth value) | `True` / `False` | A jump is either cleared or knocked down – only two states |

**Important:** `type()` shows you the hoofbeat type of any information!

```python
# Example 1: 🚶 Walk - Strings (text information)
horse_name = "Windstorm"
breed = "Hanoverian"
print(f"Horse: {horse_name}")
print(f"Breed: {breed}")
print(f"Type: {type(horse_name)}")

# Example 2: 🐎 Trot - Integers (whole numbers)
age = 7
height_cm = 168
print(f"Age: {age} years, Height: {height_cm} cm")
print(f"Type: {type(age)}")

# Example 3: 🏇 Canter - Floats (decimal numbers)
weight = 550.5
feed_amount = 4.25
print(f"Weight: {weight} kg, Feed: {feed_amount} kg")
print(f"Type: {type(weight)}")

# Example 4: 🦘 Jump - Booleans (truth values)
has_saddle = True
is_visited = False
print(f"Has saddle: {has_saddle}")
print(f"Is visited: {is_visited}")
print(f"Type: {type(has_saddle)}")
```

## Lesson 3: Operations and Conversions

**What it is:** How you combine, modify, and convert information.

**Important:** Not all types can be directly combined!

```python
# Example 1: Mathematical operations
horses_a = 12
horses_b = 8
print(f"Total horses: {horses_a + horses_b}")
print(f"Difference: {horses_a - horses_b}")
print(f"Double stables: {horses_a * 2}")
print(f"Horses per box: {horses_a / 4}")

# Example 2: String operations
name = "Star"
suffix = "light"
full_name = name + suffix
print(f"Full name: {full_name}")
print(f"Whinny: {name * 3}")

# Example 3: Type conversion
age_text = "5"
age_number = int(age_text)
print(f"Age as text: {age_text} (type: {type(age_text)})")
print(f"Age as number: {age_number} (type: {type(age_number)})")

# Example 4: Process input
# Input simulated (later with input())
entry = "15"
number = int(entry)
print(f"{number} kg of hay lasts for {number * 2} days.")
```
