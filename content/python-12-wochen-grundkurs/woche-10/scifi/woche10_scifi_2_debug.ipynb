### 🐛 Bug #1

Was ist falsch? Finde und behebe den Fehler!

```python
class raumschiff:
    def __init__(self, name):
        self.name = name

schiff = raumschiff()
print(schiff.name)
```

### 🐛 Bug #2

Was ist falsch? Finde und behebe den Fehler!

```python
class Cyborg:
    def __init(name, power_level):
        name = name
        power_level = power_level

terminator = Cyborg("T-800", 9000)
print(terminator.name)
```

### 🐛 Bug #3

Was ist falsch? Finde und behebe den Fehler!

```python
class Roboter:
    def __init__(self, bezeichnung):
        self.bezeichnung = bezeichnung
    
    def bewegen():
        print(f"{self.bezeichnung} bewegt sich! 🤖")

roboter = Roboter("R2-D2")
roboter.bewegen()
```
