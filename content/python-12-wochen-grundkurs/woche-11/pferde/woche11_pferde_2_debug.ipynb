### 🐛 Bug #1

**Was passiert:** Python kann diese Zeile nicht lesen – schau genau hin. Oft fehlt ein `:`, eine Klammer oder ein Anführungszeichen.

```python
# 🐛 Bug #1: Falsche Vererbung

class pferd:
    wiehern():
        print("Wieher! 🐴")

class araber(pferd):
    rennen():
        print("Schnell wie der Wind!")

mein_pferd = araber()
mein_pferd.wiehern()
```

### 🐛 Bug #2

**Was passiert:** Methoden in Klassen brauchen immer `self` als ersten Parameter. Ohne `self` weiß die Methode nicht, zu welchem Objekt sie gehört.

```python
# 🐛 Bug #2: Falsches super()

class tier:
    def __init__(name):
        self.name = name

class pferd(tier):
    def __init__(name, rasse):
        super().__init__(name)
        self.rasse = rasse

stormy = pferd("Stormy", "Islandpferd")
print(stormy.name)
```

### 🐛 Bug #3

**Was passiert:** Methoden in Klassen brauchen immer `self` als ersten Parameter. Ohne `self` weiß die Methode nicht, zu welchem Objekt sie gehört.

```python
# 🐛 Bug #3: Magic Method Fehler

class preis:
    def __init__(wert):
        self.wert = wert
    
    def __add__(other):
        return self.wert + other.wert

preis1 = preis(1000)
preis2 = preis(2000)
gesamt = preis1 + preis2
print(f"Gesamtpreis: {gesamt} Euro")
```
