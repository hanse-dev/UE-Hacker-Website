### ⭐⭐⭐☆☆ Mission 2: The Calculation Spell

```python
import math
import random

# Step 1: Circle spell
radius = 7
circumference = 2 * math.pi * radius
area = math.pi * radius ** 2
print("=== Magical Circle ===")
print(f"Radius:        {radius} m")
print(f"Circumference: {circumference:.2f} m")
print(f"Area:          {area:.2f} m²")

# Step 2: Random numbers and sum
mana_values = [random.randint(1, 100) for _ in range(5)]
mana_sum = sum(mana_values)
print(f"\nMana values: {mana_values}")
print(f"Sum:         {mana_sum}")

# Step 3: Angle conversion
print("\nAngle conversion:")
for degrees in [90, 180, 270]:
    radians = math.radians(degrees)
    print(f"  {degrees}° = {radians:.4f} rad")

# Step 4: Square roots
print("\nSquare roots:")
for number in [4, 9, 16, 25]:
    print(f"  √{number} = {math.sqrt(number):.1f}")

# Bonus: Trigonometry
angle = math.radians(45)
print(f"\nBonus: sin(45°) = {math.sin(angle):.4f}, cos(45°) = {math.cos(angle):.4f}")
```

### ⭐⭐⭐⭐☆ Mission 3: The Random Spell

```python
import random
import math

# Step 1: Realm selection
realms = ["Pyralia", "Frostmark", "Sunstone", "Shadowdale", "Crystal Coast", "Iron Fortress"]
chosen_realm = random.choice(realms)
print("=== Adventure World ===")
print(f"All realms: {realms}")
print(f"Chosen realm: {chosen_realm}")

# Step 2: Weather simulation
print("\nWeather report:")
regions = ["North region", "Central region", "South region"]
for region in regions:
    temperature = round(random.uniform(-50, 50), 1)
    print(f"  {region}: {temperature}°C")

# Step 3: Magic consumption
print("\nMana consumption of wizards:")
wizard_names = ["Merlin", "Gandara", "Pyrokles"]
total_mana = 0
for name in wizard_names:
    mana = random.randint(50, 150)
    total_mana += mana
    print(f"  {name}: {mana} Mana")
print(f"  Total: {total_mana} Mana")

# Step 4: Hero generator
print("\nHero rankings:")
heroes = [(name, random.randint(1, 10)) for name in ["Thorin", "Aria", "Drake"]]
for name, strength in heroes:
    print(f"  {name}: Strength {strength}/10")

# Bonus: Normal distribution
mana_gauss = round(random.gauss(100, 20), 1)
print(f"\nBonus – Magical mana value (Gauss): {mana_gauss}")
```

## 🐉 Boss Quest Solution Suggestions

### ⭐⭐⭐⭐☆ Boss Quest 1: The Dice Tower

```python
import random

players = ["Merlin", "Aria", "Dragan"]
wins = {name: 0 for name in players}

print("=== DICE TOURNAMENT ===")

for round_num in range(1, 6):
    print(f"\nRound {round_num}:")
    rolls = {}
    for name in players:
        roll = random.randint(1, 6)
        rolls[name] = roll
        print(f"  {name} rolls: {roll}")

    best = max(rolls, key=rolls.get)
    wins[best] += 1
    print(f"  -> Winner: {best}!")

print("\n=== TOURNAMENT RESULT ===")
for name, count in wins.items():
    print(f"  {name}: {count} rounds won")
champion = max(wins, key=wins.get)
print(f"\nOverall champion: {champion}!")

# Bonus: Magic dice
print("\n=== BONUS: Magic Dice Round (1-20) ===")
for name in players:
    magic_roll = random.randint(1, 20)
    print(f"  {name}: {magic_roll}")
```

### ⭐⭐⭐⭐☆ Boss Quest 3: The Random Dungeon

```python
import random
import math

room_types = ["Treasure Chamber", "Trap", "Empty Room", "Guard Post", "Altar"]
monster_list = ["Goblin", "Skeleton", "Troll", None, None]  # None = no monster

print("=== RANDOM DUNGEON ===")
monster_count = 0

for i in range(1, 6):
    room_type = random.choice(room_types)
    monster = random.choice(monster_list)
    if monster:
        monster_count += 1
        contents = f"Monster: {monster}"
    else:
        contents = "Safe – no monster"
    print(f"Room {i}: {room_type} | {contents}")

lives_needed = math.ceil(monster_count / 2)
print(f"\nMonsters total: {monster_count}")
print(f"Lives needed: {lives_needed}")

print()
print("🎉 +400 XP: Boss Quest completed!")
print("🏆 You defeated the Librarian of Infinite Spells!")
print("⭐ Title earned: Master of Libraries")
print()
print("🎊 CONGRATULATIONS! You have mastered Week 7!")
print("📚 Next week: Dictionaries and Tuples!")
```
