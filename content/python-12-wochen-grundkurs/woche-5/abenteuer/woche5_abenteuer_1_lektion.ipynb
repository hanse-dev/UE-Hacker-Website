## Zauberformel 1: Definition - Der magische Vertrag

**Was es ist:** Eine Funktion ist ein wiederverwendbarer Code-Block, den du einmal definierst und immer wieder aufrufen kannst.

**Der Aufbau einer Funktion:**
```python
def funktions_name(parameter1, parameter2):
    """Kurze Beschreibung des Zaubers"""
    # Code, der ausgeführt wird
    return ergebnis  # Optional: Magie-Ergebnis zurückgeben
```

**Schritt-für-Schritt-Erklärung:**
1. **`def`** - Das magische Wort, das die Funktions-Definition startet
2. **`funktions_name`** - Ein eindeutiger Name für deinen Zauber
3. **`(parameter)`** - Zutaten, die dein Zauber braucht (optional)
4. **`:`** - Beendet die Funktions-Definition
5. **`"""Docstring"""`** - Magie-Buch mit Beschreibung (sehr wichtig!)
6. **Eingerückter Code** - Der eigentliche Zauber-Code
7. **`return`** - Gibt das Ergebnis zurück (optional)

**Warum so nützlich:**
- 🎯 Code wiederverwenden statt kopieren
- 🔍 Logik organisieren und strukturieren
- ⚡ Fehler an einer Stelle beheben
- 🎲 Komplexe Probleme in kleine Teile zerlegen

## Zauberformel 1b: Der Schlüssel zur Magie

Eine Funktion zu *definieren* reicht nicht – sie muss auch *aufgerufen* werden! Das Aufrufen ist ein eigenes Thema mit wichtigen Regeln.

### Mit und ohne Klammern

**Mit Klammern `funktion()`:** Der Zauber wird ausgeführt! Die Klammern sagen Python: „Führe diese Funktion jetzt aus“.

**Ohne Klammern `funktion`:** Der Zauber wird NICHT ausgeführt! Du referenzierst nur das Funktionsobjekt – nichts passiert.

### Aufrufen mit Parametern

Beim Aufruf übergibst du **Argumente** in den Klammern: `funktion(argument1, argument2)`. Die Reihenfolge muss stimmen!

### Vom Argument zur Variable

**Beim Aufruf:** Du übergibst ein **Argument** (z.B. `"Aria"`)

**In der Funktion:** Daraus wird eine **Variable** mit dem Parameternamen (z.B. `name`)

Der Parameter ist ein **Platzhalter**. Beim Aufruf ersetzt Python ihn durch den übergebenen Wert – dieser steht dann als Variable in der Funktion zur Verfügung.

````python
# Mit vs. ohne Klammern
def gruesse_abenteurer():
    """Begrüßt einen tapferen Abenteurer"""
    print("Willkommen, tapferer Held!")

# MIT Klammern: Funktion wird ausgeführt
print("=== Mit Klammern gruesse_abenteurer() ===")
gruesse_abenteurer()

# OHNE Klammern: Nur Referenz, nichts passiert
print("\n=== Ohne Klammern gruesse_abenteurer ===")
print("Typ:", type(gruesse_abenteurer))
print("(Die Funktion wurde nicht ausgeführt!)")
````

````python
# Argument wird zur Variable: Schritt für Schritt
def gruesse_nach_name(name):  # "name" ist der Parameter (Platzhalter)
    """Begrüßt einen Helden mit seinem Namen"""
    # In der Funktion ist "name" jetzt eine Variable mit dem übergebenen Wert!
    print(f"In der Funktion: name = '{name}'")
    print(f"Willkommen, {name}!")

# Beim Aufruf: "Aria" ist das Argument
print("=== Aufruf: gruesse_nach_name('Aria') ===")
gruesse_nach_name("Aria")  # → name wird zu "Aria"

print("\n=== Aufruf: gruesse_nach_name('Thorin') ===")
gruesse_nach_name("Thorin")  # → name wird zu "Thorin"
````

````python
# Beispiel 1: Einfache Funktion ohne Parameter
def gruesse_abenteurer():
    """Begrüßt einen tapferen Abenteurer"""
    print("Willkommen, tapferer Held!")
    print("Dein Abenteuer beginnt jetzt!")

# Funktion aufrufen
print("=== Erster Aufruf ===")
gruesse_abenteurer()

print("\n=== Zweiter Aufruf ===")
gruesse_abenteurer()

print("\n=== Was passiert hier? ===")
print("1. def gruesse_abenteurer(): definiert die Funktion")
print("2. gruesse_abenteurer() ruft die Funktion auf")
print("3. Der Code in der Funktion wird jedes Mal ausgeführt")
````

````python
# Beispiel 2: Funktion mit einem Parameter
def gruesse_nach_name(name):
    """Begrüßt einen Helden mit seinem Namen"""
    print(f"Willkommen, {name}!")
    print(f"Viel Erfolg auf deiner Reise, {name}!")

# Funktion mit verschiedenen Namen aufrufen
print("=== Verschiedene Helden ===")
gruesse_nach_name("Aria")
gruesse_nach_name("Thorin")
gruesse_nach_name("Luna")

print("\n=== Was passiert hier? ===")
print("1. name ist ein Platzhalter für den übergebenen Wert")
print("2. Jeder Aufruf übergibt einen anderen Namen")
print("3. Die Funktion arbeitet mit dem übergebenen Wert")
````

````python
# Beispiel 3: Funktion mit mehreren Parametern
def erstelle_charakter(name, klasse, level):
    """Erstellt einen Charakter mit Name, Klasse und Level"""
    print(f"Charakter erstellt:")
    print(f"  Name: {name}")
    print(f"  Klasse: {klasse}")
    print(f"  Level: {level}")
    print(f"  Status: Bereit für Abenteuer!")

# Verschiedene Charaktere erstellen
print("=== Charakter-Erstellung ===")
erstelle_charakter("Aria", "Magierin", 5)
print()
erstelle_charakter("Thorin", "Krieger", 8)
print()
erstelle_charakter("Luna", "Schurkin", 3)
````

## 🔍 Das Geheimnis von return - Die magische Rückgabe

**return** ist der mächtigste Teil einer Funktion!

**Was return tut:**
- Beendet die Funktion sofort
- Gibt einen Wert zurück an den Aufrufer
- Dieser Wert kann in einer Variable gespeichert werden

**Ohne return:** Funktion macht etwas (Side Effect)
**Mit return:** Funktion berechnet und gibt Ergebnis zurück

````python
# 🔍 Übungen mit return

# Beispiel 1: Einfache Berechnung mit return
def berechne_schaden(basis_schaden, multiplikator):
    """Berechnet den Gesamtschaden eines Angriffs"""
    gesamtschaden = basis_schaden * multiplikator
    return gesamtschaden

# Ergebnis in Variable speichern
print("=== Schadensberechnung ===")
schaden1 = berechne_schaden(10, 2)
print(f"Schlag 1: {schaden1} Schaden")

schaden2 = berechne_schaden(15, 3)
print(f"Schlag 2: {schaden2} Schaden")

# Ergebnis direkt verwenden
print(f"Kritischer Treffer: {berechne_schaden(20, 5)} Schaden")
````

````python
# Beispiel 2: Funktion mit bedingtem return
def pruefe_level(level):
    """Prüft, ob ein Charakter für eine Quest bereit ist"""
    if level >= 10:
        return "Bereit für die Quest!"
    else:
        return f"Noch {10 - level} Level needed"

print("=== Level-Prüfung ===")
status1 = pruefe_level(5)
print(f"Level 5: {status1}")

status2 = pruefe_level(12)
print(f"Level 12: {status2}")

status3 = pruefe_level(8)
print(f"Level 8: {status3}")
````

````python
# Beispiel 3: Funktion ohne expliziten return (None)
def zeige_inventar():
    """Zeigt das Inventar (gibt nichts zurück)"""
    print("Dein Inventar:")
    print("  - Schwert")
    print("  - Schild")
    print("  - Heiltrank")
    # Kein return - gibt automatisch None zurück

print("=== Inventar anzeigen ===")
ergebnis = zeige_inventar()
print(f"Rückgabewert: {ergebnis}")
print("Funktionen ohne return geben immer None zurück!")
````

## Zauberformel 2: Die Magie-Bücher

**Was es ist:** Docstrings sind Dokumentations-Strings, die erklären, was eine Funktion tut.

**Warum extrem wichtig:**
- 📖 Andere (und du später) verstehen deine Funktion
- 🔍 IDEs zeigen dir Hilfe beim Aufruf
- 🤖 Tools können automatisch Dokumentation erstellen
- ⭐ Zeichen von professionellem Code

**Die drei Arten von Docstrings:**
1. **Einzeilig:** `"""Kurze Beschreibung"""`
2. **Mehrzeilig:** `"""Kurze Beschreibung\n\nDetails..."""`
3. **Vollständig:** `"""Beschreibung\n\nArgs:\nReturns:\n"""`

````python
# Beispiel 1: Verschiedene Docstring-Stile
def einfache_funktion():
    """Dies ist ein einzeiliger Docstring."""
    return "Einfach"

def detaillierte_funktion(name, level):
    """Erstellt eine Charakter-Beschreibung.
    
    Args:
        name (str): Der Name des Charakters
        level (int): Das Level des Charakters
    
    Returns:
        str: Eine formatierte Beschreibung
    """
    return f"{name} ist ein Level {level} Charakter."

# Docstring aufrufen (so funktioniert es!)
print("=== Docstring-Beispiele ===")
print(f"Einfacher Docstring: {einfache_funktion.__doc__}")
print()
print(f"Detaillierter Docstring: {detaillierte_funktion.__doc__}")
````

````python
# Beispiel 2: Hilfe-Funktion nutzen
print("=== Hilfe für Funktionen ===")
help(detaillierte_funktion)
````

## Zauberformel 3: Die Regeln der Magie

**Gute Funktionsnamen sind entscheidend!**

**Die Regeln:**
- **Kleinbuchstaben** mit Unterstrichen: `berechne_schaden`
- **Beschreibend**: Was die Funktion **tut** (Verb)
- **Kurz aber klar**: `get_name()` nicht `retrieve_user_data_from_database()`
- **Keine Verben im Namen**: `name_get()` falsch, `get_name()` richtig

**Beispiele:**
- ✅ `berechne_schaden()`
- ✅ `erstelle_charakter()`
- ✅ `pruefe_level()`
- ❌ `BerechneSchaden()` (CamelCase)
- ❌ `berechne_das_schaden()` (zu lang)
- ❌ `schaden_berechnen()` (Substantiv zuerst)

````python
# ✨ Beispiele für gute und schlechte Namen

# ✅ Gute Namen (klar und beschreibend)
def berechne_heilung(basis_heilung, multiplikator):
    """Berechnet die Heilungsmenge."""
    return basis_heilung * multiplikator

def finde_max_level(level_liste):
    """Findet das höchste Level in einer Liste."""
    return max(level_liste)

def ist_quest_verfuegbar(level, benoetigt_level):
    """Prüft, ob eine Quest verfügbar ist."""
    return level >= benoetigt_level

# ❌ Schlechte Namen (unklar oder falsche Konvention)
# def berechneHeilung():  # CamelCase
# def bs(b, m):  # Zu kurz, nicht beschreibend
# def healing_calculation():  # Substantiv zuerst

print("=== Gute Funktionsnamen in Aktion ===")
heilung = berechne_heilung(10, 2)
print(f"Heilung: {heilung}")

levels = [5, 8, 12, 3, 15]
max_level = finde_max_level(levels)
print(f"Höchstes Level: {max_level}")

quest_ok = ist_quest_verfuegbar(8, 10)
print(f"Quest verfügbar: {quest_ok}")
````

## Zauberformel 4: Fehler abfangen

Manchmal kann ein Programm nicht wissen, ob eine Berechnung klappt. `try/except` lässt das Programm **sicher weiterlaufen**, auch wenn etwas schiefgeht.

```python
try:
    # Code, der einen Fehler auslösen könnte
    ergebnis = int(input("Zahl eingeben: "))
except ValueError:
    # Was passiert, wenn der Fehler auftritt
    print("Das war keine gültige Zahl!")
```

**Häufige Fehlertypen:**
- `ValueError` – falscher Wert (z.B. `int("abc")`)
- `TypeError` – falscher Typ (z.B. Zahl + Text ohne `str()`)
- `ZeroDivisionError` – Division durch null
- `NameError` – Variable existiert nicht

````python
# try/except in einer Funktion
def teile_sicher(zahl, teiler):
    """Teilt zwei Zahlen – gibt None zurück wenn der Teiler 0 ist."""
    try:
        return zahl / teiler
    except ZeroDivisionError:
        print("Fehler: Kann nicht durch null teilen!")
        return None

print(teile_sicher(10, 2))   # 5.0
print(teile_sicher(10, 0))   # Fehlermeldung + None

# try/except mit Typumwandlung
def text_zu_zahl(text):
    """Wandelt Text in Zahl um – gibt None zurück bei ungültigem Wert."""
    try:
        return int(text)
    except ValueError:
        print(f"\"{text}\" ist keine gültige Zahl.")
        return None

print(text_zu_zahl("42"))    # 42
print(text_zu_zahl("abc"))  # Fehlermeldung + None
````
