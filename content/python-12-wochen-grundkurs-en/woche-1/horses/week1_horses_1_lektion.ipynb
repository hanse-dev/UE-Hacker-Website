## Lesson 1: Call

## Lesson 2: Feed Troughs

A variable is like a **labelled feed trough**: you give it a name and can store something inside – and check it again later.

```python
horse_name = "Bobby"   # the trough is called 'horse_name', inside is the text "Bobby"
age = 5                # the trough is called 'age', inside is the number 5
```

**The three most common types of values:**

| Type | Technical term | Example | How to recognise |
|------|---------------|---------|------------------|
| Text | **String** | `"Bobby"` | Quotation marks |
| Whole number | **Integer** | `5` | No quotation marks |
| Yes/No | **Boolean** | `True` / `False` | No quotation marks, capital T or F |

````python
# Example 1: Different variable types
horse_name = "Bobby"       # Text (String)
horse_age = 5              # Number (Integer)
horse_height = 150         # Number (Integer)
horse_friendly = True      # Boolean

# Example 2: Output variables
print("Horse name:")
print(horse_name)
print("Age:")
print(horse_age)
print("Height in cm:")
print(horse_height)

# Example 3: Variables can change
print("Feed before training: " + str(2) + " kg")
feed_amount = 3  # More feed after training
print("Feed after training: " + str(feed_amount) + " kg")
````

## Lesson 3

With `+` you can glue texts together – like connecting building blocks.

```python
print("Horse: " + horse_name)
```

**Why do I need `str()`?**

Python treats text and numbers as different *types*. It's like apples and oranges – you can't simply add them together. `str()` converts a number to text so it can be joined:

```python
age = 5
print("Age: " + age)       # ❌ Error – number and text don't match
print("Age: " + str(age))  # ✅ works – str() turns 5 into the text "5"
```

````python
# Example 1: Simple combination
name = "Stormy"
breed = "Haflinger"
print("Horse: " + name)
print("Breed: " + breed)

# Example 2: With numbers (str() needed!)
age = 8
print("Age: " + str(age) + " years")

# Example 3: Complex sentences
print(name + " is a " + breed + " and " + str(age) + " years old.")

# Example 4: Multiple pieces of information
print()
print("=== HORSE PROFILE ===")
print("Name: " + name)
print("Breed: " + breed)
print("Age: " + str(age) + " years")
print("===================")
````

## Lesson 4: Ask the user

`input()` pauses the program and waits for input. The answer is stored as **text (String)**.

```python
name = input("What is your horse's name? ")
print("Welcome to the ranch, " + name + "!")
```

> ⚠️ `input()` only works in a running Python environment (local or JupyterHub). In static previews it will hang.

````python
# Example 1: Simple input
horse = input("What is your horse's name? ")
print("Welcome to the ranch, " + horse + "!")

# Example 2: Multiple inputs
rider = input("Your name: ")
horse = input("Your horse: ")
print(rider + " rides on " + horse)
````
