## 🐛 Debug-Quest: Finde die Fehler!

Ein Lehrling hat versucht, Schleifen zu meistern, aber etwas läuft schief! Finde die Fehler:

### 🐛 Bug #1

**Was passiert:** Dieser Code enthält einen Fehler – findest du ihn?

```python
# 🐛 Bug #1

zaehler = 0
while zaehler < 5:
    print(f"Zaehler: {zaehler}")
```

### 🐛 Bug #2

**Was passiert:** Dieser Code enthält einen Fehler – findest du ihn?

```python
# 🐛 Bug #2

for i in range(5):
    print(f"Zahl {i+1}")
```

### 🐛 Bug #3

**Was passiert:** Python erwartet einen eingerückten Block. Alles was zu einem `if`, `for`, `while` oder `def` gehört, muss mit 4 Leerzeichen eingerückt werden.

```python
# 🐛 Bug #3

for i in range(3):
print(f"Durchlauf {i}")
print("Ende")
```
