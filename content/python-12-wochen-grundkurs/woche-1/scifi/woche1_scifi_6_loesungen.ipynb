**Erklärung:** Python liest `System` und `online` als Variablennamen – aber diese Variablen gibt es nicht. Text muss immer in Anführungszeichen stehen, damit Python weiß: das ist Text, keine Variable.

```python
# Problem: Text ohne Anführungszeichen
print("System online")
```

**Erklärung:** `level` ist eine Zahl (Integer), der Text davor ist ein String. Python kann diese zwei Typen nicht direkt mit `+` verbinden – das ist wie versuchen, einen Stein an einen Faden zu knoten. `str()` löst das: es verwandelt die Zahl in Text.

```python
# Problem: Zahl kann nicht direkt mit Text verbunden werden
level = 3
print("Sicherheitsstufe: " + str(level))
```

**Erklärung:** Die Variable heißt `status`, aber im `print()` steht `statu` – ein Buchstabe fehlt. Python sucht exakt nach dem Namen `statu` und findet ihn nicht. Tippfehler in Variablennamen sind einer der häufigsten Anfängerfehler!

```python
# Problem: Variable 'statu' existiert nicht
status = "Aktiv"
print("Systemstatus: " + status)
```
