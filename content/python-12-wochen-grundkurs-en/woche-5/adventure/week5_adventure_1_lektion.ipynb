## Spell Formula 1: Definition – The Magical Contract

**What it is:** A function is a reusable block of code that you define once and can call again and again.

**The structure of a function:**
```python
def function_name(parameter1, parameter2):
    """Short description of the spell"""
    # Code that gets executed
    return result  # Optional: return the magic result
```

**Step-by-step explanation:**
1. **`def`** – The magic word that starts the function definition
2. **`function_name`** – A unique name for your spell
3. **`(parameter)`** – Ingredients your spell needs (optional)
4. **`:`** – Ends the function definition
5. **`"""Docstring"""`** – Spell book with description (very important!)
6. **Indented code** – The actual spell code
7. **`return`** – Returns the result (optional)

**Why so useful:**
- 🎯 Reuse code instead of copying it
- 🔍 Organise and structure logic
- ⚡ Fix errors in one place
- 🎲 Break complex problems into small parts

## Spell Formula 1b: The Key to Magic

*Defining* a function is not enough – it also has to be *called*! Calling is its own topic with important rules.

### With and Without Parentheses

**With parentheses `function()`:** The spell is executed! The parentheses tell Python: "Run this function now".

**Without parentheses `function`:** The spell is NOT executed! You are only referencing the function object – nothing happens.

### Calling with Parameters

When calling, you pass **arguments** in the parentheses: `function(argument1, argument2)`. The order must match!

### From Argument to Variable

**When calling:** You pass an **argument** (e.g. `"Aria"`)

**In the function:** It becomes a **variable** with the parameter name (e.g. `name`)

The parameter is a **placeholder**. When called, Python replaces it with the passed value – which is then available as a variable inside the function.

````python
# With vs. without parentheses
def greet_adventurer():
    """Greets a brave adventurer"""
    print("Welcome, brave hero!")

# WITH parentheses: function is executed
print("=== With parentheses greet_adventurer() ===")
greet_adventurer()

# WITHOUT parentheses: just a reference, nothing happens
print("\n=== Without parentheses greet_adventurer ===")
print("Type:", type(greet_adventurer))
print("(The function was not executed!)")
````

````python
# Argument becomes variable: step by step
def greet_by_name(name):  # "name" is the parameter (placeholder)
    """Greets a hero by their name"""
    # Inside the function, "name" is now a variable with the passed value!
    print(f"In the function: name = '{name}'")
    print(f"Welcome, {name}!")

# When calling: "Aria" is the argument
print("=== Call: greet_by_name('Aria') ===")
greet_by_name("Aria")  # → name becomes "Aria"

print("\n=== Call: greet_by_name('Thorin') ===")
greet_by_name("Thorin")  # → name becomes "Thorin"
````

````python
# Example 1: Simple function without parameters
def greet_adventurer():
    """Greets a brave adventurer"""
    print("Welcome, brave hero!")
    print("Your adventure begins now!")

# Call the function
print("=== First call ===")
greet_adventurer()

print("\n=== Second call ===")
greet_adventurer()

print("\n=== What happens here? ===")
print("1. def greet_adventurer(): defines the function")
print("2. greet_adventurer() calls the function")
print("3. The code inside the function runs every time")
````

````python
# Example 2: Function with one parameter
def greet_by_name(name):
    """Greets a hero by their name"""
    print(f"Welcome, {name}!")
    print(f"Good luck on your journey, {name}!")

# Call the function with different names
print("=== Different heroes ===")
greet_by_name("Aria")
greet_by_name("Thorin")
greet_by_name("Luna")

print("\n=== What happens here? ===")
print("1. name is a placeholder for the passed value")
print("2. Each call passes a different name")
print("3. The function works with the passed value")
````

````python
# Example 3: Function with multiple parameters
def create_character(name, class_, level):
    """Creates a character with name, class and level"""
    print(f"Character created:")
    print(f"  Name: {name}")
    print(f"  Class: {class_}")
    print(f"  Level: {level}")
    print(f"  Status: Ready for adventure!")

# Create different characters
print("=== Character Creation ===")
create_character("Aria", "Mage", 5)
print()
create_character("Thorin", "Warrior", 8)
print()
create_character("Luna", "Rogue", 3)
````

## 🔍 The Secret of return – The Magical Return

**return** is the most powerful part of a function!

**What return does:**
- Ends the function immediately
- Returns a value to the caller
- This value can be stored in a variable

**Without return:** Function does something (side effect)
**With return:** Function calculates and returns a result

````python
# 🔍 Exercises with return

# Example 1: Simple calculation with return
def calculate_damage(base_damage, multiplier):
    """Calculates the total damage of an attack"""
    total_damage = base_damage * multiplier
    return total_damage

# Store result in variable
print("=== Damage Calculation ===")
damage1 = calculate_damage(10, 2)
print(f"Strike 1: {damage1} damage")

damage2 = calculate_damage(15, 3)
print(f"Strike 2: {damage2} damage")

# Use result directly
print(f"Critical hit: {calculate_damage(20, 5)} damage")
````

````python
# Example 2: Function with conditional return
def check_level(level):
    """Checks whether a character is ready for a quest"""
    if level >= 10:
        return "Ready for the quest!"
    else:
        return f"Still {10 - level} levels needed"

print("=== Level Check ===")
status1 = check_level(5)
print(f"Level 5: {status1}")

status2 = check_level(12)
print(f"Level 12: {status2}")

status3 = check_level(8)
print(f"Level 8: {status3}")
````

````python
# Example 3: Function without explicit return (None)
def show_inventory():
    """Shows the inventory (returns nothing)"""
    print("Your inventory:")
    print("  - Sword")
    print("  - Shield")
    print("  - Health potion")
    # No return – automatically returns None

print("=== Show Inventory ===")
result = show_inventory()
print(f"Return value: {result}")
print("Functions without return always return None!")
````

## Spell Formula 2: The Spell Books

**What it is:** Docstrings are documentation strings that explain what a function does.

**Why extremely important:**
- 📖 Others (and future you) understand your function
- 🔍 IDEs show you help when calling
- 🤖 Tools can automatically generate documentation
- ⭐ Sign of professional code

**The three types of docstrings:**
1. **One-line:** `"""Short description"""`
2. **Multi-line:** `"""Short description\n\nDetails..."""`
3. **Full:** `"""Description\n\nArgs:\nReturns:\n"""`

````python
# Example 1: Different docstring styles
def simple_function():
    """This is a one-line docstring."""
    return "Simple"

def detailed_function(name, level):
    """Creates a character description.
    
    Args:
        name (str): The name of the character
        level (int): The level of the character
    
    Returns:
        str: A formatted description
    """
    return f"{name} is a Level {level} character."

# Calling a docstring (this is how it works!)
print("=== Docstring Examples ===")
print(f"Simple docstring: {simple_function.__doc__}")
print()
print(f"Detailed docstring: {detailed_function.__doc__}")
````

````python
# Example 2: Using the help function
print("=== Help for Functions ===")
help(detailed_function)
````

## Spell Formula 3: The Rules of Magic

**Good function names are crucial!**

**The rules:**
- **Lowercase letters** with underscores: `calculate_damage`
- **Descriptive**: What the function **does** (verb)
- **Short but clear**: `get_name()` not `retrieve_user_data_from_database()`
- **No nouns first**: `name_get()` wrong, `get_name()` right

**Examples:**
- ✅ `calculate_damage()`
- ✅ `create_character()`
- ✅ `check_level()`
- ❌ `CalculateDamage()` (CamelCase)
- ❌ `calculate_the_damage()` (too long)
- ❌ `damage_calculate()` (noun first)

````python
# ✨ Examples of good and bad names

# ✅ Good names (clear and descriptive)
def calculate_healing(base_healing, multiplier):
    """Calculates the healing amount."""
    return base_healing * multiplier

def find_max_level(level_list):
    """Finds the highest level in a list."""
    return max(level_list)

def is_quest_available(level, required_level):
    """Checks whether a quest is available."""
    return level >= required_level

# ❌ Bad names (unclear or wrong convention)
# def calculateHealing():  # CamelCase
# def bh(b, m):  # Too short, not descriptive
# def healing_calculation():  # Noun first

print("=== Good Function Names in Action ===")
healing = calculate_healing(10, 2)
print(f"Healing: {healing}")

levels = [5, 8, 12, 3, 15]
max_level = find_max_level(levels)
print(f"Highest level: {max_level}")

quest_ok = is_quest_available(8, 10)
print(f"Quest available: {quest_ok}")
````

## Spell Formula 4: Catching Errors

Sometimes a program can't know whether a calculation will succeed. `try/except` lets the program **keep running safely**, even when something goes wrong.

```python
try:
    # Code that could raise an error
    result = int(input("Enter a number: "))
except ValueError:
    # What happens when the error occurs
    print("That was not a valid number!")
```

**Common error types:**
- `ValueError` – wrong value (e.g. `int("abc")`)
- `TypeError` – wrong type (e.g. number + text without `str()`)
- `ZeroDivisionError` – division by zero
- `NameError` – variable does not exist

````python
# try/except in a function
def divide_safe(number, divisor):
    """Divides two numbers – returns None if divisor is 0."""
    try:
        return number / divisor
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None

print(divide_safe(10, 2))   # 5.0
print(divide_safe(10, 0))   # Error message + None

# try/except with type conversion
def text_to_number(text):
    """Converts text to a number – returns None for invalid values."""
    try:
        return int(text)
    except ValueError:
        print(f'"{text}" is not a valid number.')
        return None

print(text_to_number("42"))    # 42
print(text_to_number("abc"))   # Error message + None
````
