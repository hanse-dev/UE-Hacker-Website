### 🐛 Bug #2

**Was passiert:** Methoden in Klassen brauchen immer `self` als ersten Parameter. Ohne `self` weiß die Methode nicht, zu welchem Objekt sie gehört.

```python
# 🐛 Bug #2: Falscher Konstruktor

class Reiter:
    def __init(name, level):
        name = name
        level = level

maria = Reiter("Maria", 3)
print(maria.name)
```

### 🐛 Bug #3

**Was passiert:** Methoden in Klassen brauchen immer `self` als ersten Parameter. Ohne `self` weiß die Methode nicht, zu welchem Objekt sie gehört.

```python
# 🐛 Bug #3: Methode ohne self

class Pferd:
    def __init__(self, name):
        self.name = name
    
    def galoppieren():
        print(f"{self.name} galoppiert! 🐎")

pferd = Pferd("Spirit")
pferd.galoppieren()
```
