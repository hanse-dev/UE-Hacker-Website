| Term | Meaning | Example |
|------|---------|--------|
| `import` | Load a module (toolbox) | `import math` |
| `from … import …` | Load only specific parts of a module | `from math import sqrt` |
| `import … as …` | Import a module with a short alias | `import random as r` |
| `math.sqrt()` | Calculate square root | `math.sqrt(16)` → `4.0` |
| `math.floor()` | Round down to whole number | `math.floor(3.9)` → `3` |
| `math.ceil()` | Round up to whole number | `math.ceil(3.1)` → `4` |
| `math.pi` | The number Pi | `math.pi` → `3.14159…` |
| `random.randint()` | Random integer within a range | `random.randint(1, 6)` |
| `random.choice()` | Random element from a list | `random.choice(list)` |
| `random.shuffle()` | Randomly shuffle a list (modifies the list) | `random.shuffle(list)` |

```python
import math
import random

# math module
print(math.sqrt(25))   # 5.0
print(math.floor(3.7)) # 3
print(math.ceil(3.2))  # 4
print(round(math.pi, 2))  # 3.14

# random module
roll = random.randint(1, 6)
print("Dice:", roll)

planets = ["Mars", "Venus", "Jupiter"]
print(random.choice(planets))

random.shuffle(planets)
print(planets)
```
