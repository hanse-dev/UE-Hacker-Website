| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `turtle` | Python-Modul für grafische Zeichnungen | `import turtle` |
| `turtle.Turtle()` | Ein neues Turtle-Objekt (den Zeichenstift) erstellen | `t = turtle.Turtle()` |
| `.forward()` | Turtle vorwärts bewegen (zeichnet dabei eine Linie) | `t.forward(100)` |
| `.backward()` | Turtle rückwärts bewegen | `t.backward(50)` |
| `.left()` | Turtle nach links drehen (in Grad) | `t.left(90)` |
| `.right()` | Turtle nach rechts drehen (in Grad) | `t.right(45)` |
| `.color()` | Stiftfarbe setzen | `t.color("red")` |
| `.fillcolor()` | Füllfarbe für Flächen setzen | `t.fillcolor("blue")` |
| `.begin_fill()` / `.end_fill()` | Fläche zwischen diesen Aufrufen füllen | `t.begin_fill()` … `t.end_fill()` |
| `.circle()` | Kreis mit gegebenem Radius zeichnen | `t.circle(50)` |
| `.speed()` | Zeichengeschwindigkeit setzen (0 = schnellst) | `t.speed(5)` |
| `.penup()` / `.pendown()` | Stift heben (kein Zeichnen) / senken (Zeichnen) | `t.penup()` |

```python
import turtle

t = turtle.Turtle()
t.speed(5)

# Quadrat zeichnen
for _ in range(4):
    t.forward(100)
    t.right(90)

# Farbiger Kreis
t.penup()
t.goto(0, -50)
t.pendown()
t.color("blue")
t.fillcolor("lightblue")
t.begin_fill()
t.circle(50)
t.end_fill()

turtle.done()
```
