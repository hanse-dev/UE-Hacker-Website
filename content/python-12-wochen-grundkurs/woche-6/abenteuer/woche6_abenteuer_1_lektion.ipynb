## Sammlungs-Zauber 1: Die Listen-Erstellung - Der erste Schatz

**Was es ist:** Eine Liste ist eine geordnete Sammlung von Elementen, die du mit eckigen Klammern erstellst.

**Der Aufbau einer Liste:**
```python
# Leere Liste erstellen
leere_liste = []

# Liste mit Elementen erstellen
schatzliste = ["Gold", "Edelsteine", "Zaubertränke"]
```

**Schritt-für-Schritt-Erklärung:**
1. **`[]`** - Eckige Klammern definieren eine Liste
2. **`,`** - Kommas trennen die Elemente
3. **`""`** - Strings werden in Anführungszeichen gesetzt
4. **`[]`** - Mit Index greifst du auf Elemente zu (beginnt bei 0)

**Warum so nützlich:**
- 🎯 Mehrere Werte in einer Variable speichern
- 🔍 Geordnete Struktur für Daten
- ⚡ Flexibel erweiterbar und veränderbar
- 🎲 Verschiedene Datentypen mischbar

````python
# Beispiel 1: Einfache Listen erstellen
# Leere Liste
leere_liste = []
print(f"Leere Liste: {leere_liste}")

# Liste mit Schätzen
schaetze = ["Goldmünzen", "Silberbarren", "Edelsteine"]
print(f"Schatzliste: {schaetze}")

# Liste mit Zahlen
level = [1, 5, 10, 15, 20]
print(f"Level-Liste: {level}")

# Gemischte Liste
mixed = ["Schwert", 25, True, "Schild"]
print(f"Gemischte Liste: {mixed}")
````

````python
# Beispiel 2: Auf Listenelemente zugreifen
inventar = ["Schwert", "Schild", "Trank", "Bogen", "Pfeile"]

print("=== Inventar-Zugriff ===")
print(f"Gesamtes Inventar: {inventar}")
print(f"Erstes Item: {inventar[0]}")
print(f"Zweites Item: {inventar[1]}")
print(f"Letztes Item: {inventar[-1]}")

# Länge der Liste
print(f"Anzahl Items: {len(inventar)}")

# Slicing (Ausschnitte)
print(f"Erste 3 Items: {inventar[0:3]}")
print(f"Letzte 2 Items: {inventar[-2:]}")
````

````python
# Beispiel 3: Listen in Listen (verschachtelte Listen)
helden_party = [
    ["Aria", "Magierin", 15],
    ["Thorin", "Krieger", 12],
    ["Luna", "Schurkin", 8]
]

print("=== Helden-Party ===")
print(f"Erster Held: {helden_party[0]}")
print(f"Name des ersten Helden: {helden_party[0][0]}")
print(f"Klasse des zweiten Helden: {helden_party[1][1]}")
print(f"Level des dritten Helden: {helden_party[2][2]}")
````

## Sammlungs-Zauber 2: Elemente verwalten - Der Schatzmeister

**Die wichtigsten Listen-Methoden:**

**Elemente hinzufügen:**
- `append(element)` - Fügt am Ende hinzu
- `insert(index, element)` - Fügt an bestimmter Position ein

**Elemente entfernen:**
- `remove(element)` - Entfernt erstes Vorkommen
- `pop(index)` - Entfernt und gibt Element zurück
- `clear()` - Leert die gesamte Liste

````python
# 🔍 Übungen mit append() und insert()

# Beispiel 1: Elemente hinzufügen
beutesack = []
print(f"Leerer Beutesack: {beutesack}")

# Mit append() am Ende hinzufügen
beutesack.append("Goldring")
print(f"Nach append: {beutesack}")

beutesack.append("Zauberstab")
beutesack.append("Heiltrank")
print(f"Voll gefüllt: {beutesack}")

# Mit insert() an bestimmter Position einfügen
beutesack.insert(0, "Schlüssel")  # Am Anfang
print(f"Nach insert(0): {beutesack}")

beutesack.insert(2, "Amulett")  # An Position 2
print(f"Nach insert(2): {beutesack}")
````

````python
# Beispiel 2: Elemente entfernen
inventar = ["Schwert", "Schild", "Trank", "Bogen", "Pfeile", "Trank"]
print(f"Original-Inventar: {inventar}")

# Mit remove() erstes Vorkommen entfernen
inventar.remove("Trank")
print(f"Nach remove('Trank'): {inventar}")

# Mit pop() letztes Element entfernen und zurückgeben
letztes_item = inventar.pop()
print(f"Nach pop(): {inventar}")
print(f"Entferntes Item: {letztes_item}")

# Mit pop(index) bestimmtes Element entfernen
zweites_item = inventar.pop(1)
print(f"Nach pop(1): {inventar}")
print(f"Entferntes Item: {zweites_item}")

# Mit clear() alles löschen
inventar.clear()
print(f"Nach clear(): {inventar}")
````

````python
# Beispiel 3: Listen kombinieren
schaetze = ["Gold", "Silber", "Edelsteine"]
waffen = ["Schwert", "Bogen", "Axt"]
tränke = ["Heilung", "Mana", "Stärke"]

print("=== Listen kombinieren ===")
print(f"Schätze: {schaetze}")
print(f"Waffen: {waffen}")
print(f"Tränke: {tränke}")

# Mit + Operator kombinieren
alles = schaetze + waffen + tränke
print(f"Alles kombiniert: {alles}")

# Mit extend() eine Liste zur anderen hinzufügen
schaetze.extend(waffen)
print(f"Schätze erweitert: {schaetze}")
````

## Sammlungs-Zauber 3: Listen durchsuchen und sortieren

**Suchen und prüfen:**
- `index(element)` – gibt die Position des Elements zurück
- `element in liste` – prüft ob Element vorhanden (`True`/`False`)
- `count(element)` – zählt wie oft ein Element vorkommt
- `len(liste)` – gibt die Anzahl der Elemente zurück

**Sortieren:**
- `sort()` – sortiert die Liste (verändert sie direkt)
- `sorted(liste)` – gibt eine sortierte Kopie zurück
- `reverse()` – dreht die Reihenfolge um

**Mit Index und Wert gleichzeitig:**
- `enumerate(liste)` – liefert in einer `for`-Schleife immer **Index und Wert** als Paar:

```python
for index, element in enumerate(liste):
    print(f"Position {index}: {element}")
```

````python
# Beispiel 1: In Listen suchen
inventar = ["Schwert", "Schild", "Trank", "Bogen", "Schwert", "Pfeile"]

print("=== In Listen suchen ===")
print(f"Inventar: {inventar}")

# Mit index() Position finden
position_schild = inventar.index("Schild")
print(f"Schild an Position: {position_schild}")

# Mit in() prüfen ob vorhanden
hat_bogen = "Bogen" in inventar
hat_zauberstab = "Zauberstab" in inventar
print(f"Bogen vorhanden: {hat_bogen}")
print(f"Zauberstab vorhanden: {hat_zauberstab}")

# Mit count() zählen
anzahl_schwerter = inventar.count("Schwert")
print(f"Anzahl Schwerter: {anzahl_schwerter}")
````

````python
# Beispiel 2: Listen sortieren
zahlen = [5, 2, 8, 1, 9, 3]
namen = ["Zara", "Anna", "Max", "Berta", "Klaus"]

print("=== Listen sortieren ===")
print(f"Original-Zahlen: {zahlen}")
print(f"Original-Namen: {namen}")

# Mit sort() direkt sortieren
zahlen.sort()
print(f"Zahlen nach sort(): {zahlen}")

# Mit sorted() Kopie sortieren
sortierte_namen = sorted(namen)
print(f"Namen Original: {namen}")
print(f"Namen sortiert: {sortierte_namen}")

# Umgekehrt sortieren
zahlen.sort(reverse=True)
print(f"Zahlen absteigend: {zahlen}")

# Mit reverse() Reihenfolge umkehren
namen.reverse()
print(f"Namen umgekehrt: {namen}")
````

````python
# Beispiel 3: Nützliche Listen-Operationen
monster = ["Goblin", "Ork", "Troll", "Drache", "Goblin", "Ork"]

print("=== Listen-Operationen ===")
print(f"Monster-Liste: {monster}")
print(f"Anzahl Monster: {len(monster)}")

# Einzigartige Elemente: set() entfernt Duplikate
# set ist eine Sammlung ohne Reihenfolge und ohne Duplikate
einzigartige_monster = list(set(monster))
print(f"Einzigartige Monster: {einzigartige_monster}")

# Mit enumerate() Index und Wert gleichzeitig
print("\nAlle Monster mit Position:")
for index, m in enumerate(monster):
    print(f"  Position {index}: {m}")
````

## Sammlungs-Zauber 4: break und continue – Schleifen steuern

Wenn du eine Liste durchläufst, kannst du die Schleife gezielt steuern:

- **`break`** – Schleife sofort beenden (z.B. wenn du gefunden hast was du suchst)
- **`continue`** – aktuellen Durchlauf überspringen und mit dem nächsten weitermachen

```python
for item in liste:
    if item == "Gesuchtes":
        break       # gefunden – aufhören
    if item == "Überspringen":
        continue    # dieses Element ignorieren
    print(item)
```

````python
inventar = ["Trank", "Schwert", "Schlüssel", "Schatz", "Karte"]

# break: Suche beenden sobald Schatz gefunden
print("=== Schatzsuche ===")
for gegenstand in inventar:
    print(f"Prüfe: {gegenstand}")
    if gegenstand == "Schatz":
        print("🎉 Schatz gefunden! Suche beendet.")
        break

# continue: Tränke beim Ausgeben überspringen
print("\n=== Inventar ohne Tränke ===")
for gegenstand in inventar:
    if gegenstand == "Trank":
        continue
    print(f"  - {gegenstand}")
````
