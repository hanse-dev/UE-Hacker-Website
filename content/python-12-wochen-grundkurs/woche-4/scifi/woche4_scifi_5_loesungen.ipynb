## 🐛 Debug-Quest Lösungen

```python
# ✅ Lösung Bug #1: Endlosschleife

# Problem: zeit wird nicht erhöht
zeit = 0
while zeit < 5:
    print(f"Zeit {zeit}")
    zeit += 1  # Das fehlte!
```

```python
# ✅ Lösung Bug #2: Falsche range-Grenze

# Problem: range(5) gibt 0-4, aber wir wollen 1-5
for i in range(1, 6):  # Start bei 1, ende bei 6 (exklusiv)
    print(f"System {i}")
```

```python
# ✅ Lösung Bug #3: Vergessene Einrückung

# Problem: Code nach for nicht eingerückt
for i in range(3):
    print(f"Protokoll {i}")
print("Ende")
```
