## System Log 1: The Loop – The Time Loop

**What it is:** A for loop repeats code for every element in a sequence.

**Structure of a for loop:**
```python
for variable in sequence:
    # Code that runs for every element
```

**Step-by-step explanation:**
1. **`for`** – The command that starts the time loop
2. **`variable`** – A placeholder that holds the current element on each pass
3. **`in`** – Connects the variable to the sequence
4. **`sequence`** – The collection we iterate through (e.g. list, string, range())
5. **`:`** – Ends the loop definition
6. **Indented code** – Runs on every pass

**The iterator:**
- The iterator is like an invisible time tracker that remembers where we are in the sequence
- It starts at the first element
- After each pass it automatically jumps to the next element
- The loop ends when no elements are left

**Why so useful:**
- 🎯 Automate repeating tasks
- 🔍 Process each element individually
- ⚡ Make code more efficient and shorter
- 🎲 Create patterns and sequences

````python
# Example 1: Simple for loop with range()
print("=== Example 1: 5 Time Cycles ===")
# range(5) generates the numbers 0, 1, 2, 3, 4
for i in range(5):
    print(f"Time Cycle {i+1}: Systems synchronized")

print("\n=== What happens here? ===")
print("1. range(5) creates: [0, 1, 2, 3, 4]")
print("2. i is set to the next element on each pass")
print("3. We print i+1 to count from 1 to 5")
print("4. The loop ends after the last element")
````

````python
# Example 2: For loop with start and end
print("=== Example 2: Scanning Sectors 3 to 7 ===")
# range(3, 8) generates the numbers 3, 4, 5, 6, 7 (8 is exclusive!)
for sector in range(3, 8):
    print(f"Sector {sector}: Scan complete")

print("\n=== Important observation ===")
print("range(start, end) includes start, but not end!")
print("Like coordinates: from 3 up to BEFORE 8")
````

````python
# Example 3: For loop with step size
print("=== Example 3: Every Second Sector ===")
# range(0, 11, 2) generates: 0, 2, 4, 6, 8, 10
for sector in range(0, 11, 2):
    print(f"Sector {sector}: Reinforcement activated")

print("\n=== The step size ===")
print("range(start, end, step) skips sectors")
print("Here: skip every other sector")
````

````python
# Example 4: For loop over a string
print("=== Example 4: Letters of a Code ===")
code = "ALPHA"
print(f"The code is: {code}")
print("Letter by letter:")

for letter in code:
    print(f"  - {letter}")

print("\n=== What happens here? ===")
print("String is treated as a sequence of letters")
print("Each letter is iterated individually")
````

## 🔍 The Secret of range() – The Time Generator

**range() is your most powerful tool for time sequences!**

### The three forms of range():

1. **`range(n)`** – Numbers from 0 to n-1
   ```python
   range(5) → 0, 1, 2, 3, 4
   ```

2. **`range(start, end)`** – Numbers from start to end-1
   ```python
   range(2, 7) → 2, 3, 4, 5, 6
   ```

3. **`range(start, end, step)`** – Numbers with intervals
   ```python
   range(1, 10, 2) → 1, 3, 5, 7, 9
   ```

**Important rules:**
- The **upper bound is always exclusive** (not counted)
- **Negative steps** are possible for counting backwards
- range() is **very memory efficient** (generates numbers on demand)

````python
# 🔍 Exercises with range()

# Exercise 1: Countdown to jump
print("=== Countdown to Hyperspace Jump ===")
for i in range(5, 0, -1):
    print(f"{i}...")
print("🚀 HYPERSPACE JUMP!")

# Exercise 2: Even timestamps
print("\n=== Even Timestamps ===")
for time in range(0, 21, 2):
    print(f"T{time:02d}", end=" ")
print()

# Exercise 3: Energy cycles
print("\n=== Energy Cycles ===")
for cycle in range(1, 6):
    print(f"Cycle {cycle}: Energy waves active")
````

## System Log 2: The Loop – The Continuous Scanner

**What it is:** A while loop repeats code as long as a condition is true.

**Structure of a while loop:**
```python
while condition:
    # Code that repeats as long as condition is true
    # IMPORTANT: The condition must eventually become false!
```

**⚠️ Caution:** Infinite loops occur when the condition never becomes false!

````python
# Example 1: Simple while loop
print("=== Example 1: Scanning up to 10 ===")
scan = 0
while scan <= 10:
    print(f"Scan {scan}: Data received")
    scan += 1  # IMPORTANT: Increment, otherwise infinite loop!

print("\n=== What happens here? ===")
print("1. scan starts at 0")
print("2. As long as scan <= 10, the code runs")
print("3. scan += 1 increases the value on each pass")
print("4. At scan = 11 the condition is false → loop ends")
````

````python
# Example 2: Shield generator
print("=== Example 2: Shield Generator ===")
energy = 100
minute = 1

while energy > 20:
    consumption = 15
    energy -= consumption
    print(f"Minute {minute}: -{consumption} energy, remaining: {energy}%")
    minute += 1

print(f"\n⚠️ Shield critical after {minute-1} minutes!")
print("Recharge energy!")
````

````python
# Example 3: Access codes
print("=== Example 3: Access Codes ===")
code = ""
attempts = 0
max_attempts = 3

while code != "NEXUS" and attempts < max_attempts:
    code = input("Enter the access code: ")
    attempts += 1
    
    if code == "NEXUS":
        print("✅ Access granted!")
    elif attempts >= max_attempts:
        print("❌ Too many attempts! System locked.")
    else:
        print(f"Wrong! {max_attempts - attempts} attempts remaining")
````

## System Log 3: Nested Loops – The Time-Within-Time Cycle

**What it is:** A loop can contain another loop inside it – this is called nesting. The inner loop runs completely through on every single pass of the outer loop.

```python
for outer in range(3):
    for inner in range(2):
        # This code runs 3 × 2 = 6 times
```

**Important:** Each additional nesting level needs one more level of indentation!

````python
# Example 1: The sector scans
print("=== Scanning all decks ===")
for deck in range(1, 4):
    print(f"Deck {deck}:")
    for room in range(1, 3):
        print(f"  Room {room} is being scanned...")

# Example 2: A hologram pattern
print("\n=== A hologram pattern ===")
for row in range(3):
    for col in range(4):
        print("✨", end=" ")
    print()
````
