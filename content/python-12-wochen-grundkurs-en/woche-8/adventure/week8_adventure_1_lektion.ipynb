## Data Structure 1: Attributes with Names

**What it is:** A dictionary is like a **profile**: every **attribute** (key) has a **value**. You create it with curly braces.

**The structure of a profile:**
```python
# Create an empty profile
empty_profile = {}

# Fill a profile with attributes
hero = {"name": "Aria", "hero_class": "Mage", "level": 15}
```

**Step-by-step explanation:**
1. **`{}`** – Curly braces define a profile
2. **`:`** – Colon separates the attribute name (key) from the value
3. **`,`** – Commas separate the entries
4. **`[key]`** – Use the key to access the value

**Why it's so useful:**
- 🎯 Attributes with names instead of positions (like in lists)
- 🔍 Fast access via the key
- ⚡ Flexible to extend and modify
- 🎲 Different value types per attribute possible

````python
# Example 1: Creating simple profiles
# Empty profile
empty_profile = {}
print(f"Empty profile: {empty_profile}")

# Hero profile
hero = {
    "name": "Aria",
    "hero_class": "Mage",
    "level": 15,
    "hit_points": 120
}
print(f"Hero profile: {hero}")

# Monster profile
monster = {
    "name": "Dragon",
    "type": "Fire",
    "strength": 50,
    "loot": True
}
print(f"Monster profile: {monster}")
````

````python
# Example 2: Accessing profiles
hero = {
    "name": "Aria",
    "hero_class": "Mage",
    "level": 15,
    "hit_points": 120
}

print("=== Reading the Hero Profile ===")
print(f"Hero name: {hero['name']}")
print(f"Class: {hero['hero_class']}")
print(f"Level: {hero['level']}")
print(f"Hit points: {hero['hit_points']}")

# Using the get() method (safer)
print(f"\nWith get(): {hero.get('name')}")
print(f"Non-existent: {hero.get('mana', 'Unknown')}")
````

````python
# Example 3: Modifying and extending profiles
hero = {
    "name": "Aria",
    "hero_class": "Mage",
    "level": 15
}

print(f"Original: {hero}")

# Change a value
hero["level"] = 16
print(f"After level-up: {hero}")

# Add a new attribute
hero["mana"] = 80
print(f"After adding mana: {hero}")

# Remove an entry
removed = hero.pop("hero_class")
print(f"Removed: {removed}")
print(f"After removal: {hero}")
````

## Data Structure 1b

To process all entries of a dictionary, there are three methods:

| Method | Returns | Example |
|--------|---------|-------|
| `.keys()` | all keys | `for k in hero.keys()` |
| `.values()` | all values | `for v in hero.values()` |
| `.items()` | key + value as pair | `for k, v in hero.items()` |

`.items()` is the most useful – you get both at once.

````python
hero = {
    "name": "Aria",
    "hero_class": "Mage",
    "level": 15,
    "hit_points": 120
}

# .keys() – keys only
print("Attributes:", list(hero.keys()))

# .values() – values only
print("Values:", list(hero.values()))

# .items() – key + value
print("\n=== Hero Profile ===")
for attribute, value in hero.items():
    print(f"  {attribute}: {value}")
````

## Data Structure 2: The Magic Capsule

**The most important artifact properties:**

**Creation:**
- `artifact = (1, 2, 3)` - With round brackets
- `artifact = 1, 2, 3` - Without brackets
- `artifact = (1,)` - One element with a comma

**Immutability:**
- Once created, artifacts cannot be changed
- Perfect for eternal treasures
- Magically protected from modification

````python
# 🔍 Exercises with Artifacts

# Example 1: Creating artifacts
coordinates = (100, 200)
print(f"Coordinates: {coordinates}")

# Without brackets
position = 10, 20
print(f"Position: {position}")

# One element
single = (42,)
print(f"Single artifact: {single}")

# Access elements
print(f"\nX coordinate: {coordinates[0]}")
print(f"Y coordinate: {coordinates[1]}")
````

````python
# Example 2: Artifacts are immutable
spell_status = ("Active", "Stable", "Powerful")
print(f"Spell status: {spell_status}")

# Access works
print(f"Status: {spell_status[0]}")

# Modification does NOT work!
try:
    spell_status[0] = "Inactive"
except TypeError as e:
    print(f"Error: {e}")

# But artifacts inside lists can be replaced
spells = [
    ("Fireball", "Active"),
    ("Healing", "Stable"),
    ("Shield", "Powerful")
]
spells[1] = ("Healing", "Inactive")
print(f"\nSpells after change: {spells}")
````

````python
# Example 3: Artifact unpacking and methods
hero_data = ("Aria", "Mage", 15, 120)

# Artifact unpacking
name, hero_class, level, hit_points = hero_data
print("=== Artifact Unpacking ===")
print(f"Name: {name}")
print(f"Class: {hero_class}")
print(f"Level: {level}")
print(f"Hit points: {hit_points}")

# Artifact methods
print(f"\nNumber of elements: {len(hero_data)}")
print(f"Index of 'Mage': {hero_data.index('Mage')}")
print(f"Count of 'Mage': {hero_data.count('Mage')}")
````

## ✨ Nested Structures – Profiles within Profiles

**The most powerful combinations:**

**Profile within a profile:**
- `person = {"name": "Alex", "equipment": {"weapon": "X", "armor": "Y"}}`

**List of profiles:**
- `heroes = [{"name": "A", "hero_class": "M"}, {"name": "B", "hero_class": "K"}]`

**Artifact (Tuple) as a dictionary key:**
- `positions = {(10, 20): "Point A", (30, 40): "Point B"}`

````python
# Example 1: Profile within a profile
dungeon = {
    "name": "Damnation Fortress",
    "location": {
        "region": "Shadow Lands",
        "coordinates": (100, 200),
        "realm": "Pyralia"
    },
    "inhabitants": {
        "boss": "Lich King",
        "monsters": "Skeletons",
        "traps": "Magical"
    }
}

print("=== Nested Profile ===")
print(f"Dungeon: {dungeon['name']}")
print(f"Region: {dungeon['location']['region']}")
print(f"Coordinates: {dungeon['location']['coordinates']}")
print(f"Boss: {dungeon['inhabitants']['boss']}")
````

````python
# Example 2: List of profiles
heroes = [
    {
        "name": "Aria",
        "hero_class": "Mage",
        "level": 15,
        "experience": 2500
    },
    {
        "name": "Thorin",
        "hero_class": "Warrior",
        "level": 18,
        "experience": 3200
    },
    {
        "name": "Luna",
        "hero_class": "Rogue",
        "level": 12,
        "experience": 1800
    }
]

print("=== Hero List ===")
for hero in heroes:
    print(f"{hero['name']} - {hero['hero_class']} (Level {hero['level']})")

# Filter by level
strong = [h for h in heroes if h['level'] > 14]
print(f"\nStrong heroes: {[h['name'] for h in strong]}")
````

````python
# Example 3: Complex structures
# Artifact (Tuple) as dictionary key
treasure_map = {
    (100, 200): "Gold Chest",
    (300, 400): "Magic Chest",
    (500, 600): "Artifact Chest"
}

print("=== Treasure Map (Tuple as Key) ===")
for position, treasure in treasure_map.items():
    print(f"{position}: {treasure}")

# Complex quest system
quests = {
    "active": [
        {
            "id": 1,
            "goal": "Defeat dragon",
            "reward": ("Gold", 1000, "XP", 500),
            "status": "In Progress"
        }
    ],
    "completed": [
        {
            "id": 2,
            "goal": "Find treasure",
            "reward": ("Silver", 500, "XP", 250),
            "status": "Success"
        }
    ]
}

print(f"\nActive quests: {len(quests['active'])}")
print(f"Completed quests: {len(quests['completed'])}")
````
