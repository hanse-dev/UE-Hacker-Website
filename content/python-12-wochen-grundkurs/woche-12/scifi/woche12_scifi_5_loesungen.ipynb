## 🐛 Debug-Quest Lösungen

```python
# ✅ Lösung Bug #1: Sterne sind unsichtbar

# Problem: Schwarze Sterne auf weißem Grund
import turtle

screen = turtle.Screen()
screen.bgcolor("black")  # Schwarzer Hintergrund

stern = turtle.Turtle()
stern.color("yellow")  # Gelbe Sterne
stern.begin_fill()
stern.fillcolor("white")  # Weiße Füllung
stern.circle(5)
stern.end_fill()
turtle.done()
```

```python
# ✅ Lösung Bug #2: Raumschiff fliegt nicht

# Problem: turtle.done() fehlt
import turtle

schiff = turtle.Turtle()
schiff.shape("turtle")
schiff.color("red")
schiff.penup()  # Stift heben
schiff.goto(-100, 0)
schiff.forward(200)
turtle.done()  # Fehlende Zeile
```

```python
# ✅ Lösung Bug #3: Planet hat keinen Ring

# Problem: pendown() fehlt für den Ring
import turtle

planet = turtle.Turtle()
planet.penup()
planet.goto(0, 0)
planet.pendown()
planet.circle(50)
planet.penup()
planet.goto(0, -50)
planet.pendown()  # Fehlende Zeile
planet.width(5)
planet.circle(50)
turtle.done()
```
