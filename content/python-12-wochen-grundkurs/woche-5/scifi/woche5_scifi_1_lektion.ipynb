## Systemprotokoll 1: Definition - Der Befehlscode

**Was es ist:** Eine Funktion ist ein wiederverwendbarer Protokoll-Block, den du einmal definierst und immer wieder aufrufen kannst.

**Der Aufbau einer Funktion:**
```python
def protokoll_name(parameter1, parameter2):
    """Beschreibung des Protokolls"""
    # Protokoll-Code
    return ergebnis  # Optional: Ergebnis zurückgeben
```

**Schritt-für-Schritt-Erklärung:**
1. **`def`** - Das Kommando, das die Protokoll-Definition startet
2. **`protokoll_name`** - Ein eindeutiger Name für dein Protokoll
3. **`(parameter)`** - System-Variablen (optional)
4. **`:`** - Beendet die Protokoll-Definition
5. **`"""Docstring"""`** - Systemhandbuch (sehr wichtig!)
6. **Eingerückter Code** - Der eigentliche Protokoll-Code
7. **`return`** - Gibt das Ergebnis zurück (optional)

**Warum so nützlich:**
- 🎯 Protokolle wiederverwenden statt neu programmieren
- 🔍 System organisieren und strukturieren
- ⚡ Fehler an einer Stelle beheben
- 🎲 Komplexe Systeme in kleine Teile zerlegen

## Systemprotokoll 1b: Der Schlüssel zur Ausführung

Ein Protokoll zu *definieren* reicht nicht – es muss auch *aufgerufen* werden! Das Aufrufen ist ein eigenes Thema mit wichtigen Regeln.

### Mit und ohne Klammern

**Mit Klammern `funktion()`:** Das Protokoll wird ausgeführt! Die Klammern sagen Python: „Führe diese Funktion jetzt aus“.

**Ohne Klammern `funktion`:** Das Protokoll wird NICHT ausgeführt! Du referenzierst nur das Funktionsobjekt – nichts passiert.

### Aufrufen mit Parametern

Beim Aufruf übergibst du **Argumente** in den Klammern: `funktion(argument1, argument2)`. Die Reihenfolge muss stimmen!

### Vom Argument zur Variable

**Beim Aufruf:** Du übergibst ein **Argument** (z.B. `"Alex"`)

**In der Funktion:** Daraus wird eine **Variable** mit dem Parameternamen (z.B. `name`)

Der Parameter ist ein **Platzhalter**. Beim Aufruf ersetzt Python ihn durch den übergebenen Wert – dieser steht dann als Variable in der Funktion zur Verfügung.

````python
# Mit vs. ohne Klammern
def system_begruessung():
    """Begrüßt Besucher auf der Raumstation"""
    print("Willkommen an Bord der Nebula-7!")

# MIT Klammern: Funktion wird ausgeführt
print("=== Mit Klammern system_begruessung() ===")
system_begruessung()

# OHNE Klammern: Nur Referenz, nichts passiert
print("\n=== Ohne Klammern system_begruessung ===")
print("Typ:", type(system_begruessung))
print("(Die Funktion wurde nicht ausgeführt!)")
````

````python
# Argument wird zur Variable: Schritt für Schritt
def begruesse_piloten(name):  # "name" ist der Parameter (Platzhalter)
    """Begrüßt einen Piloten mit seinem Namen"""
    # In der Funktion ist "name" jetzt eine Variable mit dem übergebenen Wert!
    print(f"In der Funktion: name = '{name}'")
    print(f"Hallo {name}!")

# Beim Aufruf: "Alex" ist das Argument
print("=== Aufruf: begruesse_piloten('Alex') ===")
begruesse_piloten("Alex")  # → name wird zu "Alex"

print("\n=== Aufruf: begruesse_piloten('Zara') ===")
begruesse_piloten("Zara")  # → name wird zu "Zara"
````

````python
# Beispiel 1: Einfache Funktion ohne Parameter
def system_begruessung():
    """Begrüßt Besitzer auf der Raumstation"""
    print("Willkommen an Bord der Nebula-7!")
    print("Deine Mission beginnt jetzt!")

# Funktion aufrufen
print("=== Erster Besucher ===")
system_begruessung()

print("\n=== Zweiter Besucher ===")
system_begruessung()

print("\n=== Was passiert hier? ===")
print("1. def system_begruessung(): definiert die Funktion")
print("2. system_begruessung() ruft die Funktion auf")
print("3. Der Code in der Funktion wird jedes Mal ausgeführt")
````

````python
# Beispiel 2: Funktion mit einem Parameter
def begruesse_piloten(name):
    """Begrüßt einen Piloten mit seinem Namen"""
    print(f"Hallo {name}!")
    print(f"Viel Erfolg bei deiner Mission, {name}!")

# Funktion mit verschiedenen Namen aufrufen
print("=== Verschiedene Piloten ===")
begruesse_piloten("Alex")
begruesse_piloten("Zara")
begruesse_piloten("Nova")

print("\n=== Was passiert hier? ===")
print("1. name ist ein Platzhalter für den übergebenen Wert")
print("2. Jeder Aufruf übergibt einen anderen Namen")
print("3. Die Funktion arbeitet mit dem übergebenen Wert")
````

````python
# Beispiel 3: Funktion mit mehreren Parametern
def erstelle_schiff_profil(name, klasse, besatzung):
    """Erstellt ein Schiff-Profil mit Name, Klasse und Besatzung"""
    print(f"Schiff-Profil:")
    print(f"  Name: {name}")
    print(f"  Klasse: {klasse}")
    print(f"  Besatzung: {besatzung} Personen")
    print(f"  Status: Bereit für den Einsatz!")

# Verschiedene Schiff-Profile erstellen
print("=== Schiff-Profile ===")
erstelle_schiff_profil("Nebula-Explorer", "Forschungsschiff", 150)
print()
erstelle_schiff_profil("Star-Fighter", "Jäger", 2)
print()
erstelle_schiff_profil("Cargo-Hauler", "Frachter", 25)
````

## 🔍 Das Geheimnis von return - Das Systemergebnis

**return** ist der mächtigste Teil einer Funktion!

**Was return tut:**
- Beendet die Funktion sofort
- Gibt einen Wert zurück an den Aufrufer
- Dieser Wert kann in einer Variable gespeichert werden

**Ohne return:** Funktion macht etwas (Side Effect)
**Mit return:** Funktion berechnet und gibt Ergebnis zurück

````python
# 🔍 Übungen mit return

# Beispiel 1: Einfache Berechnung mit return
def berechne_warp_distanz(warp_faktor, zeit):
    """Berechnet die zurückgelegte Distanz im Warp"""
    distanz = warp_faktor * zeit * 1000
    return distanz

# Ergebnis in Variable speichern
print("=== Warp-Distanz-Berechnung ===")
distanz1 = berechne_warp_distanz(5, 10)
print(f"Sprung 1: {distanz1} Lichtjahre")

distanz2 = berechne_warp_distanz(8, 15)
print(f"Sprung 2: {distanz2} Lichtjahre")

# Ergebnis direkt verwenden
print(f"Notfall-Sprung: {berechne_warp_distanz(10, 5)} Lichtjahre")
````

````python
# Beispiel 2: Funktion mit bedingtem return
def pruefe_system_status(energie_level):
    """Prüft, ob das System für eine Mission bereit ist"""
    if energie_level >= 75:
        return "System bereit für Mission!"
    else:
        return f"Noch {75 - energie_level}% Energie needed"

print("=== System-Status-Prüfung ===")
status1 = pruefe_system_status(60)
print(f"Energie 60%: {status1}")

status2 = pruefe_system_status(90)
print(f"Energie 90%: {status2}")

status3 = pruefe_system_status(70)
print(f"Energie 70%: {status3}")
````

````python
# Beispiel 3: Funktion ohne expliziten return (None)
def zeige_ausruestung():
    """Zeigt die Standard-Ausrüstung (gibt nichts zurück)"""
    print("Deine Ausrüstung:")
    print("  - Phaser")
    print("  - Tricorder")
    print("  - Kommunikator")
    # Kein return - gibt automatisch None zurück

print("=== Ausrüstung anzeigen ===")
ergebnis = zeige_ausruestung()
print(f"Rückgabewert: {ergebnis}")
print("Funktionen ohne return geben immer None zurück!")
````

## Systemprotokoll 2: Die Systemhandbücher

**Was es ist:** Docstrings sind Dokumentations-Strings, die erklären, was eine Funktion tut.

**Warum extrem wichtig:**
- 📖 Andere (und du später) verstehen deine Funktion
- 🔍 IDEs zeigen dir Hilfe beim Aufruf
- 🤖 Tools können automatisch Dokumentation erstellen
- ⭐ Zeichen von professionellem Code

**Die drei Arten von Docstrings:**
1. **Einzeilig:** `"""Kurze Beschreibung"""`
2. **Mehrzeilig:** `"""Kurze Beschreibung\n\nDetails..."""`
3. **Vollständig:** `"""Beschreibung\n\nArgs:\nReturns:\n"""`

````python
# Beispiel 1: Verschiedene Docstring-Stile
def einfache_funktion():
    """Dies ist ein einzeiliger Docstring."""
    return "Einfach"

def detaillierte_funktion(schiff_name, flotten_id):
    """Erstellt eine Missions-Beschreibung.
    
    Args:
        schiff_name (str): Der Name des Schiffs
        flotten_id (int): Die ID der Flotte
    
    Returns:
        str: Eine formatierte Beschreibung
    """
    return f"{schiff_name} gehört zur Flotte {flotten_id}."

# Docstring aufrufen (so funktioniert es!)
print("=== Docstring-Beispiele ===")
print(f"Einfacher Docstring: {einfache_funktion.__doc__}")
print()
print(f"Detaillierter Docstring: {detaillierte_funktion.__doc__}")
````

````python
# Beispiel 2: Hilfe-Funktion nutzen
print("=== Hilfe für Funktionen ===")
help(detaillierte_funktion)
````

## Systemprotokoll 3: Die Regeln des Codes

**Gute Funktionsnamen sind entscheidend!**

**Die Regeln:**
- **Kleinbuchstaben** mit Unterstrichen: `berechne_warp_distanz`
- **Beschreibend**: Was die Funktion **tut** (Verb)
- **Kurz aber klar**: `get_name()` nicht `retrieve_starship_data_from_database()`
- **Keine Verben im Namen**: `name_get()` falsch, `get_name()` richtig

**Beispiele:**
- ✅ `berechne_warp_distanz()`
- ✅ `erstelle_schiff_profil()`
- ✅ `pruefe_system_status()`
- ❌ `BerechneWarpDistanz()` (CamelCase)
- ❌ `berechne_das_warp()` (zu lang)
- ❌ `warp_berechnen()` (Substantiv zuerst)

````python
# ✨ Beispiele für gute und schlechte Namen

# ✅ Gute Namen (klar und beschreibend)
def berechne_schild_effizienz(energie, flaeche):
    """Berechnet die Effizienz der Schilde."""
    return (energie / flaeche) * 100

def finde_naechsten_hafen(schiff_position, hafen_liste):
    """Findet den nächsten Hafen zur aktuellen Position."""
    return min(hafen_liste, key=lambda h: abs(h - schiff_position))

def ist_kommunikation_moeglich(reichweite, distanz):
    """Prüft, ob Kommunikation möglich ist."""
    return distanz <= reichweite

# ❌ Schlechte Namen (unklar oder falsche Konvention)
# def berechneSchildEffizienz():  # CamelCase
# def se(e, f):  # Zu kurz, nicht beschreibend
# def shield_calculation():  # Substantiv zuerst

print("=== Gute Funktionsnamen in Aktion ===")
effizienz = berechne_schild_effizienz(1000, 50)
print(f"Schild-Effizienz: {effizienz}%")

position = 100
haefen = [50, 200, 350, 500]
naechster_hafen = finde_naechsten_hafen(position, haefen)
print(f"Nächster Hafen: {naechster_hafen}")

komm_ok = ist_kommunikation_moeglich(1000, 750)
print(f"Kommunikation möglich: {komm_ok}")
````
