## 🗣️ Kapitel 1: Die gemeinsame Sprache – f-Strings

Bevor du Elemente kombinierst, brauchst du eine Sprache, um über sie zu sprechen. Ein **f-String** ist die **modernere und lesbarere** Art, Variablen in Text einzubauen. Das `f` vor den Anführungszeichen sagt Python: *"Schau in die geschweiften Klammern!"*

**f-String vs. `+`-Verkettung – was soll ich benutzen?**

| Methode | Beispiel | Empfehlung |
|---------|---------|------------|
| `+`-Verkettung | `"Hallo " + name` | funktioniert, braucht `str()` für Zahlen |
| f-String | `f"Hallo {name}"` | moderner, kürzer, Zahlen funktionieren direkt |

> ✅ Benutze f-Strings – sie sind klarer zu lesen und weniger fehleranfällig.
> Die `+`-Methode aus Woche 1 funktioniert aber weiterhin.

````python
# Beispiel 1: Einfache f-Strings
name = "Aria"
level = 5
print(f"Hallo {name}!")
print(f"Du bist Level {level}!")

# Beispiel 2: Mit Berechnungen
gold = 100
bonus = 25
print(f"Dein Gold: {gold + bonus}")

# Beispiel 3: Mehrere Variablen
held = "Thorin"
klasse = "Krieger"
waffe = "Axt"
print(f"{held} der {klasse} schwingt seine {waffe}!")
````

## 🔥🪨💧💨 Kapitel 2: Die vier Elemente

Die Turmwächterin zeigt dir die vier Elementsteine. Jeder hat eine eigene **Natur** – in Python nennen wir das den **Typ** einer Variable.

| Element | Typ | Fachbegriff | Beispiel | Natur |
|---------|-----|------------|---------|-------|
| 🔥 Feuer | Text | `str` | `"Feuerball"` | ausgesprochene Worte, Namen, Zaubersprüche |
| 🪨 Erde | Ganze Zahl | `int` | `42` | fest und zählbar – Level, Gold, Punkte |
| 💧 Wasser | Kommazahl | `float` | `3.14` | fließend, nie ganz genau – Prozente, Schaden |
| 💨 Luft | Wahrheitswert | `bool` | `True` / `False` | unsichtbar, nur da oder nicht da – Schalter |

**Zwei nützliche Hilfsbefehle, um ein Element zu bestimmen:**
- `type(x)` – zeigt dir das Element (den Typ) einer Variable an
- `len(x)` – zeigt dir die Länge eines Feuer-Elements (Textes)

```python
name = "Aria"
print(type(name))   # <class 'str'> -> Feuer-Element
print(len(name))    # 4
```

````python
# Beispiel 1: 🔥 Feuer - Text-Elemente
zauber = "Feuerball"
beschreibung = "Ein mächtiger Feuerzauber"
print(f"Zauber: {zauber}")
print(f"Element: {type(zauber)}")

# Beispiel 2: 🪨 Erde - Ganze Zahlen
level = 12
lebenspunkte = 150
print(f"Level: {level}, LP: {lebenspunkte}")
print(f"Element: {type(level)}")

# Beispiel 3: 💧 Wasser - Kommazahlen
schaden = 23.5
prozent = 87.25
print(f"Schaden: {schaden}, Trefferchance: {prozent}%")
print(f"Element: {type(schaden)}")

# Beispiel 4: 💨 Luft - Wahrheitswerte
magie_aktiv = True
vergiftet = False
print(f"Magie aktiv: {magie_aktiv}")
print(f"Vergiftet: {vergiftet}")
print(f"Element: {type(magie_aktiv)}")
````

## ⚗️ Kapitel 3: Elemente verschmelzen

**Was es ist:** Wie du Elemente kombinierst, veränderst und von einer Natur in eine andere wandelst.

**Wichtig:** Nicht alle Elemente können direkt kombiniert werden! 🔥 Feuer (Text) und 🪨 Erde (Zahl) widersetzen sich einer direkten Verschmelzung – du brauchst eine Umwandlung (`str()`, `int()`, `float()`).

````python
# Beispiel 1: 🪨 Erde verschmelzen - mathematische Operationen
gold_a = 150
gold_b = 75
print(f"Gesamtgold: {gold_a + gold_b}")
print(f"Differenz: {gold_a - gold_b}")
print(f"Doppeltes Gold: {gold_a * 2}")
print(f"Geteilt: {gold_a / 3}")

# Beispiel 2: 🔥 Feuer verschmelzen - String-Operationen
vorname = "Luna"
nachname = "Silbermond"
titel = "die Weise"
voller_name = vorname + " " + nachname + " " + titel
print(f"Voller Name: {voller_name}")
print(f"Name mal 3: {vorname * 3}")

# Beispiel 3: Elemente umwandeln (Erde <-> Feuer)
alter_str = "25"
alter_int = int(alter_str)
print(f"Alter als Feuer-Element (Text): {alter_str} ({type(alter_str)})")
print(f"Alter als Erd-Element (Zahl): {alter_int} ({type(alter_int)})")
````

## 🔮 Kapitel 4: Das Orakel befragt

`input()` lässt dich das Orakel befragen. Die Antwort ist immer ein **🔥 Feuer-Element (String)** – auch wenn jemand eine Zahl tippt. Für 🪨 Erde oder 💧 Wasser musst du den Wert also noch umwandeln:

```python
name = input("Wie heißt du? ")        # ergibt ein Feuer-Element (str)
alter = int(input("Wie alt bist du? ")) # Feuer -> Erd-Element (str -> int)
```

> ⚠️ In diesen Notebooks läuft `input()` nur, wenn du das Notebook lokal ausführst oder in einer interaktiven Umgebung bist.

````python
# Beispiel: Das Orakel befragen und die Antwort umwandeln
# (lokal ausführen – in statischen Umgebungen ggf. kommentieren)

name = input("Wie heißt dein Held? ")
level_str = input("Welches Level hat er? ")
level = int(level_str)   # Feuer -> Erde (str -> int)

print(f"Willkommen, {name}!")
print(f"{name} ist auf Level {level}.")
print(f"In 5 Levels erreichst du Level {level + 5}.")
````
