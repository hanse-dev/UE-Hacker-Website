## 🐛 Debug Solution Suggestions

**Explanation:** Python reads `System` and `online` as variable names – but these variables don't exist. Text must always be in quotation marks so Python knows: this is text, not a variable.

```python
# Problem: text without quotation marks
print("System online")
```

**Explanation:** `level` is a number (Integer), the text before it is a String. Python cannot directly connect these two types with `+` – it's like trying to tie a stone to a thread. `str()` solves this: it converts the number to text.

```python
# Problem: number cannot be directly connected with text
level = 3
print("Security level: " + str(level))
```

**Explanation:** The variable is called `status`, but in `print()` it says `statu` – one letter missing. Python searches exactly for the name `statu` and cannot find it. Typos in variable names are one of the most common beginner errors!

```python
# Problem: variable 'statu' does not exist
status = "Active"
print("System status: " + status)
```
