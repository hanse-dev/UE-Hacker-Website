| Term | Meaning | Example |
|------|---------|---------|
| `import` | Load a module (toolbox) | `import random` |
| `import ... as ...` | Load a module with a nickname | `import random as r` |
| `from ... import ...` | Load only certain spells | `from string import ascii_letters` |
| `random.randint(a, b)` | Random whole number from a to b | `random.randint(1, 6)` |
| `random.choice(list)` | Random element from a list | `random.choice(paths)` |
| `random.shuffle(list)` | Shuffle a list randomly (changes the list) | `random.shuffle(cards)` |
| `random.sample(list, k)` | k random, distinct elements | `random.sample(list, 3)` |
| `string.ascii_letters` | All letters a-z and A-Z as text | Ingredient for rune passwords |
| `string.digits` | All digits 0-9 as text | Ingredient for rune passwords |
| `time.sleep(seconds)` | Pause the program for a while | `time.sleep(2)` |
| `datetime.now()` | Current date and time | `datetime.now().hour` |

```python
import random
import string
import time

# random module
roll = random.randint(1, 6)
print("Dice roll:", roll)

heroes = ["Aria", "Borin", "Lena"]
print("Randomly chosen:", random.choice(heroes))

random.shuffle(heroes)
print("Shuffled:", heroes)

# string module
rune = random.choice(string.ascii_letters)
print("Random rune:", rune)

# time module
print("Casting spell...")
time.sleep(1)
print("Done!")
```
