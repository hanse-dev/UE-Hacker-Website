```python
class Pferd:
    def __init__(self, name, rasse="Unbekannt"):
        self.name = name
        self.rasse = rasse

    def vorstellen(self):
        print(f"Ich bin {self.name}, ein {self.rasse}!")

    def trainieren(self):
        self.rasse = self.rasse  # Platzhalter
        print(f"{self.name} trainiert!")

# Objekte erstellen
pferd1 = Pferd("Bobby")
pferd2 = Pferd("Blitz", rasse="Haflinger")

pferd1.vorstellen()   # Ich bin Bobby, ein Unbekannt!
pferd1.trainieren()   # Bobby trainiert!
print(pferd2.rasse)   # Haflinger
```
