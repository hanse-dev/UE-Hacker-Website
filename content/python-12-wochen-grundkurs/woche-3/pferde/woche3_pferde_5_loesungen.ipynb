## 🐛 Debug-Quest Lösungen

```python
# ✅ Lösung Bug #1: Falsche Einrückung

# Problem: Code nach if nicht eingerückt
level = 10
if level >= 10:
    print("Reitfertigkeit erreicht!")
    print("Glückwunsch!")
```

```python
# ✅ Lösung Bug #2: Falscher Vergleich

# Problem: = statt == für Vergleich verwendet
name = "Thunder"
if name == "Thunder":
    print("Hallo Thunder!")
```

```python
# ✅ Lösung Bug #3: Vergessene Bedingung

# Problem: elif braucht immer eine Bedingung
alter = 8
if alter >= 5:
    print("Ausgebildet")
else:
    print("In Ausbildung")
```
