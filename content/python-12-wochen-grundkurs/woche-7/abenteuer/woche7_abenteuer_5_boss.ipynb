### ⭐⭐⭐⭐☆ Boss-Quest 1: Der Würfelturm des Archivars

**Belohnung:** 500 XP + Simulations-Kristall

Im Herzen der Bibliothek steht ein alter Würfelturm – wer die höchste Zahl würfelt, gewinnt die Schatzsuche!

**Schritt 1 – Würfeln:**
Erzeuge für 3 Spieler je einen Würfelwurf (1–6) mit `random.randint()` und gib die Ergebnisse aus.

**Schritt 2 – Gewinner:**
Finde heraus, wer die höchste Zahl gewürfelt hat, und kündige den Gewinner an.

**Schritt 3 – Turnier:**
Lass 5 Runden laufen und zähle, wer am häufigsten gewinnt. Gib am Ende die Gesamtstatistik aus.

**Bonus:** Füge einen "Magiewürfel" (1–20) als Joker-Runde hinzu.

### ⭐⭐⭐⭐⭐ Boss-Quest 2: Das Namens-Orakel

**Belohnung:** 600 XP + Elementarstab

Das Orakel der Bibliothek liest dein Schicksal aus deinem Namen und den Runen des Zufalls.

**Schritt 1 – Glücks-Rune:**
Lass den Benutzer seinen Namen eingeben. Ziehe mit `random.choice()` so viele zufällige Runen aus `string.ascii_uppercase`, wie der Name Buchstaben hat, und füge sie zu einer "Glücks-Rune" zusammen.

**Schritt 2 – Prophezeiung:**
Erstelle eine Liste mit 6 Prophezeiungen und wähle mit `random.choice()` eine passende aus.

**Schritt 3 – Horoskop:**
Kombiniere Name, Glücks-Rune und Prophezeiung zu einem persönlichen Horoskop-Text und gib ihn aus.

**Bonus:** Zeige an, wie hoch die Wahrscheinlichkeit dieser Prophezeiung war (in %, gerundet).
