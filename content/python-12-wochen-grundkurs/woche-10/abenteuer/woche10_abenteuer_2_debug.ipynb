### 🐛 Bug #2

**Was passiert:** Methoden in Klassen brauchen immer `self` als ersten Parameter. Ohne `self` weiß die Methode nicht, zu welchem Objekt sie gehört.

```python
# 🐛 Bug #2: Falscher Konstruktor

class Magier:
    def __init(name, mana):
        name = name
        mana = mana

merlin = Magier("Merlin", 150)
print(merlin.name)
```

### 🐛 Bug #3

**Was passiert:** Methoden in Klassen brauchen immer `self` als ersten Parameter. Ohne `self` weiß die Methode nicht, zu welchem Objekt sie gehört.

```python
# 🐛 Bug #3: Methode ohne self

class Drache:
    def __init__(self, name):
        self.name = name
    
    def vorstellen():
        print(f"Ich bin {self.name}!")

drache = Drache("Smaug")
drache.vorstellen()
```
