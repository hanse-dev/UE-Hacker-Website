A young wizard tried to create their first classes, but something went wrong! Find the bugs:

### 🐛 Bug #1

What is wrong? Find and fix the bug!

```python
class swordsman:
    def __init__(self, name):
        self.name = name

fighter = swordsman()
print(fighter.name)
```

### 🐛 Bug #2

What is wrong? Find and fix the bug!

```python
class Wizard:
    def __init(name, mana):
        name = name
        mana = mana

merlin = Wizard("Merlin", 150)
print(merlin.name)
```

### 🐛 Bug #3

What is wrong? Find and fix the bug!

```python
class Dragon:
    def __init__(self, name):
        self.name = name
    
    def introduce():
        print(f"I am {self.name}!")

dragon = Dragon("Smaug")
dragon.introduce()
```
