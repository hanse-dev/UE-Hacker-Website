## Systemprotokoll 1: String-Befehl

**Was es ist:** Der f-String ist die moderne Methode, Text und Variablen zu präzisen Befehlen zu fusionieren.

**Wichtig:** Immer ein `f` vor die Anführungszeichen setzen!

```python
# Beispiel 1: Einfache f-Strings
pilot = "Zara"
rang = 5
print(f"Hallo {pilot}!")
print(f"Dein Rang: {rang}")

# Beispiel 2: Mit Berechnungen
energie_a = 75
energie_b = 25
print(f"Gesamtenergie: {energie_a + energie_b}%")

# Beispiel 3: Mehrere Variablen
schiff = "Nebula-Falcon"
kapitaen = "Orion"
mission = "Expedition Alpha"
print(f"{schiff} unter {kapitaen} startet {mission}!")
```

## Systemprotokoll 2: Typen

**Was es ist:** Jede Information auf der Nebula-7 hat einen Quanten-Typ - wie die vier Aggregatzustände der Materie.

**Wichtig:** `type()` zeigt dir den Quanten-Typ jeder Information!

```python
# Beispiel 1: Strings - Textdaten
schiff_name = "Starlight"
klasse = "Kreuzer"
print(f"Schiff: {schiff_name}")
print(f"Klasse: {klasse}")
print(f"Typ: {type(schiff_name)}")

# Beispiel 2: Integers - Ganze Zahlen
crew_anzahl = 250
deck_anzahl = 15
print(f"Crew: {crew_anzahl}, Decks: {deck_anzahl}")
print(f"Typ: {type(crew_anzahl)}")

# Beispiel 3: Floats - Präzise Werte
geschwindigkeit = 2.5
koordinate_x = 127.35
print(f"Geschwindigkeit: {geschwindigkeit} Lichtjahre/h")
print(f"Position X: {koordinate_x}")
print(f"Typ: {type(geschwindigkeit)}")

# Beispiel 4: Booleans - Systemzustände
system_online = True
alarm_aktiv = False
print(f"System online: {system_online}")
print(f"Alarm aktiv: {alarm_aktiv}")
print(f"Typ: {type(system_online)}")
```

## Systemprotokoll 3: Operationen und Transformationen

**Was es ist:** Wie du Daten kombinierst, veränderst und transformierst.

**Wichtig:** Nicht alle Datentypen können direkt kombiniert werden!

```python
# Beispiel 1: Mathematische Operationen
planeten_a = 8
planeten_b = 4
print(f"Gesamtplaneten: {planeten_a + planeten_b}")
print(f"Differenz: {planeten_a - planeten_b}")
print(f"Verdopplung: {planeten_a * 2}")
print(f"Pro Sektor: {planeten_a / 2}")

# Beispiel 2: String-Operationen
prefix = "Neo"
suffix = "tron"
system_name = prefix + suffix
print(f"Systemname: {system_name}")
print(f"Signal: {prefix * 3}")

# Beispiel 3: Datentransformation
temperatur_str = "273"
temperatur_int = int(temperatur_str)
print(f"Temperatur als Text: {temperatur_str}K (Typ: {type(temperatur_str)})")
print(f"Temperatur als Zahl: {temperatur_int}K (Typ: {type(temperatur_int)})")

# Beispiel 4: Sensordaten verarbeiten
# Sensor-Eingabe simuliert (später mit input())
sensor_wert = "42.5"
wert_float = float(sensor_wert)
print(f"Sensorwert {wert_float} verdoppelt: {wert_float * 2}")
```
