## 📚 Werkzeug-Sammlung 1: Der Import-Kasten - Das erste Modul

**Was es ist:** Ein Modul ist eine Sammlung von vordefinierten Werkzeugen, die du in dein Programm laden kannst.

**Der Aufbau eines Imports:**
```python
# Ganzes Modul importieren
import math

# Mit Alias importieren
import math as m

# Nur bestimmte Werkzeuge importieren
from math import pi, sqrt
```

**Schritt-für-Schritt-Erklärung:**
1. **`import`** - Schlüsselwort zum Laden
2. **`math`** - Name der Werkzeugkiste
3. **`as`** - Optionaler Spitzname
4. **`from`** - Importiert spezifische Werkzeuge

**Warum so nützlich:**
- 🎯 Werkzeuge nicht neu bauen müssen
- 🔍 Professionelle, optimierte Ausrüstung
- ⚡ Dein Stall bleibt übersichtlich
- 🎲 Zugriff auf spezialisierte Werkzeuge

````python
# Beispiel 1: Module importieren
# Ganzes Modul importieren
import math
print(f"Pi aus math: {math.pi}")

# Mit Alias importieren
import random as r
print(f"Zufallszahl: {r.randint(1, 10)}")

# Nur bestimmte Werkzeuge importieren
from math import sqrt, sin, cos
print(f"Wurzel aus 16: {sqrt(16)}")
print(f"Sinus 90°: {sin(1.57)}")
print(f"Kosinus 0°: {cos(0)}")
````

````python
# Beispiel 2: Verschiedene Import-Methoden vergleichen
# Methode 1: import math
import math
print("Methode 1 - import math:")
print(f"  math.pi: {math.pi}")
print(f"  math.sqrt(25): {math.sqrt(25)}")

# Methode 2: import math as m
import math as m
print("\nMethode 2 - import math as m:")
print(f"  m.pi: {m.pi}")
print(f"  m.sqrt(25): {m.sqrt(25)}")

# Methode 3: from math import pi, sqrt
from math import pi, sqrt
print("\nMethode 3 - from math import pi, sqrt:")
print(f"  pi: {pi}")
print(f"  sqrt(25): {sqrt(25)}")
````

````python
# Beispiel 3: Mehrere Module importieren
import math
import random
import time

# Berechne eine zufällige Reitdauer
distanz = 1000  # Meter
geschwindigkeit = random.uniform(0.1, 0.9)  # Bruchteil der Maximalgeschwindigkeit
reitzeit = distanz / geschwindigkeit

print("=== Reit-Simulation ===")
print(f"Distanz: {distanz} Meter")
print(f"Geschwindigkeit: {geschwindigkeit:.2f} m/s")
print(f"Reitzeit: {reitzeit:.1f} Minuten")
print(f"Warte 2 Sekunden...")
time.sleep(2)
print(f"Ankunft am Ziel!")
````

## 📦 Werkzeug-Sammlung 2: Das math-Modul - Der Rechen-Bock

**Die wichtigsten math-Werkzeuge:**

**Konstanten:**
- `math.pi` - Die Zahl Pi (3.14159...)
- `math.e` - Eulersche Zahl (2.71828...)
- `math.inf` - Unendlich

**Grundwerkzeuge:**
- `math.sqrt(x)` - Quadratwurzel
- `math.pow(x, y)` - x hoch y
- `math.ceil(x)` - Aufrunden
- `math.floor(x)` - Abrunden
- `math.fabs(x)` - Absolutwert

````python
# 🔍 Übungen mit math-Konstanten
import math

print("=== Mathematische Konstanten ===")
print(f"Pi (π): {math.pi}")
print(f"Eulersche Zahl (e): {math.e}")
print(f"Unendlich: {math.inf}")
print(f"Tau (2π): {math.tau}")

# Praktische Anwendung: Kreisberechnungen
radius = 10
umfang = 2 * math.pi * radius
flaeche = math.pi * radius ** 2
print(f"\nKreis mit r={radius}:")
print(f"  Umfang: {umfang:.2f}")
print(f"  Fläche: {flaeche:.2f}")
````

````python
# Beispiel 2: Mathematische Grundwerkzeuge
import math

print("=== Mathematische Werkzeuge ===")

# Wurzel und Potenzen
print(f"sqrt(16): {math.sqrt(16)}")
print(f"pow(2, 3): {math.pow(2, 3)}")
print(f"2**3: {2**3}")

# Runden
zahl = 3.7
print(f"\nZahl: {zahl}")
print(f"round({zahl}): {round(zahl)}")
print(f"math.ceil({zahl}): {math.ceil(zahl)}")
print(f"math.floor({zahl}): {math.floor(zahl)}")

# Absolutwert
print(f"\nmath.fabs(-5.5): {math.fabs(-5.5)}")
print(f"abs(-5.5): {abs(-5.5)}")
````

````python
# Beispiel 3: Fortgeschrittene Werkzeuge
import math

print("=== Fortgeschrittene Werkzeuge ===")

# Winkelfunktionen (in Bogenmaß!)
winkel_grad = 45
winkel_rad = math.radians(winkel_grad)
print(f"{winkel_grad}° = {winkel_rad:.3f} rad")
print(f"sin({winkel_grad}°): {math.sin(winkel_rad):.3f}")
print(f"cos({winkel_grad}°): {math.cos(winkel_rad):.3f}")
print(f"tan({winkel_grad}°): {math.tan(winkel_rad):.3f}")

# Logarithmen
print(f"\nlog(10): {math.log(10):.3f}")
print(f"log10(100): {math.log10(100)}")
print(f"log2(8): {math.log2(8)}")

# Spezielle Werkzeuge
print(f"\nfactorial(5): {math.factorial(5)}")
print(f"gcd(24, 36): {math.gcd(24, 36)}")
print(f"lcm(12, 15): {math.lcm(12, 15)}")
````

## ✨ Werkzeug-Sammlung 3: Das random-Modul - Das Zufalls-Pferd

**Die mächtigsten Zufalls-Werkzeuge:**

**Zahlen erzeugen:**
- `random.randint(a, b)` - Zufällige Ganzzahl von a bis b
- `random.uniform(a, b)` - Zufällige Kommazahl von a bis b
- `random.random()` - Zufallszahl zwischen 0 und 1

**Aus Listen wählen:**
- `random.choice(liste)` - Ein zufälliges Element
- `random.choices(liste, k=n)` - n zufällige Elemente
- `random.shuffle(liste)` - Liste mischen

**Verteilungen:**
- `random.gauss(mu, sigma)` - Normalverteilung
- `random.expovariate(lambda)` - Exponentialverteilung

````python
# Beispiel 1: Zufallszahlen erzeugen
import random

print("=== Zufallszahlen ===")

# Ganzzahlen
print(f"Zufällige Zahl 1-6: {random.randint(1, 6)}")
print(f"Zufällige Zahl 0-100: {random.randint(0, 100)}")

# Kommazahlen
print(f"Zufallszahl 0-1: {random.random():.3f}")
print(f"Zufallszahl 10-20: {random.uniform(10, 20):.2f}")

# Mit Seeds reproduzierbar
random.seed(42)
print(f"Mit Seed 42: {random.randint(1, 100)}")
random.seed(42)
print(f"Mit Seed 42: {random.randint(1, 100)} (gleiches Ergebnis!)")
````

````python
# Beispiel 2: Aus Listen wählen
import random

print("=== Zufallsauswahl aus Listen ===")

# Pferde auf dem Hof
pferde = ["Thunder", "Stormy", "Sunny", "Shadow", "Blaze", "Misty"]
print(f"Alle Pferde: {pferde}")

# Ein zufälliges Pferd
ziel = random.choice(pferde)
print(f"Zielpferd: {ziel}")

# Drei zufällige Pferde
gruppe = random.choices(pferde, k=3)
print(f"Reitgruppe: {gruppe}")

# Liste mischen
reihenfolge = pferde.copy()
random.shuffle(reihenfolge)
print(f"Gemischte Reihenfolge: {reihenfolge}")
````

````python
# Beispiel 3: Fortgeschrittene Zufallswerkzeuge
import random
import math

print("=== Fortgeschrittene Zufallswerkzeuge ===")

# Normalverteilung für Futterwerte
print("Futterverbrauch (Normalverteilung):")
for i in range(5):
    futter = random.gauss(100, 20)  # Mittelwert 100, Stdabw 20
    print(f"  Pferd {i+1}: {max(0, futter):.1f} kg")

# Exponentialverteilung für Wartezeiten
print("\nWartezeiten (Exponentialverteilung):")
for i in range(5):
    warten = random.expovariate(0.1)  # Lambda = 0.1
    print(f"  Wartezeit {i+1}: {warten:.1f} Minuten")

# Zufälliger Punkt im Kreis
def zufalls_punkt_kreis(radius):
    winkel = random.uniform(0, 2 * math.pi)
    r = radius * math.sqrt(random.random())
    x = r * math.cos(winkel)
    y = r * math.sin(winkel)
    return x, y

x, y = zufalls_punkt_kreis(10)
print(f"\nZufälliger Punkt im Kreis r=10: ({x:.2f}, {y:.2f})")
````
