### ⭐⭐⭐⭐☆ Mission 3: The Champion's Arena

```python
# Step 1 – Create heroes
hero1_name = "Thorin"
hero1_level = 12
hero1_strength = 90
hero1_intelligence = 45

hero2_name = "Lyra"
hero2_level = 11
hero2_strength = 55
hero2_intelligence = 95

# Step 2 – Calculate total stats
total1 = hero1_strength + hero1_intelligence
total2 = hero2_strength + hero2_intelligence

# Step 3 – Determine stronger hero
print(f"{hero1_name} (Level {hero1_level}): total stats {total1}")
print(f"{hero2_name} (Level {hero2_level}): total stats {total2}")
print()

if total1 > total2:
    winner = hero1_name
else:
    winner = hero2_name
print(f"The stronger hero is: {winner}!")

# Step 4 – Battle announcement
print(f"{winner} wins the arena challenge!")
average_strength = (total1 + total2) / 2
print(f"Average total strength: {average_strength}")

# Bonus
difference = abs(total1 - total2)
print(f"The winner is stronger by {difference} points.")
```

### ⭐⭐⭐⭐⭐ Boss Quest 2: The Magic Workshop

```python
# Step 1 – Create spells
s1_name = "Phoenix Fire"
s1_power = 92.5
s1_complexity = 5
s1_duration = 8

s2_name = "Frost Breath"
s2_power = 68.0
s2_complexity = 3
s2_duration = 4

s3_name = "Thunder Call"
s3_power = 75.5
s3_complexity = 4
s3_duration = 6

s4_name = "Light Shard"
s4_power = 55.0
s4_complexity = 2
s4_duration = 3

s5_name = "Shadow Wave"
s5_power = 83.0
s5_complexity = 4
s5_duration = 7

# Step 2 – Output descriptions
print("=== MAGIC WORKSHOP PROTOCOL ===")
print(f"{s1_name}: Power {s1_power}, Complexity {s1_complexity}, Duration {s1_duration}s")
print(f"{s2_name}: Power {s2_power}, Complexity {s2_complexity}, Duration {s2_duration}s")
print(f"{s3_name}: Power {s3_power}, Complexity {s3_complexity}, Duration {s3_duration}s")
print(f"{s4_name}: Power {s4_power}, Complexity {s4_complexity}, Duration {s4_duration}s")
print(f"{s5_name}: Power {s5_power}, Complexity {s5_complexity}, Duration {s5_duration}s")
print()

# Step 3 – Sum complexity
total_complexity = s1_complexity + s2_complexity + s3_complexity + s4_complexity + s5_complexity
print(f"Total complexity: {total_complexity}")

# Step 4 – Most powerful and most complex
print(f"Most powerful spell: {s1_name} ({s1_power})")
print(f"Most complex spell: {s1_name} (complexity {s1_complexity})")

# Step 5 – Average power
total_power = s1_power + s2_power + s3_power + s4_power + s5_power
average_power = total_power / 5
print(f"Average power: {average_power}")

# Step 6 – Boolean marking
s1_above = s1_power > average_power
s2_above = s2_power > average_power
s3_above = s3_power > average_power
s4_above = s4_power > average_power
s5_above = s5_power > average_power
print(f"{s1_name} above average: {s1_above}")
print(f"{s2_name} above average: {s2_above}")
print(f"{s3_name} above average: {s3_above}")
print(f"{s4_name} above average: {s4_above}")
print(f"{s5_name} above average: {s5_above}")

# Bonus: efficiency
print()
print(f"Efficiency {s1_name}: {s1_power / s1_complexity:.2f}")
print(f"Efficiency {s2_name}: {s2_power / s2_complexity:.2f}")
print(f"Efficiency {s3_name}: {s3_power / s3_complexity:.2f}")
print(f"Efficiency {s4_name}: {s4_power / s4_complexity:.2f}")
print(f"Efficiency {s5_name}: {s5_power / s5_complexity:.2f}")
```

### ⭐⭐⭐⭐☆ Boss Quest 3: The Quest Planning

```python
# Step 1 – Define quests
q1_goal = "Clear the dragon's lair"
q1_days = 7
q1_reward = 1200
q1_group_size = 5
q1_success = 0.8

q2_goal = "Recover lost relics"
q2_days = 3
q2_reward = 600
q2_group_size = 3
q2_success = 0.95

# Step 2 – Quest protocol
print("=== QUEST PROTOCOL ===")
print(f"Quest 1: {q1_goal}")
print(f"  Duration: {q1_days} days, Reward: {q1_reward} gold, Group: {q1_group_size}, Success: {q1_success * 100}%")
print(f"Quest 2: {q2_goal}")
print(f"  Duration: {q2_days} days, Reward: {q2_reward} gold, Group: {q2_group_size}, Success: {q2_success * 100}%")
print()

# Step 3 – Totals
total_reward = q1_reward + q2_reward
total_group = q1_group_size + q2_group_size
print(f"Total reward: {total_reward} gold")
print(f"Total group size: {total_group} people")
print()

# Step 4 – Forecast success
expected1 = q1_reward * q1_success
expected2 = q2_reward * q2_success
print(f"Expected reward Quest 1: {expected1} gold")
print(f"Expected reward Quest 2: {expected2} gold")
print()

# Step 5 – Compare quests
print(f"Reward difference: {q1_reward - q2_reward} gold")
print(f"Duration difference: {q1_days - q2_days} days")
print(f"Success difference: {(q2_success - q1_success) * 100}%")

# Step 6 – Profitability
q1_more_profitable = expected1 > expected2
print(f"Quest 1 more profitable: {q1_more_profitable}")

# Bonus: reward per group member
print()
print(f"Reward per person Quest 1: {q1_reward / q1_group_size} gold")
print(f"Reward per person Quest 2: {q2_reward / q2_group_size} gold")

print()
print("🎉 +400 XP: Boss Quest complete!")
print("🏆 You defeated the Golem of Confused Forms!")
print("⭐ Title earned: Master of Elements")
print()
print("🎊 CONGRATULATIONS! You have mastered Week 2!")
print("📚 Next week: Conditions (if-else)!")
```
