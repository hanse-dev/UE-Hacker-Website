## 🗣️ Chapter 1: The Common Tongue – f-strings

Before you combine elements, you need a language to talk about them. An **f-string** is the **more modern and readable** way to embed variables in text. The `f` before the quotation marks tells Python: *"Look inside the curly braces!"*

**f-string vs. `+` concatenation – which should I use?**

| Method | Example | Recommendation |
|--------|---------|-----------------|
| `+` concatenation | `"Hello " + name` | works, needs `str()` for numbers |
| f-string | `f"Hello {name}"` | more modern, shorter, numbers work directly |

> ✅ Use f-strings – they're clearer to read and less error-prone.
> The `+` method from Week 1 still works, though.

````python
# Example 1: Simple f-strings
name = "Aria"
level = 5
print(f"Hello {name}!")
print(f"You are Level {level}!")

# Example 2: With calculations
gold = 100
bonus = 25
print(f"Your gold: {gold + bonus}")

# Example 3: Multiple variables
hero = "Thorin"
character_class = "Warrior"
weapon = "Axe"
print(f"{hero} the {character_class} swings their {weapon}!")
````

## 🔥🪨💧💨 Chapter 2: The Four Elements

The Tower Guardian shows you the four elemental stones. Each has its own **nature** – in Python we call this the **type** of a variable.

| Element | Type | Term | Example | Nature |
|---------|------|------|---------|--------|
| 🔥 Fire | Text | `str` | `"Fireball"` | spoken words, names, spells |
| 🪨 Earth | Whole number | `int` | `42` | solid and countable - level, gold, points |
| 💧 Water | Decimal number | `float` | `3.14` | flowing, never quite exact - percentages, damage |
| 💨 Air | Truth value | `bool` | `True` / `False` | invisible, either there or not - switches |

**Two useful commands to determine an element:**
- `type(x)` - shows you the element (type) of a variable
- `len(x)` - shows you the length of a Fire element (text)

```python
name = "Aria"
print(type(name))   # <class 'str'> -> Fire element
print(len(name))    # 4
```

````python
# Example 1: 🔥 Fire - text elements
spell = "Fireball"
description = "A powerful fire spell"
print(f"Spell: {spell}")
print(f"Element: {type(spell)}")

# Example 2: 🪨 Earth - whole numbers
level = 12
hit_points = 150
print(f"Level: {level}, HP: {hit_points}")
print(f"Element: {type(level)}")

# Example 3: 💧 Water - decimal numbers
damage = 23.5
percent = 87.25
print(f"Damage: {damage}, Hit chance: {percent}%")
print(f"Element: {type(damage)}")

# Example 4: 💨 Air - truth values
magic_active = True
poisoned = False
print(f"Magic active: {magic_active}")
print(f"Poisoned: {poisoned}")
print(f"Element: {type(magic_active)}")
````

## ⚗️ Chapter 3: Fusing Elements

**What it is:** How you combine, change and convert elements from one nature into another.

**Important:** Not all elements can be fused directly! 🔥 Fire (text) and 🪨 Earth (number) resist direct fusion - you need a conversion (`str()`, `int()`, `float()`).

````python
# Example 1: Fusing 🪨 Earth - mathematical operations
gold_a = 150
gold_b = 75
print(f"Total gold: {gold_a + gold_b}")
print(f"Difference: {gold_a - gold_b}")
print(f"Double gold: {gold_a * 2}")
print(f"Divided: {gold_a / 3}")

# Example 2: Fusing 🔥 Fire - string operations
first_name = "Luna"
last_name = "Silvermoon"
title = "the Wise"
full_name = first_name + " " + last_name + " " + title
print(f"Full name: {full_name}")
print(f"Name x3: {first_name * 3}")

# Example 3: Converting elements (Earth <-> Fire)
age_str = "25"
age_int = int(age_str)
print(f"Age as Fire element (text): {age_str} ({type(age_str)})")
print(f"Age as Earth element (number): {age_int} ({type(age_int)})")
````

## 🔮 Chapter 4: Consulting the Oracle

`input()` lets you consult the oracle. The answer is always a **🔥 Fire element (string)** - even if someone types a number. For 🪨 Earth or 💧 Water you still need to convert the value:

```python
name = input("What is your name? ")        # gives a Fire element (str)
age = int(input("How old are you? "))       # Fire -> Earth element (str -> int)
```

> ⚠️ In these notebooks `input()` only works when you run the notebook locally or in an interactive environment.

````python
# Example: Consulting the oracle and converting the answer
# (run locally - comment out in static environments if needed)

name = input("What is your hero's name? ")
level_str = input("What level are they? ")
level = int(level_str)   # Fire -> Earth (str -> int)

print(f"Welcome, {name}!")
print(f"{name} is at level {level}.")
print(f"In 5 levels you will reach level {level + 5}.")
````
