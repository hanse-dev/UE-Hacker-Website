# 📖 Gesamtglossar – Python 12-Wochen-Grundkurs

> Hier findest du alle Begriffe aus allen 12 Wochen auf einen Blick.
> Sortiert nach Woche – so siehst du auch, wann ein Begriff eingeführt wurde.

## Woche 1 – Grundlagen

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `print()` | Gibt Text auf dem Bildschirm aus | `print("Hallo")` |
| **Variable** | Behälter mit einem Namen, der einen Wert speichert | `name = "Aria"` |
| **String** | Text – immer in Anführungszeichen | `"Hallo Welt"` |
| **Integer** | Ganze Zahl ohne Komma | `42` |
| **Boolean** | Ja/Nein-Wert – entweder `True` oder `False` | `bereit = True` |
| `str()` | Wandelt eine Zahl in Text um | `str(42)` → `"42"` |
| `input()` | Liest eine Eingabe vom Benutzer ein (immer String) | `name = input("Name? ")` |
| **Kommentar** | Erklärung im Code mit `#` – wird nicht ausgeführt | `# Das ist ein Kommentar` |
| **Fehler/Bug** | Wenn der Code nicht funktioniert | `NameError`, `TypeError` |


## Woche 2 – Datentypen & Strings

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `f-String` | Moderne Textformatierung mit Variablen direkt im Text | `f"Hallo {name}"` |
| `Float` | Kommazahl | `3.14` |
| `type()` | Datentyp einer Variable prüfen | `type(42)` → `<class 'int'>` |
| `len()` | Länge eines Textes oder einer Liste | `len("Hallo")` → `5` |
| `int()` | Umwandlung in ganze Zahl | `int("42")` → `42` |
| `float()` | Umwandlung in Kommazahl | `float("3")` → `3.0` |
| `bool()` | Umwandlung in Wahrheitswert | `bool(0)` → `False` |
| `.upper()` | Text in Großbuchstaben | `"hallo".upper()` → `"HALLO"` |
| `.lower()` | Text in Kleinbuchstaben | `"HALLO".lower()` → `"hallo"` |
| `.strip()` | Leerzeichen am Rand entfernen | `" hi ".strip()` → `"hi"` |
| `.split()` | Text an einem Zeichen aufteilen | `"a,b".split(",")` → `["a","b"]` |
| `.replace()` | Teiltext durch anderen ersetzen | `"Hallo".replace("H","h")` → `"hallo"` |


## Woche 3 – Bedingungen

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `if` | Wenn-Bedingung | `if x > 5:` |
| `elif` | Sonst-Wenn – weitere Bedingung | `elif x == 5:` |
| `else` | Sonst – Code wenn alle Bedingungen falsch sind | `else:` |
| `==` | Ist gleich (Vergleich, nicht Zuweisung!) | `x == 5` |
| `!=` | Ist ungleich | `x != 5` |
| `<` / `>` | Kleiner als / Größer als | `x < 10` |
| `and` | Beide Bedingungen müssen wahr sein | `x > 0 and x < 10` |
| `or` | Mindestens eine Bedingung muss wahr sein | `x < 0 or x > 10` |
| `not` | Kehrt einen Wahrheitswert um | `not True` → `False` |


## Woche 4 – Schleifen & Listen-Grundlagen

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `for` | Schleife – geht jedes Element einer Sammlung durch | `for item in liste:` |
| `range()` | Erzeugt eine Zahlenfolge | `range(5)` → `0,1,2,3,4` |
| `while` | Schleife – wiederholt solange Bedingung wahr ist | `while x < 10:` |
| **Liste** `[]` | Geordnete Sammlung von Werten | `["Aria", "Borin"]` |
| `.append()` | Element am Ende der Liste hinzufügen | `liste.append("neu")` |
| **Index** | Position eines Elements in der Liste (beginnt bei 0) | `liste[0]` → erstes Element |


## Woche 5 – Funktionen

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `def` | Funktion definieren – wiederverwendbaren Code-Block erstellen | `def begrüße(name):` |
| **Parameter** | Eingabewert einer Funktion | `def add(a, b):` |
| `return` | Ergebnis aus der Funktion zurückgeben | `return a + b` |
| **Docstring** | Beschreibung einer Funktion in dreifachen Anführungszeichen | `"""Was die Funktion tut"""` |
| `snake_case` | Schreibregel: Wörter mit Unterstrich verbinden | `meine_funktion()` |
| `try` | Code ausführen und auf Fehler vorbereiten | `try:` |
| `except` | Fehler abfangen und behandeln | `except ValueError:` |


## Woche 6 – Listen (vollständig)

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `.insert()` | Element an einer bestimmten Position einfügen | `liste.insert(1, "x")` |
| `.remove()` | Erstes Vorkommen eines Elements entfernen | `liste.remove("x")` |
| `.pop()` | Letztes Element entfernen und zurückgeben | `liste.pop()` |
| `.index()` | Position eines Elements finden | `liste.index("x")` |
| `.count()` | Wie oft ein Element vorkommt | `liste.count("x")` |
| `in` | Prüfen ob ein Element in der Liste vorhanden ist | `"x" in liste` |
| `.sort()` | Liste aufsteigend sortieren (verändert die Liste) | `liste.sort()` |
| `sorted()` | Gibt eine sortierte Kopie zurück | `sorted(liste)` |
| `enumerate()` | Index und Wert gleichzeitig beim Durchlaufen | `for i, x in enumerate(liste):` |
| `break` | Schleife sofort beenden | `if x == 5: break` |
| `continue` | Aktuelle Runde überspringen | `if x == 0: continue` |
| `set` | Menge – wie Liste, aber ohne Duplikate | `set([1,1,2])` → `{1, 2}` |
| **Algorithmus *(= Schritt-für-Schritt-Lösungsweg)*** | Feste Abfolge von Schritten um ein Problem zu lösen | `for x in liste: if x > max: max = x` |


## Woche 7 – Module & Bibliotheken

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `import` | Ein Modul (Werkzeugkasten) laden | `import math` |
| `from … import` | Nur bestimmte Teile eines Moduls laden | `from math import sqrt` |
| `import … as` | Modul mit Kurzname laden | `import random as r` |
| `math.sqrt()` | Quadratwurzel berechnen | `math.sqrt(16)` → `4.0` |
| `math.floor()` | Abrunden auf ganze Zahl | `math.floor(3.9)` → `3` |
| `math.ceil()` | Aufrunden auf ganze Zahl | `math.ceil(3.1)` → `4` |
| `math.pi` | Die Kreiszahl Pi | `math.pi` → `3.14159…` |
| `random.randint()` | Zufällige ganze Zahl in einem Bereich | `random.randint(1, 6)` |
| `random.choice()` | Zufälliges Element aus einer Liste | `random.choice(liste)` |
| `random.shuffle()` | Liste zufällig durchmischen | `random.shuffle(liste)` |


## Woche 8 – Dictionaries & Tupel

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| **Dictionary** `{}` | Sammlung von Schlüssel-Wert-Paaren | `{"name": "Aria", "level": 5}` |
| **Schlüssel (Key)** | Name eines Eintrags im Dictionary | `held["name"]` |
| **Wert (Value)** | Inhalt eines Eintrags | `held["level"]` → `5` |
| `.get()` | Sicherer Zugriff – kein Fehler wenn Schlüssel fehlt | `held.get("xp", 0)` |
| `.pop()` | Eintrag löschen und zurückgeben | `held.pop("level")` |
| `.update()` | Mehrere Einträge auf einmal ändern | `held.update({"level": 2})` |
| `.keys()` | Alle Schlüssel | `held.keys()` |
| `.values()` | Alle Werte | `held.values()` |
| `.items()` | Alle Schlüssel-Wert-Paare – nützlich für Schleifen | `for k, v in held.items():` |
| **Tupel** `()` | Unveränderliche geordnete Sammlung | `punkt = (3, 5)` |
| **Tupel-Unpacking** | Tupel-Werte direkt in Variablen aufteilen | `x, y = (3, 5)` |


## Woche 9 – Dateien & I/O

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `open()` | Datei öffnen | `open("datei.txt", "r")` |
| `with` | Datei sicher öffnen und automatisch schließen | `with open("datei.txt") as f:` |
| `"r"` / `"w"` / `"a"` | Lese- / Schreib- / Anhänge-Modus | `open("f.txt", "w")` |
| `.read()` | Gesamten Datei-Inhalt einlesen | `inhalt = f.read()` |
| `.write()` | Text in Datei schreiben | `f.write("Hallo")` |
| `json.load()` | JSON-Datei einlesen → Python-Dictionary | `daten = json.load(f)` |
| `json.dump()` | Python-Dictionary als JSON speichern | `json.dump(daten, f)` |
| `csv.reader` | CSV-Datei zeilenweise einlesen | `reader = csv.reader(f)` |
| `FileNotFoundError` | Fehler wenn eine Datei nicht gefunden wird | `except FileNotFoundError:` |


## Woche 10 – OOP Grundlagen

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `class` | Eine eigene Klasse (Bauplan) erstellen | `class Held:` |
| **Objekt** | Ein einzelnes erstelltes Exemplar einer Klasse | `held = Held("Aria")` |
| **Konstruktor** *(= Methode die beim Erstellen automatisch startet)* | `__init__` wird beim Erstellen jedes Objekts aufgerufen | `def __init__(self, name):` |
| `self` | Verweis auf das eigene Objekt innerhalb der Klasse | `self.name = name` |
| **Attribut** | Eigenschaft / Variable eines Objekts | `held.name` |
| **Methode** | Funktion die zu einem Objekt gehört | `held.angreifen()` |
| **Instanz** | Ein einzelnes erstelltes Objekt einer Klasse | `held = Held()` |


## Woche 11 – OOP Fortgeschritten

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| **Vererbung** | Eine Klasse übernimmt alle Eigenschaften einer anderen | `class Magier(Held):` |
| `super()` | Auf die Elternklasse zugreifen | `super().__init__(name)` |
| **Polymorphismus** *(= gleiche Methode, unterschiedliches Ergebnis)* | Gleiche Methode – je nach Klasse ein anderes Verhalten | `tier.laut()` → Hund: Wuff, Katze: Miau |
| `__str__` | Textdarstellung – aufgerufen von `print()` und `str()` | `def __str__(self): return f"{self.name}"` |
| `__add__` | `+`-Operator für eigene Klassen | `def __add__(self, other):` |
| `__eq__` | `==`-Operator für eigene Klassen | `def __eq__(self, other):` |


## Woche 12 – Grafik mit Turtle

| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `turtle` | Python-Modul für grafische Zeichnungen | `import turtle` |
| `turtle.Turtle()` | Ein neues Turtle-Objekt (Zeichenstift) erstellen | `t = turtle.Turtle()` |
| `.forward()` | Turtle vorwärts bewegen | `t.forward(100)` |
| `.left()` / `.right()` | Turtle drehen (in Grad) | `t.left(90)` |
| `.color()` | Stiftfarbe setzen | `t.color("red")` |
| `.begin_fill()` / `.end_fill()` | Fläche füllen | `t.begin_fill()` … `t.end_fill()` |
| `.circle()` | Kreis zeichnen | `t.circle(50)` |
| `.speed()` | Zeichengeschwindigkeit setzen | `t.speed(5)` |
| `.penup()` / `.pendown()` | Stift heben / senken | `t.penup()` |
