**Erklärung:** Python liest `Hallo` und `Welt` als Variablennamen – aber diese Variablen gibt es nicht. Text muss immer in Anführungszeichen stehen, damit Python weiß: das ist Text, keine Variable.

```python
# Problem: Text ohne Anführungszeichen
print("Hallo Welt")
```

**Erklärung:** `alter` ist eine Zahl (Integer), der Text davor und danach ist ein String. Python kann diese zwei Typen nicht direkt mit `+` verbinden – das ist wie versuchen, einen Stein an einen Faden zu knoten. `str()` löst das: es verwandelt die Zahl in Text.

```python
# Problem: Zahl kann nicht direkt mit Text verbunden werden
alter = 25
print("Ich bin " + str(alter) + " Jahre alt.")
```

**Erklärung:** Die Variable heißt `name`, aber im `print()` steht `nam` – ein Buchstabe fehlt. Python sucht exakt nach dem Namen `nam` und findet ihn nicht. Tippfehler in Variablennamen sind einer der häufigsten Anfängerfehler!

```python
# Problem: Variable 'nam' existiert nicht
name = "Gandalf"
print("Willkommen, " + name)
```
