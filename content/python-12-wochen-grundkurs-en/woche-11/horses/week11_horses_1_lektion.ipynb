## Lesson 1: Inheritance

**What it is:** Inheritance lets you derive one horse class from another. The child class gets all the properties of the parent class and can add its own.

**Important:** Use `super()` to call the parent methods!

```python
# Example 1: Simple inheritance
class Horse:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def eat(self):
        print(f"{self.name} eats hay")

class RidingHorse(Horse):
    def ride(self, speed):
        print(f"{self.name} rides at {speed} km/h")

# Example 2: Overriding methods
class Thoroughbred(RidingHorse):
    def eat(self):
        print(f"{self.name} eats special high-energy feed")

    def gallop(self):
        print(f"{self.name} gallops like the wind! 🐎")

# Example 3: Using super()
class Coldblood(Horse):
    def __init__(self, name, age, pulling_force):
        super().__init__(name, age)
        self.pulling_force = pulling_force

    def pull(self, load):
        print(f"{self.name} pulls {load} with {self.pulling_force}kg of force")
```

## Lesson 2

**What it is:** Polymorphism means "many forms". Different horse types can react differently to the same instruction.

**Important:** Duck Typing – "If it walks like a horse and neighs like a horse, it's a horse!"

```python
# Example 1: Method polymorphism
class DressageHorse:
    def special_ability(self):
        print("Elegant Piaffe! 💃")

class JumpingHorse:
    def special_ability(self):
        print("High Jump! 🦘")

class WesternHorse:
    def special_ability(self):
        print("Sliding Stop! 🤠")

# Example 2: Polymorphic function
def complete_training(horse):
    print(f"Training for {horse.name}:")
    horse.special_ability()

# Example 3: Operator polymorphism
class Horsepower:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Horsepower(self.value + other.value)

    def __str__(self):
        return f"{self.value} horsepowers"
```

## Lesson 3

**What it is:** Magic Methods (dunders) are special methods with __ at the beginning and end. They give your horse classes special abilities.

**Important:** The most important Magic Methods are __init__, __str__, __add__, __len__!

```python
# Example 1: __str__ and __repr__
class Horse:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def __str__(self):
        return f"🐴 {self.name} ({self.breed})"

    def __repr__(self):
        return f"Horse('{self.name}', '{self.breed}')"

# Example 2: __len__ and __getitem__
class Herd:
    def __init__(self):
        self.horses = []

    def __len__(self):
        return len(self.horses)

    def __getitem__(self, index):
        return self.horses[index]

    def add_horse(self, horse):
        self.horses.append(horse)

# Example 3: __eq__ and __lt__
class Champion:
    def __init__(self, name, points):
        self.name = name
        self.points = points

    def __eq__(self, other):
        return self.points == other.points

    def __lt__(self, other):
        return self.points < other.points

    def __str__(self):
        return f"{self.name} ({self.points} points)"
```
