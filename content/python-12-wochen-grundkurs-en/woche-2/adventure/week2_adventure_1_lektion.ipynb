## Spell Formula 1: Strings – modern text formatting

An f-String is the **more modern and readable** way to embed variables in text. The `f` before the quotation marks tells Python: *"Look inside the curly braces!"*

**f-String vs. `+` concatenation – which should I use?**

| Method | Example | Recommendation |
|--------|---------|---------------|
| `+` concatenation | `"Hello " + name` | works, needs `str()` for numbers |
| f-String | `f"Hello {name}"` | more modern, shorter, numbers work directly |

> ✅ Use f-Strings – they are clearer to read and less error-prone.
> The `+` method from Week 1 still works too.

````python
# Example 1: Simple f-Strings
name = "Aria"
level = 5
print(f"Hello {name}!")
print(f"You are Level {level}!")

# Example 2: With calculations
gold = 100
bonus = 25
print(f"Your gold: {gold + bonus}")

# Example 3: Multiple variables
hero = "Thorin"
hero_class = "Warrior"
weapon = "Axe"
print(f"{hero} the {hero_class} swings his {weapon}!")
````

## Spell Formula 2

Every variable in Python has a **type** – it determines what you can do with it.

| Type | Technical term | Example | For |
|------|---------------|---------|-----|
| Text | `str` | `"Hello"` | Names, texts, spells |
| Whole number | `int` | `42` | Level, gold, points |
| Decimal number | `float` | `3.14` | Percentages, damage values |
| Yes/No | `bool` | `True` / `False` | Switches, states |

**Two useful helper commands:**
- `type(x)` – shows you the type of a variable
- `len(x)` – shows you the length of a text (or later a list)

```python
name = "Aria"
print(type(name))   # <class 'str'>
print(len(name))    # 4
```

````python
# Example 1: Strings – text elements
spell = "Fireball"
description = "A powerful fire spell"
print(f"Spell: {spell}")
print(f"Type: {type(spell)}")

# Example 2: Integers – whole numbers
level = 12
hit_points = 150
print(f"Level: {level}, HP: {hit_points}")
print(f"Type: {type(level)}")

# Example 3: Floats – decimal numbers
damage = 23.5
hit_chance = 87.25
print(f"Damage: {damage}, Hit chance: {hit_chance}%")
print(f"Type: {type(damage)}")

# Example 4: Booleans – truth values
magic_active = True
poisoned = False
print(f"Magic active: {magic_active}")
print(f"Poisoned: {poisoned}")
print(f"Type: {type(magic_active)}")
````

## Spell Formula 3: Operations and Conversions

**What it is:** How you combine, modify, and convert elements from one form to another.

**Important:** Not all elements can be directly combined!

````python
# Example 1: Mathematical operations
gold_a = 150
gold_b = 75
print(f"Total gold: {gold_a + gold_b}")
print(f"Difference: {gold_a - gold_b}")
print(f"Double gold: {gold_a * 2}")
print(f"Divided: {gold_a / 3}")

# Example 2: String operations
first_name = "Luna"
last_name = "Silverwood"
title = "the Wise"
full_name = first_name + " " + last_name + " " + title
print(f"Full name: {full_name}")
print(f"Name x3: {first_name * 3}")

# Example 3: Type conversion
age_str = "25"
age_int = int(age_str)
print(f"Age as text: {age_str} (type: {type(age_str)})")
print(f"Age as number: {age_int} (type: {type(age_int)})")

# Example 4: Process input
# Input simulated (later with input())
user_input = "42"
number = int(user_input)
print(f"The number {number} doubled is {number * 2}")
````

## Spell Formula 4: Consult the Oracle

`input()` reads input from the user. The result is always a **String** – even if someone types a number. For numbers you therefore need to convert the value:

```python
name = input("What is your name? ")         # gives a String
age = int(input("How old are you? "))       # String → Integer
```

> ⚠️ In these notebooks, `input()` only runs when you execute the notebook locally or in an interactive environment.

````python
# Example: Read and process input
# (run locally – may need to comment out in static environments)

name = input("What is your hero's name? ")
level_str = input("What level is your hero? ")
level = int(level_str)   # String → Integer

print(f"Welcome, {name}!")
print(f"{name} is at level {level}.")
print(f"In 5 levels you will reach level {level + 5}.")
````
