## 📚 Zauberformel 1: Die for-Schleife – Den Zirkel der Dinge

Eine `for`-Schleife führt Code für **jedes Element** in einer Sammlung aus. Das natürlichste Muster ist das Durchlaufen einer Liste:

```python
for element in liste:
    # Code für jedes Element
```

**Schritt für Schritt:**
1. Python nimmt das erste Element aus der Liste
2. Speichert es in `element`
3. Führt den eingerückten Code aus
4. Wiederholt das für jedes weitere Element

`range(n)` ist ein Sonderfall – er erzeugt eine Folge von Zahlen (0, 1, 2 … n-1) wenn du keine echte Liste hast, aber eine bestimmte Anzahl Wiederholungen brauchst.

````python
# Beispiel 1: Einfache for-Schleife mit range()
print("=== Beispiel 1: Zählen von 0 bis 4 ===")
# range(5) erzeugt die Zahlen 0, 1, 2, 3, 4
for i in range(5):
    print(f"Durchlauf {i}: Der Wert von i ist {i}")

print("\n=== Was passiert hier? ===")
print("1. range(5) erstellt: [0, 1, 2, 3, 4]")
print("2. i wird bei jedem Durchlauf auf das nächste Element gesetzt")
print("3. Die Schleife endet nach dem letzten Element")
````

````python
# Beispiel 2: For-Schleife mit Start und Ende
print("=== Beispiel 2: Zählen von 3 bis 7 ===")
# range(3, 8) erzeugt die Zahlen 3, 4, 5, 6, 7 (8 ist exklusiv!)
for zahl in range(3, 8):
    print(f"Zahl: {zahl}")

print("\n=== Wichtige Beobachtung ===")
print("range(start, ende) enthält start, aber nicht ende!")
print("Das ist wie bei Slices: [start:ende] ist ende-exklusiv")
````

````python
# Beispiel 3: For-Schleife mit Schrittweite
print("=== Beispiel 3: Jede zweite Zahl von 0 bis 10 ===")
# range(0, 11, 2) erzeugt: 0, 2, 4, 6, 8, 10
for gerade in range(0, 11, 2):
    print(f"Gerade Zahl: {gerade}")

print("\n=== Die Schrittweite ===")
print("range(start, ende, schritt) springt immer 'schritt' Positionen weiter")
print("Hier: start=0, ende=11, schritt=2 → 0, 2, 4, 6, 8, 10")
````

````python
# Beispiel 4: For-Schleife über einen String
print("=== Beispiel 4: Buchstaben eines Namens ===")
name = "Aria"
print(f"Der Name ist: {name}")
print("Buchstaben für Buchstabe:")

for buchstabe in name:
    print(f"  - {buchstabe}")

print("\n=== Was passiert hier? ===")
print("String wird als Sequenz von Buchstaben behandelt")
print("Jeder Buchstabe wird einzeln durchlaufen")
````

````python
# Beispiel 5: For-Schleife über eine Liste
print("=== Beispiel 5: Inventar durchsuchen ===")
inventar = ["Schwert", "Schild", "Trank", "Schlüssel"]
print("Dein Inventar:")

for gegenstand in inventar:
    print(f"  - {gegenstand}")

print("\n=== Mit Index und Wert ===")
# enumerate gibt uns sowohl Index als auch Wert
for index, gegenstand in enumerate(inventar):
    print(f"  Slot {index}: {gegenstand}")
````

## 🔍 Das Geheimnis von range() - Der Zahlenzauber

**range() ist dein mächtigster Verbündeter für Zahlenfolgen!**

### Die drei Formen von range():

1. **`range(n)`** - Zahlen von 0 bis n-1
   ```python
   range(5) → 0, 1, 2, 3, 4
   ```

2. **`range(start, ende)`** - Zahlen von start bis ende-1
   ```python
   range(2, 7) → 2, 3, 4, 5, 6
   ```

3. **`range(start, ende, schritt)`** - Zahlen mit Abständen
   ```python
   range(1, 10, 2) → 1, 3, 5, 7, 9
   ```

**Wichtige Regeln:**
- Die **obere Grenze ist immer exklusiv** (wird nicht mitgezählt)
- **Negative Schritte** sind möglich für rückwärts zählen
- range() ist **sehr speichereffizient** (erzeugt Zahlen erst bei Bedarf)

````python
# 🔍 Übungen mit range()

# Übung 1: countdown
print("=== Countdown von 5 bis 1 ===")
for i in range(5, 0, -1):
    print(f"{i}...")
print("🚀 Start!")

# Übung 2: Gerade Zahlen finden
print("\n=== Gerade Zahlen bis 20 ===")
for zahl in range(0, 21, 2):
    print(zahl, end=" ")
print()

# Übung 3: Vielfache von 3
print("\n=== Vielfache von 3 bis 30 ===")
for i in range(3, 31, 3):
    print(f"3 × {i//3} = {i}")
````

## 📦 Zauberformel 2: Die while-Schleife - Der Wächter der Bedingung

**Was es ist:** Eine while-Schleife wiederholt Code, solange eine Bedingung wahr ist.

**Die Anatomie einer while-Schleife:**
```python
while bedingung:
    # Code, der wiederholt wird, solange bedingung wahr ist
    # WICHTIG: Die Bedingung muss irgendwann falsch werden!
```

**Warum nützlich:**
- ⏳ Wiederhole, bis etwas eintritt
- 🎮 Spiele-Schleifen (solange spieler lebt)
- 🔍 Nutzeingaben validieren
- ⚡ Unbekannte Anzahl von Wiederholungen

**⚠️ Vorsicht:** Endlosschleifen entstehen, wenn die Bedingung niemals falsch wird!

````python
# Beispiel 1: Einfache while-Schleife
print("=== Beispiel 1: Bis 5 zählen ===")
zaehler = 0
while zaehler <= 5:
    print(f"Zaehler: {zaehler}")
    zaehler += 1  # WICHTIG: Hochzählen, sonst Endlosschleife!

print("\n=== Was passiert hier? ===")
print("1. zaehler startet bei 0")
print("2. Solange zaehler <= 5, wird der Code ausgeführt")
print("3. zaehler += 1 erhöht den Wert bei jedem Durchlauf")
print("4. Bei zaehler = 6 wird die Bedingung falsch → Schleife endet")
````

````python
# Beispiel 2: Spiel - Leben und Energie
print("=== Beispiel 2: Spiel-Schleife ===")
leben = 100
runde = 1

while leben > 0:
    schaden = 20
    leben -= schaden
    print(f"Runde {runde}: -{schaden} Leben, übrig: {leben}")
    runde += 1

print(f"\n💀 Spiel vorbei nach {runde-1} Runden!")
````

````python
# Beispiel 3: Benutzereingabe validieren
print("=== Beispiel 3: Passwort-Abfrage ===")
passwort = ""
versuche = 0
max_versuche = 3

while passwort != "Geheim" and versuche < max_versuche:
    passwort = input("Gib das Passwort ein: ")
    versuche += 1
    
    if passwort == "Geheim":
        print("✅ Zugang gewährt!")
    elif versuche >= max_versuche:
        print("❌ Zu viele Versuche!")
    else:
        print(f"Falsch! Noch {max_versuche - versuche} Versuche")
````

## 📦 Zauberformel 3: Listen-Grundlagen – Die erste Sammlung

Eine Liste speichert mehrere Werte unter einem Namen. Damit werden `for`-Schleifen erst richtig nützlich.

```python
inventar = ["Schwert", "Schild", "Trank"]  # Liste erstellen
inventar.append("Schlüssel")               # Element hinzufügen
print(inventar[0])                         # Erstes Element: "Schwert"
print(len(inventar))                       # Anzahl: 4
```

**Das Wichtigste:**
- Eckige Klammern `[]` – Liste erstellen
- `.append(wert)` – Element ans Ende anfügen
- `liste[0]` – erstes Element (Zählung startet bei 0)
- `len(liste)` – Anzahl der Elemente

> 💡 Alle weiteren Listen-Methoden kommen in Woche 6.

````python
# Listen erstellen und befüllen
inventar = ["Schwert", "Schild", "Trank"]
print(f"Inventar: {inventar}")
print(f"Anzahl: {len(inventar)}")

# Element hinzufügen
inventar.append("Schlüssel")
print(f"Nach append: {inventar}")

# Per Index zugreifen (Zählung startet bei 0!)
print(f"Erstes Element:  {inventar[0]}")
print(f"Zweites Element: {inventar[1]}")
print(f"Letztes Element: {inventar[-1]}")

# for-Schleife über eine Liste
print("\nDein Inventar:")
for gegenstand in inventar:
    print(f"  - {gegenstand}")
````

**Beispiel-Code zum Starten:**
```python
import random

print("=== ENDLOSER DUNGEON ===")

tiefe = 0
leben = 100
mana = 100

while True:
    tiefe += 1
    raum_schatz = random.randint(1, 100)
    leben -= 5

    print(f"Raum {tiefe}: Schatzwert {raum_schatz}, Leben {leben}, Mana {mana}")

    if leben <= 0:
        print("💀 Leben erschöpft!")
        break
```

````python
# ✅ Lösung Bug #2: Falsche range-Grenze

# Problem: range(5) gibt 0-4, aber wir wollen 1-5
for i in range(1, 6):  # Start bei 1, ende bei 6 (exklusiv)
    print(f"Zahl {i}")
````
