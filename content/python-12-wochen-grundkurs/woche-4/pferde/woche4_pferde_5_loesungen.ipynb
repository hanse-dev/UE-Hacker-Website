**Erklärung:** Dieser Code enthält einen Fehler – findest du ihn?

```python
# Problem: range(5) gibt 0-4, aber wir wollen 1-5
for i in range(1, 6):  # Start bei 1, ende bei 6 (exklusiv)
    print(f"Pferd {i}")
```

**Erklärung:** Python erwartet einen eingerückten Block. Alles was zu einem `if`, `for`, `while` oder `def` gehört, muss mit 4 Leerzeichen eingerückt werden.

```python
# Problem: Code nach for nicht eingerückt
for i in range(3):
    print(f"Übung {i}")
print("Training beendet")
```
