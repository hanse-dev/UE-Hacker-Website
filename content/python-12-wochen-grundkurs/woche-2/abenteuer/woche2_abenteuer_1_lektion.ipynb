## Zauberformel 1: Strings – die moderne Textformatierung

Ein f-String ist die **modernere und lesbarere** Art, Variablen in Text einzubauen. Das `f` vor den Anführungszeichen sagt Python: *"Schau in die geschweiften Klammern!"*

**f-String vs. `+`-Verkettung – was soll ich benutzen?**

| Methode | Beispiel | Empfehlung |
|---------|---------|------------|
| `+`-Verkettung | `"Hallo " + name` | funktioniert, braucht `str()` für Zahlen |
| f-String | `f"Hallo {name}"` | moderner, kürzer, Zahlen funktionieren direkt |

> ✅ Benutze f-Strings – sie sind klarer zu lesen und weniger fehleranfällig.
> Die `+`-Methode aus Woche 1 funktioniert aber weiterhin.

````python
# Beispiel 1: Einfache f-Strings
name = "Aria"
level = 5
print(f"Hallo {name}!")
print(f"Du bist Level {level}!")

# Beispiel 2: Mit Berechnungen
gold = 100
bonus = 25
print(f"Dein Gold: {gold + bonus}")

# Beispiel 3: Mehrere Variablen
held = "Thorin"
klasse = "Krieger"
waffe = "Axt"
print(f"{held} der {klasse} schwingt seine {waffe}!")
````

## Zauberformel 2

Jede Variable in Python hat einen **Typ** – er bestimmt, was man damit tun kann.

| Typ | Fachbegriff | Beispiel | Wofür |
|-----|------------|---------|-------|
| Text | `str` | `"Hallo"` | Namen, Texte, Zaubersprüche |
| Ganze Zahl | `int` | `42` | Level, Gold, Punkte |
| Kommazahl | `float` | `3.14` | Prozente, Schadenswerte |
| Ja/Nein | `bool` | `True` / `False` | Schalter, Zustände |

**Zwei nützliche Hilfsbefehle:**
- `type(x)` – zeigt dir den Typ einer Variable an
- `len(x)` – zeigt dir die Länge eines Textes (oder später einer Liste)

```python
name = "Aria"
print(type(name))   # <class 'str'>
print(len(name))    # 4
```

````python
# Beispiel 1: Strings - Textelemente
zauber = "Feuerball"
beschreibung = "Ein mächtiger Feuerzauber"
print(f"Zauber: {zauber}")
print(f"Typ: {type(zauber)}")

# Beispiel 2: Integers - Ganze Zahlen
level = 12
lebenspunkte = 150
print(f"Level: {level}, LP: {lebenspunkte}")
print(f"Typ: {type(level)}")

# Beispiel 3: Floats - Kommazahlen
schaden = 23.5
prozent = 87.25
print(f"Schaden: {schaden}, Trefferchance: {prozent}%")
print(f"Typ: {type(schaden)}")

# Beispiel 4: Booleans - Wahrheitswerte
magie_aktiv = True
vergiftet = False
print(f"Magie aktiv: {magie_aktiv}")
print(f"Vergiftet: {vergiftet}")
print(f"Typ: {type(magie_aktiv)}")
````

## Zauberformel 3: Operationen und Umwandlungen

**Was es ist:** Wie du Elemente kombinierst, veränderst und von einer Form in eine andere wandelst.

**Wichtig:** Nicht alle Elemente können direkt kombiniert werden!

````python
# Beispiel 1: Mathematische Operationen
gold_a = 150
gold_b = 75
print(f"Gesamtgold: {gold_a + gold_b}")
print(f"Differenz: {gold_a - gold_b}")
print(f"Doppeltes Gold: {gold_a * 2}")
print(f"Geteilt: {gold_a / 3}")

# Beispiel 2: String-Operationen
vorname = "Luna"
nachname = "Silbermond"
titel = "die Weise"
voller_name = vorname + " " + nachname + " " + titel
print(f"Voller Name: {voller_name}")
print(f"Name mal 3: {vorname * 3}")

# Beispiel 3: Typumwandlung
alter_str = "25"
alter_int = int(alter_str)
print(f"Alter als Text: {alter_str} (Typ: {type(alter_str)})")
print(f"Alter als Zahl: {alter_int} (Typ: {type(alter_int)})")

# Beispiel 4: Eingaben verarbeiten
# Eingabe simuliert (später mit input())
eingabe = "42"
zahl = int(eingabe)
print(f"Die Zahl {zahl} verdoppelt ist {zahl * 2}")
````
