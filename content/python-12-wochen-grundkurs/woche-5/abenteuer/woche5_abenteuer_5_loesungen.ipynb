**Erklärung:** Dieser Code enthält einen Fehler – findest du ihn?

```python
# Problem: return vergessen
def addiere(a, b):
    ergebnis = a + b
    return ergebnis  # Das fehlte!

summe = addiere(5, 3)
print(f"Summe: {summe}")  # Jetzt gibt es 8 aus!
```
