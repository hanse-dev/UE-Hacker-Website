## 🔬 What is a system protocol?

A **system protocol** is like a command set for the space station – a fixed instruction that the computer understands and can execute.

**Imagine:**
- 📡 The command bridge sends protocols to the ship systems
- 🤖 Every protocol has a specific function and syntax
- ⚡ The systems execute the commands precisely
- 📊 Without protocols, communication would break down

**In programming we learn:**
- The basic protocols (commands) of Python
- How to combine these protocols
- How to create your own protocols (later)

**Every system protocol you learn is like a new tool for your spaceship!**

## System Protocol 1: Command

**What it is:** `print()` is your first command. It displays text on the screen.

**Important:** Text must always be in quotation marks `"` or `'`!

````python
# Example 1: Simple text output
print("Hello World!")
print("Welcome aboard the Nebula-7!")

# Example 2: Multiple lines
print("This is your first mission.")
print("You are now a Commander!")

# Example 3: Empty line for better readability
print()
print("=== SYSTEM STARTING ===")
````

## System Protocol 2: Data Storage

A variable is like a **labelled data storage unit**: you give it a name and can store something inside – and retrieve it again later.

```python
ship_name = "Enterprise"   # the storage is called 'ship_name', inside is the text "Enterprise"
crew = 150                 # the storage is called 'crew', inside is the number 150
```

**The three most common types of values:**

| Type | Technical term | Example | How to recognise |
|------|---------------|---------|------------------|
| Text | **String** | `"Enterprise"` | Quotation marks |
| Whole number | **Integer** | `150` | No quotation marks |
| Yes/No | **Boolean** | `True` / `False` | No quotation marks, capital T or F |

````python
# Example 1: Different variable types
ship_name = "Enterprise"     # Text (String)
crew = 150                   # Number (Integer)
speed = 5                    # Number (Integer)
system_online = True         # Boolean

# Example 2: Output variables
print("Ship name:")
print(ship_name)
print("Crew:")
print(crew)
print("Light speed factor:")
print(speed)

# Example 3: Variables can change
print("Energy before jump: " + str(80) + "%")
energy = 45  # After jump
print("Energy after jump: " + str(energy) + "%")
````

## System Protocol 3

With `+` you can glue texts together – like connecting building blocks.

```python
print("Ship: " + ship_name)
```

**Why do I need `str()`?**

Python treats text and numbers as different *types*. It's like apples and oranges – you can't simply add them together. `str()` converts a number to text so it can be joined:

```python
mission = 5
print("Mission: " + mission)       # ❌ Error – number and text don't match
print("Mission: " + str(mission))  # ✅ works – str() turns 5 into the text "5"
```

````python
# Example 1: Simple combination
pilot = "Kirk"
rank = "Captain"
print("Name: " + pilot)
print("Rank: " + rank)

# Example 2: With numbers (str() needed!)
mission = 5
print("Mission number: " + str(mission))

# Example 3: Complex sentences
print(pilot + " is " + rank + " of Mission " + str(mission))

# Example 4: Multiple pieces of information
print()
print("=== MISSION PROFILE ===")
print("Captain: " + pilot)
print("Mission: " + str(mission))
print("Status: Active")
print("=======================")
````

## System Protocol 4: User Input

`input()` pauses the program and waits for input. The answer is stored as **text (String)**.

```python
name = input("What is your name, Commander? ")
print("Welcome aboard, " + name + "!")
```

> ⚠️ `input()` only works in a running Python environment (local or JupyterHub). In static previews it will hang.

````python
# Example 1: Simple input
name = input("What is your name, Commander? ")
print("Welcome aboard, " + name + "!")

# Example 2: Multiple inputs
pilot = input("Name of the pilot: ")
ship = input("Name of the ship: ")
print(pilot + " flies the " + ship)
````

## System Protocol 5: Log Entries

With `#` you write a **comment**. Python skips everything after it on that line – comments are just for you (and other crew members), so you can understand the code later.

```python
# This is a comment - it is not executed
print("Hello!")  # Comments can also follow code on the same line
```

````python
# Example 1: Comment above the code
# This line prints the ship's name
print("Enterprise")

# Example 2: Comment at the end of a line
print("150 crew members")  # shows the crew size

# Example 3: Comments to switch off code
print("This will run")
# print("This will NOT run")
````
