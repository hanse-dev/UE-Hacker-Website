## System Protocol 1: Definition – The Command Code

**What it is:** A function is a reusable protocol block that you define once and can call again and again.

**The structure of a function:**
```python
def protocol_name(parameter1, parameter2):
    """Description of the protocol"""
    # Protocol code
    return result  # Optional: return a result
```

**Step-by-step explanation:**
1. **`def`** – The command that starts the protocol definition
2. **`protocol_name`** – A unique name for your protocol
3. **`(parameter)`** – System variables (optional)
4. **`:`** – Ends the protocol definition line
5. **`"""Docstring"""`** – System manual (very important!)
6. **Indented code** – The actual protocol code
7. **`return`** – Sends the result back (optional)

**Why so useful:**
- 🎯 Reuse protocols instead of reprogramming them
- 🔍 Organize and structure the system
- ⚡ Fix errors in one place
- 🎲 Break complex systems into small parts

## System Protocol 1b: The Key to Execution

*Defining* a protocol is not enough – it must also be *called*! Calling is its own topic with important rules.

### With and Without Parentheses

**With parentheses `function()`:** The protocol is executed! The parentheses tell Python: "Run this function now".

**Without parentheses `function`:** The protocol is NOT executed! You are only referencing the function object – nothing happens.

### Calling with Parameters

When calling, you pass **arguments** inside the parentheses: `function(argument1, argument2)`. The order must match!

### From Argument to Variable

**At the call:** You pass an **argument** (e.g. `"Alex"`)

**Inside the function:** It becomes a **variable** with the parameter name (e.g. `name`)

The parameter is a **placeholder**. When called, Python replaces it with the passed value – which is then available as a variable inside the function.

````python
# With vs. without parentheses
def system_greeting():
    """Greets visitors on the space station"""
    print("Welcome aboard Nebula-7!")

# WITH parentheses: function is executed
print("=== With parentheses system_greeting() ===")
system_greeting()

# WITHOUT parentheses: only a reference, nothing happens
print("\n=== Without parentheses system_greeting ===")
print("Type:", type(system_greeting))
print("(The function was not executed!)")
````

````python
# Argument becomes a variable: step by step
def greet_pilot(name):  # "name" is the parameter (placeholder)
    """Greets a pilot by name"""
    # Inside the function, "name" is now a variable with the passed value!
    print(f"Inside the function: name = '{name}'")
    print(f"Hello {name}!")

# At the call: "Alex" is the argument
print("=== Call: greet_pilot('Alex') ===")
greet_pilot("Alex")  # → name becomes "Alex"

print("\n=== Call: greet_pilot('Zara') ===")
greet_pilot("Zara")  # → name becomes "Zara"
````

````python
# Example 1: Simple function without parameters
def system_greeting():
    """Greets visitors on the space station"""
    print("Welcome aboard Nebula-7!")
    print("Your mission begins now!")

# Call the function
print("=== First Visitor ===")
system_greeting()

print("\n=== Second Visitor ===")
system_greeting()

print("\n=== What happens here? ===")
print("1. def system_greeting(): defines the function")
print("2. system_greeting() calls the function")
print("3. The code inside the function runs each time")
````

````python
# Example 2: Function with one parameter
def greet_pilot(name):
    """Greets a pilot by name"""
    print(f"Hello {name}!")
    print(f"Good luck on your mission, {name}!")

# Call the function with different names
print("=== Various Pilots ===")
greet_pilot("Alex")
greet_pilot("Zara")
greet_pilot("Nova")

print("\n=== What happens here? ===")
print("1. name is a placeholder for the passed value")
print("2. Each call passes a different name")
print("3. The function works with the passed value")
````

````python
# Example 3: Function with multiple parameters
def create_ship_profile(name, ship_class, crew):
    """Creates a ship profile with name, class, and crew"""
    print(f"Ship Profile:")
    print(f"  Name: {name}")
    print(f"  Class: {ship_class}")
    print(f"  Crew: {crew} members")
    print(f"  Status: Ready for deployment!")

# Create different ship profiles
print("=== Ship Profiles ===")
create_ship_profile("Nebula-Explorer", "Research Vessel", 150)
print()
create_ship_profile("Star-Fighter", "Fighter", 2)
print()
create_ship_profile("Cargo-Hauler", "Freighter", 25)
````

## 🔍 The Secret of return – The System Result

**return** is the most powerful part of a function!

**What return does:**
- Ends the function immediately
- Sends a value back to the caller
- That value can be stored in a variable

**Without return:** Function does something (side effect)
**With return:** Function calculates and returns a result

````python
# 🔍 Exercises with return

# Example 1: Simple calculation with return
def calculate_warp_distance(warp_factor, time):
    """Calculates the distance traveled in warp"""
    distance = warp_factor * time * 1000
    return distance

# Store result in a variable
print("=== Warp Distance Calculation ===")
distance1 = calculate_warp_distance(5, 10)
print(f"Jump 1: {distance1} light-years")

distance2 = calculate_warp_distance(8, 15)
print(f"Jump 2: {distance2} light-years")

# Use result directly
print(f"Emergency jump: {calculate_warp_distance(10, 5)} light-years")
````

````python
# Example 2: Function with conditional return
def check_system_status(energy_level):
    """Checks whether the system is ready for a mission"""
    if energy_level >= 75:
        return "System ready for mission!"
    else:
        return f"Still {75 - energy_level}% energy needed"

print("=== System Status Check ===")
status1 = check_system_status(60)
print(f"Energy 60%: {status1}")

status2 = check_system_status(90)
print(f"Energy 90%: {status2}")

status3 = check_system_status(70)
print(f"Energy 70%: {status3}")
````

````python
# Example 3: Function without explicit return (None)
def show_equipment():
    """Shows the standard equipment (returns nothing)"""
    print("Your equipment:")
    print("  - Phaser")
    print("  - Tricorder")
    print("  - Communicator")
    # No return – automatically returns None

print("=== Show Equipment ===")
result = show_equipment()
print(f"Return value: {result}")
print("Functions without return always return None!")
````

## System Protocol 2: The System Manuals

**What it is:** Docstrings are documentation strings that explain what a function does.

**Why extremely important:**
- 📖 Others (and you later) understand your function
- 🔍 IDEs show you help when calling
- 🤖 Tools can automatically generate documentation
- ⭐ Sign of professional code

**The three types of docstrings:**
1. **Single-line:** `"""Short description"""`
2. **Multi-line:** `"""Short description\n\nDetails..."""`
3. **Full:** `"""Description\n\nArgs:\nReturns:\n"""`

````python
# Example 1: Different docstring styles
def simple_function():
    """This is a single-line docstring."""
    return "Simple"

def detailed_function(ship_name, fleet_id):
    """Creates a mission description.
    
    Args:
        ship_name (str): The name of the ship
        fleet_id (int): The fleet ID
    
    Returns:
        str: A formatted description
    """
    return f"{ship_name} belongs to fleet {fleet_id}."

# Accessing a docstring (this is how it works!)
print("=== Docstring Examples ===")
print(f"Simple docstring: {simple_function.__doc__}")
print()
print(f"Detailed docstring: {detailed_function.__doc__}")
````

````python
# Example 2: Using the help function
print("=== Help for Functions ===")
help(detailed_function)
````

## System Protocol 3: The Rules of the Code

**Good function names are essential!**

**The rules:**
- **Lowercase letters** with underscores: `calculate_warp_distance`
- **Descriptive**: What the function **does** (verb first)
- **Short but clear**: `get_name()` not `retrieve_starship_data_from_database()`
- **No nouns first**: `name_get()` wrong, `get_name()` right

**Examples:**
- ✅ `calculate_warp_distance()`
- ✅ `create_ship_profile()`
- ✅ `check_system_status()`
- ❌ `CalculateWarpDistance()` (CamelCase)
- ❌ `calculate_the_warp()` (too long)
- ❌ `warp_calculate()` (noun first)

````python
# ✨ Examples of good and bad names

# ✅ Good names (clear and descriptive)
def calculate_shield_efficiency(energy, area):
    """Calculates the efficiency of the shields."""
    return (energy / area) * 100

def find_nearest_dock(ship_position, dock_list):
    """Finds the nearest dock to the current position."""
    return min(dock_list, key=lambda d: abs(d - ship_position))

def is_communication_possible(range_, distance):
    """Checks whether communication is possible."""
    return distance <= range_

# ❌ Bad names (unclear or wrong convention)
# def CalculateShieldEfficiency():  # CamelCase
# def se(e, f):  # Too short, not descriptive
# def shield_calculation():  # Noun first

print("=== Good Function Names in Action ===")
efficiency = calculate_shield_efficiency(1000, 50)
print(f"Shield efficiency: {efficiency}%")

position = 100
docks = [50, 200, 350, 500]
nearest_dock = find_nearest_dock(position, docks)
print(f"Nearest dock: {nearest_dock}")

comm_ok = is_communication_possible(1000, 750)
print(f"Communication possible: {comm_ok}")
````
