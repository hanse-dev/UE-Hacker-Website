## Lesson 1

**What it is:** A riding arena is a rectangular area with markings and lines.

**Important:** Use coordinates for precise placement!

```python
# Example 1: A simple riding arena
import turtle

screen = turtle.Screen()
screen.title("Tournament Ground")
screen.bgcolor("green")

arena = turtle.Turtle()
arena.speed(5)
arena.color("white")
arena.pensize(3)

# Draw outer arena
arena.penup()
arena.goto(-200, -150)
arena.pendown()
for _ in range(2):
    arena.forward(400)
    arena.left(90)
    arena.forward(300)
    arena.left(90)

# Centre line
arena.penup()
arena.goto(0, -150)
arena.pendown()
arena.goto(0, 150)

turtle.done()
```

## Lesson 2

**What it is:** Obstacles are the challenges in a tournament.

**Important:** Use functions for repeated obstacles!

```python
# Example 2: Drawing obstacles
import turtle

screen = turtle.Screen()
screen.bgcolor("lightgreen")

obstacle_pen = turtle.Turtle()
obstacle_pen.speed(3)
obstacle_pen.color("brown")

def draw_obstacle(x, y, height):
    obstacle_pen.penup()
    obstacle_pen.goto(x, y)
    obstacle_pen.pendown()
    obstacle_pen.begin_fill()
    obstacle_pen.fillcolor("saddlebrown")
    for _ in range(2):
        obstacle_pen.forward(40)
        obstacle_pen.left(90)
        obstacle_pen.forward(height)
        obstacle_pen.left(90)
    obstacle_pen.end_fill()

draw_obstacle(-150, -100, 60)
draw_obstacle(-50, -100, 80)
draw_obstacle(50, -100, 70)
draw_obstacle(150, -100, 90)

turtle.done()
```

## Lesson 3

**What it is:** Horses as moving objects on the track.

**Important:** Use `shape()` for Turtle shapes!

```python
# Example 3: Horse on the track
import turtle
import time

screen = turtle.Screen()
screen.bgcolor("green")

horse = turtle.Turtle()
horse.shape("turtle")
horse.color("black")
horse.shapesize(2, 1.5)
horse.penup()

horse.goto(-180, 0)

for _ in range(10):
    horse.forward(40)
    time.sleep(0.5)

turtle.done()
```
