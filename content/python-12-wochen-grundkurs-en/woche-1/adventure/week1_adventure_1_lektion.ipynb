## Spell Formula 1: Incantation

**What it is:** `print()` is your first command. It displays text on the screen.

**Important:** Text must always be in quotation marks `"` or `'`!

````python
# Example 1: Simple text output
print("Hello, world!")
print("Welcome to Pyralia!")

# Example 2: Multiple lines
print("This is your first day.")
print("You are now a hero!")

# Example 3: Empty line for better readability
print()
print("=== YOUR ADVENTURE BEGINS ===")
````

## Spell Formula 2: Magic Chest

A variable is like a **labelled magic chest**: you give it a name and can store something inside – and look it up again later.

```python
hero_name = "Aria"    # the chest is called 'hero_name', inside is the text "Aria"
hero_level = 1        # the chest is called 'hero_level', inside is the number 1
```

**The three most common types of values:**

| Type | Technical term | Example | How to recognise |
|------|---------------|---------|------------------|
| Text | **String** | `"Aria"` | Quotation marks |
| Whole number | **Integer** | `1` | No quotation marks |
| Yes/No | **Boolean** | `True` / `False` | No quotation marks, capital T or F |

````python
# Example 1: Different variable types
hero_name = "Aria"        # Text (String)
hero_level = 1            # Number (Integer)
hero_gold = 50            # Number (Integer)
hero_warrior = True       # Boolean

# Example 2: Output variables
print("Hero name:")
print(hero_name)
print("Level:")
print(hero_level)
print("Gold:")
print(hero_gold)

# Example 3: Variables can change
print("XP before battle: " + str(0) + " XP")
hero_xp = 150  # After first battle
print("XP after battle: " + str(hero_xp) + " XP")
````

## Spell Formula 3

With `+` you can glue texts together – like connecting building blocks.

```python
print("Hero: " + hero_name)
```

**Why do I need `str()`?**

Python treats text and numbers as different *types*. It's like apples and oranges – you can't simply add them together. `str()` converts a number to text so it can be joined:

```python
level = 5
print("Level: " + level)       # ❌ Error – number and text don't match
print("Level: " + str(level))  # ✅ works – str() turns 5 into the text "5"
```

````python
# Example 1: Simple combination
name = "Legolas"
weapon = "Bow"
print("Hero: " + name)
print("Weapon: " + weapon)

# Example 2: With numbers (str() needed!)
arrows = 20
print("Arrows: " + str(arrows))

# Example 3: Complex sentences
print(name + " fights with " + weapon + " and has " + str(arrows) + " arrows.")

# Example 4: Multiple pieces of information
print()
print("=== HERO PROFILE ===")
print("Name: " + name)
print("Weapon: " + weapon)
print("Arrows: " + str(arrows))
print("===================")
````

## Spell Formula 4: Ask the user

`input()` pauses the program and waits for input. The answer is stored as **text (String)**.

```python
name = input("What is your hero's name? ")
print("Welcome to Pyralia, " + name + "!")
```

> ⚠️ `input()` only works in a running Python environment (local or JupyterHub). In static previews it will hang.

````python
# Example 1: Simple input
hero = input("What is your hero's name? ")
print("Welcome to Pyralia, " + hero + "!")

# Example 2: Multiple inputs
hero_name = input("Your hero's name: ")
weapon = input("Your weapon: ")
print(hero_name + " wields a " + weapon)
````

## Spell Formula 5: Secret Notes

With `#` you write a **comment**. Python skips everything after it on that line – comments are just for you (and other adventurers), so you can understand the spell later.

```python
# This is a comment - it is not executed
print("Hello!")  # Comments can also follow code on the same line
```

````python
# Example 1: Comment above the code
# This line prints the hero's name
print("Aragorn")

# Example 2: Comment at the end of a line
print("Level 5")  # shows the current level

# Example 3: Comments to switch off code
print("This will run")
# print("This will NOT run")
````
