## 📚 Trainings-Methode 1: Die Funktions-Definition - Der Trainingsplan

**Was es ist:** Eine Funktion ist ein wiederverwendbarer Trainings-Block, den du einmal definierst und immer wieder aufrufen kannst.

**Der Aufbau einer Funktion:**
```python
def trainings_name(parameter1, parameter2):
    """Beschreibung des Trainings"""
    # Trainings-Code
    return ergebnis  # Optional: Ergebnis zurückgeben
```

**Schritt-für-Schritt-Erklärung:**
1. **`def`** - Das Kommando, das die Trainings-Definition startet
2. **`trainings_name`** - Ein eindeutiger Name für dein Training
3. **`(parameter)`** - Trainings-Variablen (optional)
4. **`:`** - Beendet die Trainings-Definition
5. **`"""Docstring"""`** - Trainingshandbuch (sehr wichtig!)
6. **Eingerückter Code** - Der eigentliche Trainings-Code
7. **`return`** - Gibt das Ergebnis zurück (optional)

**Warum so nützlich:**
- 🎯 Trainings wiederverwenden statt neu schreiben
- 🔍 Training organisieren und strukturieren
- ⚡ Fehler an einer Stelle beheben
- 🎲 Komplexe Trainings in kleine Teile zerlegen

## 🏇 Trainings-Methode 1b: Trainings aufrufen - Der Schlüssel zur Anwendung

Eine Trainingsroutine zu *definieren* reicht nicht – sie muss auch *aufgerufen* werden! Das Aufrufen ist ein eigenes Thema mit wichtigen Regeln.

### Mit und ohne Klammern

**Mit Klammern `funktion()`:** Das Training wird ausgeführt! Die Klammern sagen Python: „Führe diese Funktion jetzt aus“.

**Ohne Klammern `funktion`:** Das Training wird NICHT ausgeführt! Du referenzierst nur das Funktionsobjekt – nichts passiert.

### Aufrufen mit Parametern

Beim Aufruf übergibst du **Argumente** in den Klammern: `funktion(argument1, argument2)`. Die Reihenfolge muss stimmen!

### Vom Argument zur Variable

**Beim Aufruf:** Du übergibst ein **Argument** (z.B. `"Anna"`)

**In der Funktion:** Daraus wird eine **Variable** mit dem Parameternamen (z.B. `name`)

Der Parameter ist ein **Platzhalter**. Beim Aufruf ersetzt Python ihn durch den übergebenen Wert – dieser steht dann als Variable in der Funktion zur Verfügung.

````python
# Mit vs. ohne Klammern
def begruessung_reiterhof():
    """Begrüßt Besucher auf dem Reiterhof"""
    print("Willkommen auf dem Reiterhof Sonnental!")

# MIT Klammern: Funktion wird ausgeführt
print("=== Mit Klammern begruessung_reiterhof() ===")
begruessung_reiterhof()

# OHNE Klammern: Nur Referenz, nichts passiert
print("\n=== Ohne Klammern begruessung_reiterhof ===")
print("Typ:", type(begruessung_reiterhof))
print("(Die Funktion wurde nicht ausgeführt!)")
````

````python
# Argument wird zur Variable: Schritt für Schritt
def begruesse_reiter(name):  # "name" ist der Parameter (Platzhalter)
    """Begrüßt einen Reiter mit seinem Namen"""
    # In der Funktion ist "name" jetzt eine Variable mit dem übergebenen Wert!
    print(f"In der Funktion: name = '{name}'")
    print(f"Hallo {name}!")

# Beim Aufruf: "Anna" ist das Argument
print("=== Aufruf: begruesse_reiter('Anna') ===")
begruesse_reiter("Anna")  # → name wird zu "Anna"

print("\n=== Aufruf: begruesse_reiter('Max') ===")
begruesse_reiter("Max")  # → name wird zu "Max"
````

````python
# Beispiel 1: Einfache Funktion ohne Parameter
def begruessung_reiterhof():
    """Begrüßt Besucher auf dem Reiterhof"""
    print("Willkommen auf dem Reiterhof Sonnental!")
    print("Dein Training beginnt jetzt!")

# Funktion aufrufen
print("=== Erster Besucher ===")
begruessung_reiterhof()

print("\n=== Zweiter Besucher ===")
begruessung_reiterhof()

print("\n=== Was passiert hier? ===")
print("1. def begruessung_reiterhof(): definiert die Funktion")
print("2. begruessung_reiterhof() ruft die Funktion auf")
print("3. Der Code in der Funktion wird jedes Mal ausgeführt")
````

````python
# Beispiel 2: Funktion mit einem Parameter
def begruesse_reiter(name):
    """Begrüßt einen Reiter mit seinem Namen"""
    print(f"Hallo {name}!")
    print(f"Viel Erfolg beim Training, {name}!")

# Funktion mit verschiedenen Namen aufrufen
print("=== Verschiedene Reiter ===")
begruesse_reiter("Anna")
begruesse_reiter("Max")
begruesse_reiter("Lena")

print("\n=== Was passiert hier? ===")
print("1. name ist ein Platzhalter für den übergebenen Wert")
print("2. Jeder Aufruf übergibt einen anderen Namen")
print("3. Die Funktion arbeitet mit dem übergebenen Wert")
````

````python
# Beispiel 3: Funktion mit mehreren Parametern
def erstelle_pferd_profil(name, rasse, alter):
    """Erstellt ein Pferd-Profil mit Name, Rasse und Alter"""
    print(f"Pferd-Profil:")
    print(f"  Name: {name}")
    print(f"  Rasse: {rasse}")
    print(f"  Alter: {alter} Jahre")
    print(f"  Status: Bereit für das Training!")

# Verschiedene Pferde-Profile erstellen
print("=== Pferd-Profile ===")
erstelle_pferd_profil("Thunder", "Hannoveraner", 8)
print()
erstelle_pferd_profil("Luna", "Isländer", 5)
print()
erstelle_pferd_profil("Storm", "Quarter Horse", 6)
````

## 🔍 Das Geheimnis von return - Das Trainingsergebnis

**return** ist der mächtigste Teil einer Funktion!

**Was return tut:**
- Beendet die Funktion sofort
- Gibt einen Wert zurück an den Aufrufer
- Dieser Wert kann in einer Variable gespeichert werden

**Ohne return:** Funktion macht etwas (Side Effect)
**Mit return:** Funktion berechnet und gibt Ergebnis zurück

````python
# 🔍 Übungen mit return

# Beispiel 1: Einfache Berechnung mit return
def berechne_trainingszeit(einheiten, dauer_pro_einheit):
    """Berechnet die gesamte Trainingszeit"""
    gesamtzeit = einheiten * dauer_pro_einheit
    return gesamtzeit

# Ergebnis in Variable speichern
print("=== Trainingszeit-Berechnung ===")
zeit1 = berechne_trainingszeit(5, 15)
print(f"Training 1: {zeit1} Minuten")

zeit2 = berechne_trainingszeit(3, 20)
print(f"Training 2: {zeit2} Minuten")

# Ergebnis direkt verwenden
print(f"Intensivtraining: {berechne_trainingszeit(10, 30)} Minuten")
````

````python
# Beispiel 2: Funktion mit bedingtem return
def pruefe_training_level(erfahrung):
    """Prüft, ob ein Pferd für fortgeschrittenes Training bereit ist"""
    if erfahrung >= 50:
        return "Bereit für fortgeschrittenes Training!"
    else:
        return f"Noch {50 - erfahrung} Trainingseinheiten needed"

print("=== Training-Level-Prüfung ===")
status1 = pruefe_training_level(25)
print(f"Erfahrung 25: {status1}")

status2 = pruefe_training_level(75)
print(f"Erfahrung 75: {status2}")

status3 = pruefe_training_level(40)
print(f"Erfahrung 40: {status3}")
````

````python
# Beispiel 3: Funktion ohne expliziten return (None)
def zeige_ausruestung():
    """Zeigt die Standard-Ausrüstung (gibt nichts zurück)"""
    print("Deine Ausrüstung:")
    print("  - Sattel")
    print("  - Zaumzeug")
    print("  - Gerte")
    # Kein return - gibt automatisch None zurück

print("=== Ausrüstung anzeigen ===")
ergebnis = zeige_ausruestung()
print(f"Rückgabewert: {ergebnis}")
print("Funktionen ohne return geben immer None zurück!")
````

## 📦 Trainings-Methode 2: Docstrings - Die Trainingshandbücher

**Was es ist:** Docstrings sind Dokumentations-Strings, die erklären, was eine Funktion tut.

**Warum extrem wichtig:**
- 📖 Andere (und du später) verstehen deine Funktion
- 🔍 IDEs zeigen dir Hilfe beim Aufruf
- 🤖 Tools können automatisch Dokumentation erstellen
- ⭐ Zeichen von professionellem Code

**Die drei Arten von Docstrings:**
1. **Einzeilig:** `"""Kurze Beschreibung"""`
2. **Mehrzeilig:** `"""Kurze Beschreibung\n\nDetails..."""`
3. **Vollständig:** `"""Beschreibung\n\nArgs:\nReturns:\n"""`

````python
# Beispiel 1: Verschiedene Docstring-Stile
def einfache_funktion():
    """Dies ist ein einzeiliger Docstring."""
    return "Einfach"

def detaillierte_funktion(pferd_name, training_level):
    """Erstellt eine Trainings-Beschreibung.
    
    Args:
        pferd_name (str): Der Name des Pferdes
        training_level (int): Das Trainings-Level des Pferdes
    
    Returns:
        str: Eine formatierte Beschreibung
    """
    return f"{pferd_name} hat Trainings-Level {training_level}."

# Docstring aufrufen (so funktioniert es!)
print("=== Docstring-Beispiele ===")
print(f"Einfacher Docstring: {einfache_funktion.__doc__}")
print()
print(f"Detaillierter Docstring: {detaillierte_funktion.__doc__}")
````

````python
# Beispiel 2: Hilfe-Funktion nutzen
print("=== Hilfe für Funktionen ===")
help(detaillierte_funktion)
````

## ✨ Trainings-Methode 3: Schreibregeln - Die Regeln des Reitstalls

**Gute Funktionsnamen sind entscheidend!**

**Die Regeln:**
- **Kleinbuchstaben** mit Unterstrichen: `berechne_trainingszeit`
- **Beschreibend**: Was die Funktion **tut** (Verb)
- **Kurz aber klar**: `get_name()` nicht `retrieve_horse_data_from_stable()`
- **Keine Verben im Namen**: `name_get()` falsch, `get_name()` richtig

**Beispiele:**
- ✅ `berechne_trainingszeit()`
- ✅ `erstelle_pferd_profil()`
- ✅ `pruefe_training_level()`
- ❌ `BerechneTrainingszeit()` (CamelCase)
- ❌ `berechne_das_training()` (zu lang)
- ❌ `training_berechnen()` (Substantiv zuerst)

````python
# ✨ Beispiele für gute und schlechte Namen

# ✅ Gute Namen (klar und beschreibend)
def berechne_futter_menge(gewicht, aktivitaet):
    """Berechnet die tägliche Futtermenge."""
    return gewicht * aktivitaet * 0.02

def finde_bestes_training(pferde_liste):
    """Findet das beste Training für eine Pferdeliste."""
    return max(pferde_liste)

def ist_pferd_verfuegbar(pferd_id, zeitpunkt):
    """Prüft, ob ein Pferd verfügbar ist."""
    return pferd_id not in belegte_pferde[zeitpunkt]

# ❌ Schlechte Namen (unklar oder falsche Konvention)
# def berechneFutterMenge():  # CamelCase
# def fm(g, a):  # Zu kurz, nicht beschreibend
# def feeding_calculation():  # Substantiv zuerst

print("=== Gute Funktionsnamen in Aktion ===")
futter = berechne_futter_menge(500, 1.5)
print(f"Futtermenge: {futter} kg")

trainings = ["Dressur", "Springen", "Western"]
bestes_training = finde_bestes_training(trainings)
print(f"Bestes Training: {bestes_training}")

# Beispiel-Daten für Verfügbarkeitsprüfung
belegte_pferde = {"morgen": ["Thunder", "Luna"], "mittag": []}
verfuegbar = ist_pferd_verfuegbar("Storm", "morgen")
print(f"Pferd verfügbar: {verfuegbar}")
````
