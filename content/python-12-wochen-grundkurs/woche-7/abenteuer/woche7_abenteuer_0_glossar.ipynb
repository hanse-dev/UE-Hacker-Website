| Begriff | Bedeutung | Beispiel |
|---------|-----------|----------|
| `import` | Ein Modul (Werkzeugkasten) laden | `import random` |
| `import ... as ...` | Modul mit Spitznamen laden | `import random as r` |
| `from ... import ...` | Nur bestimmte Zauber laden | `from string import ascii_letters` |
| `random.randint(a, b)` | Zufällige Ganzzahl von a bis b | `random.randint(1, 6)` |
| `random.choice(liste)` | Zufälliges Element aus einer Liste | `random.choice(pfade)` |
| `random.shuffle(liste)` | Liste zufällig mischen (verändert die Liste) | `random.shuffle(karten)` |
| `random.sample(liste, k)` | k zufällige, unterschiedliche Elemente | `random.sample(liste, 3)` |
| `string.ascii_letters` | Alle Buchstaben a-z und A-Z als Text | Zutat für Runen-Passwörter |
| `string.digits` | Alle Ziffern 0-9 als Text | Zutat für Runen-Passwörter |
| `time.sleep(sekunden)` | Programm für eine Zeit pausieren | `time.sleep(2)` |
| `datetime.now()` | Aktuelles Datum und Uhrzeit | `datetime.now().hour` |

```python
import random
import string
import time

# random-Modul
wurf = random.randint(1, 6)
print("Würfel:", wurf)

helden = ["Aria", "Borin", "Lena"]
print("Zufällig gewählt:", random.choice(helden))

random.shuffle(helden)
print("Gemischt:", helden)

# string-Modul
rune = random.choice(string.ascii_letters)
print("Zufällige Rune:", rune)

# time-Modul
print("Zauber wird gewirkt...")
time.sleep(1)
print("Fertig!")
```
