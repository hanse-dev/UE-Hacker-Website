What is wrong? Find and fix the error!

```python
class machine:
    def __init__(id):
        self.id = id

class android(machine):
    def __init__(id, name):
        super().__init__(id)
        self.name = name

unit7 = android("007", "Seven")
print(unit7.id)
```

### 🐛 Bug #3

What is wrong? Find and fix the error!

```python
class data_packet:
    def __init__(size):
        self.size = size

    def __add__(other):
        return self.size + other.size

packet1 = data_packet(1024)
packet2 = data_packet(2048)
total = packet1 + packet2
print(f"Total size: {total} MB")
```
