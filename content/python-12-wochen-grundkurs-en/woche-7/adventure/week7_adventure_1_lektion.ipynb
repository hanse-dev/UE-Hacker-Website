## 📚 Magic Collection 1: The Spell Import - The First Module

**What it is:** A module is a collection of predefined spells that you can load into your program.

**The structure of an import:**
```python
# Import the entire module
import math

# Import with alias
import math as m

# Import only specific spells
from math import pi, sqrt
```

**Step-by-step explanation:**
1. **`import`** - Magic word for loading
2. **`math`** - Name of the library
3. **`as`** - Optional nickname
4. **`from`** - Imports specific spells

**Why so useful:**
- 🎯 No need to reinvent spells
- 🔍 Professional, optimised incantations
- ⚡ Your code stays organised
- 🎲 Access to specialised magic

````python
# Example 1: Importing modules
# Import the entire module
import math
print(f"Pi from math: {math.pi}")

# Import with alias
import random as r
print(f"Random number: {r.randint(1, 10)}")

# Import only specific spells
from math import sqrt, sin, cos
print(f"Square root of 16: {sqrt(16)}")
print(f"Sine 90°: {sin(1.57)}")
print(f"Cosine 0°: {cos(0)}")
````

````python
# Example 2: Comparing different import methods
# Method 1: import math
import math
print("Method 1 - import math:")
print(f"  math.pi: {math.pi}")
print(f"  math.sqrt(25): {math.sqrt(25)}")

# Method 2: import math as m
import math as m
print("\nMethod 2 - import math as m:")
print(f"  m.pi: {m.pi}")
print(f"  m.sqrt(25): {m.sqrt(25)}")

# Method 3: from math import pi, sqrt
from math import pi, sqrt
print("\nMethod 3 - from math import pi, sqrt:")
print(f"  pi: {pi}")
print(f"  sqrt(25): {sqrt(25)}")
````

````python
# Example 3: Importing multiple modules
import math
import random
import time

# Calculate a random flight duration
distance = 1000  # miles
speed = random.uniform(0.1, 0.9)  # fraction of the speed of light
travel_time = distance / speed

print("=== Magical Journey Simulation ===")
print(f"Distance: {distance} miles")
print(f"Speed: {speed:.2f}c")
print(f"Travel time: {travel_time:.1f} hours")
print(f"Waiting 2 seconds...")
time.sleep(2)
print(f"Arrived at destination!")
````

## 📦 Magic Collection 2: The math Module - The Calculation Spell

````python
# 🔍 Exercises with math constants
import math

print("=== Magical Constants ===")
print(f"Pi (π): {math.pi}")
print(f"Euler's number (e): {math.e}")
print(f"Infinity: {math.inf}")
print(f"Tau (2π): {math.tau}")

# Practical application: circle calculations
radius = 10
circumference = 2 * math.pi * radius
area = math.pi * radius ** 2
print(f"\nCircle with r={radius}:")
print(f"  Circumference: {circumference:.2f}")
print(f"  Area: {area:.2f}")
````

````python
# Example 2: Basic mathematical spells
import math

print("=== Mathematical Spells ===")

# Roots and powers
print(f"sqrt(16): {math.sqrt(16)}")
print(f"pow(2, 3): {math.pow(2, 3)}")
print(f"2**3: {2**3}")

# Rounding
number = 3.7
print(f"\nNumber: {number}")
print(f"round({number}): {round(number)}")
print(f"math.ceil({number}): {math.ceil(number)}")
print(f"math.floor({number}): {math.floor(number)}")

# Absolute value
print(f"\nmath.fabs(-5.5): {math.fabs(-5.5)}")
print(f"abs(-5.5): {abs(-5.5)}")
````

## ✨ Magic Collection 3: The random Module - The Random Spell

**The most powerful random spells:**

**Generating numbers:**
- `random.randint(a, b)` - Random integer from a to b
- `random.uniform(a, b)` - Random float from a to b
- `random.random()` - Random number between 0 and 1

**Choosing from lists:**
- `random.choice(list)` - One random element
- `random.choices(list, k=n)` - n random elements
- `random.shuffle(list)` - Shuffle a list

**Distributions:**
- `random.gauss(mu, sigma)` - Normal distribution
- `random.expovariate(lambda)` - Exponential distribution

````python
# Example 1: Generating random numbers
import random

print("=== Random Numbers ===")

# Integers
print(f"Random number 1-6: {random.randint(1, 6)}")
print(f"Random number 0-100: {random.randint(0, 100)}")

# Floats
print(f"Random number 0-1: {random.random():.3f}")
print(f"Random number 10-20: {random.uniform(10, 20):.2f}")

# Reproducible with seeds
random.seed(42)
print(f"With seed 42: {random.randint(1, 100)}")
random.seed(42)
print(f"With seed 42: {random.randint(1, 100)} (same result!)")
````

````python
# Example 2: Choosing from lists
import random

print("=== Random Selection from Lists ===")

# Realms in the kingdom
realms = ["Pyralia", "Aquaria", "Terrania", "Aeris", "Ignis", "Natura"]
print(f"All realms: {realms}")

# One random realm
destination = random.choice(realms)
print(f"Destination realm: {destination}")

# Three random realms
fleet = random.choices(realms, k=3)
print(f"Travel destinations: {fleet}")

# Shuffle list
order = realms.copy()
random.shuffle(order)
print(f"Shuffled order: {order}")
````

````python
# Example 3: Advanced random spells
import random
import math

print("=== Advanced Random Spells ===")

# Normal distribution for energy values
print("Magical energy (normal distribution):")
for i in range(5):
    energy = random.gauss(100, 20)  # mean 100, std dev 20
    print(f"  Mage {i+1}: {max(0, energy):.1f} Mana")

# Exponential distribution for waiting times
print("\nWaiting times (exponential distribution):")
for i in range(5):
    wait = random.expovariate(0.1)  # lambda = 0.1
    print(f"  Wait time {i+1}: {wait:.1f} minutes")

# Random point in a circle
def random_point_in_circle(radius):
    angle = random.uniform(0, 2 * math.pi)
    r = radius * math.sqrt(random.random())
    x = r * math.cos(angle)
    y = r * math.sin(angle)
    return x, y

x, y = random_point_in_circle(10)
print(f"\nRandom point in circle r=10: ({x:.2f}, {y:.2f})")
````
