## Magic Formula 1

**What it is:** A class is like a magic blueprint for objects. It describes what properties and abilities something should have.

**Important:** Class names always start with a capital letter!

```python
# Example 1: Create first class
class Sword:
    pass

# Example 2: Class with attributes
class Wizard:
    name = "Unknown"
    mana = 100

# Example 3: Class with method
class Dragon:
    def roar(self):
        print("ROAR! 🐉")
```

## Magic Formula 2

**What it is:** An object is a concrete instance of a class – like a single sword forged from the blueprint.

**Important:** Every object is unique, even if it comes from the same class!

```python
# Example 1: Create objects
my_sword = Sword()
your_sword = Sword()

# Example 2: Set attributes
gandalf = Wizard()
gandalf.name = "Gandalf"
gandalf.mana = 200

# Example 3: Multiple objects
smaug = Dragon()
falcor = Dragon()
smaug.roar()
```

## Magic Formula 3: __init__

**What it is:** The constructor *(= special method that runs automatically when an object is created)* is a special method that is automatically executed when an object is created.

**Important:** `__init__` always has `self` as its first parameter! (Why `self` is needed is explained in the next section.)

```python
# Example 1: Simple constructor
class Hero:
    def __init__(self, name):
        self.name = name

# Example 2: Constructor with multiple values
class Potion:
    def __init__(self, name, color, strength):
        self.name = name
        self.color = color
        self.strength = strength

# Example 3: Constructor with default values
class Goblin:
    def __init__(self, name="Grog", hp=10):
        self.name = name
        self.hp = hp
```

## 🪞 Why `self`? – The Key to Your Objects

**What is `self`?** `self` is the reference to **the concrete object** that is currently being addressed. When you call `smaug.roar()`, `self` inside the method is exactly the object `smaug`.

**Why do we need it at all?** A class can create many objects. Python needs to know: *which* of these objects is meant? With 10 dragons the method needs to know: "I am working for THIS dragon here."

**In the constructor `__init__`:**
- `self.name = name` means: **Save** the passed `name` on THIS object.
- Without `self` Python would not know where the variable belongs – it would only exist locally and disappear after the call.

**In methods:**
- `self.name` means: **Read the name** from THIS object.
- Without `self` the method could not access the object's attributes – it would not know whether it's Smaug, Falcor or another dragon.

**How the call works in the background:**
- `smaug.roar()` is translated by Python to `Dragon.roar(smaug)`.
- The first argument is always the object itself – that's why the parameter is called `self`.
