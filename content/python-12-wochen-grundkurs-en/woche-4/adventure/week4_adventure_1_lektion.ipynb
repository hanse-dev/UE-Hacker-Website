## Spell 1: The for Loop – Circle of Things

A `for` loop runs code for **every element** in a collection. The most natural pattern is to iterate over a list:

```python
for element in list_:
    # Code for every element
```

**Step by step:**
1. Python takes the first element from the list
2. Stores it in `element`
3. Runs the indented code
4. Repeats that for every further element

`range(n)` is a special case – it creates a sequence of numbers (0, 1, 2 … n-1) when you don't have an actual list but need a specific number of repetitions.

````python
# Example 1: Simple for loop with range()
print("=== Example 1: Counting from 0 to 4 ===")
# range(5) creates the numbers 0, 1, 2, 3, 4
for i in range(5):
    print(f"Run {i}: The value of i is {i}")

print("\n=== What happens here? ===")
print("1. range(5) creates: [0, 1, 2, 3, 4]")
print("2. i is set to the next element each run")
print("3. The loop ends after the last element")
````

````python
# Example 2: For loop with start and end
print("=== Example 2: Counting from 3 to 7 ===")
# range(3, 8) creates the numbers 3, 4, 5, 6, 7 (8 is exclusive!)
for number in range(3, 8):
    print(f"Number: {number}")

print("\n=== Important observation ===")
print("range(start, end) includes start, but not end!")
print("Just like slices: [start:end] is end-exclusive")
````

````python
# Example 3: For loop with step size
print("=== Example 3: Every other number from 0 to 10 ===")
# range(0, 11, 2) creates: 0, 2, 4, 6, 8, 10
for even in range(0, 11, 2):
    print(f"Even number: {even}")

print("\n=== The step size ===")
print("range(start, end, step) jumps 'step' positions each time")
print("Here: start=0, end=11, step=2 → 0, 2, 4, 6, 8, 10")
````

````python
# Example 4: For loop over a string
print("=== Example 4: Letters of a name ===")
name = "Aria"
print(f"The name is: {name}")
print("Letter by letter:")

for letter in name:
    print(f"  - {letter}")

print("\n=== What happens here? ===")
print("The string is treated as a sequence of letters")
print("Each letter is visited one at a time")
````

````python
# Example 5: For loop over a list
print("=== Example 5: Searching the inventory ===")
inventory = ["Sword", "Shield", "Potion", "Key"]
print("Your inventory:")

for item in inventory:
    print(f"  - {item}")

print("\n=== With index and value ===")
# enumerate gives us both index and value
for index, item in enumerate(inventory):
    print(f"  Slot {index}: {item}")
````

## 🔍 The Secret of range() – The Number Spell

**range() is your most powerful ally for number sequences!**

### The three forms of range():

1. **`range(n)`** – Numbers from 0 to n-1
   ```python
   range(5) → 0, 1, 2, 3, 4
   ```

2. **`range(start, end)`** – Numbers from start to end-1
   ```python
   range(2, 7) → 2, 3, 4, 5, 6
   ```

3. **`range(start, end, step)`** – Numbers with gaps
   ```python
   range(1, 10, 2) → 1, 3, 5, 7, 9
   ```

**Important rules:**
- The **upper bound is always exclusive** (not counted)
- **Negative steps** are possible for counting backwards
- range() is **very memory-efficient** (generates numbers only when needed)

````python
# 🔍 Exercises with range()

# Exercise 1: countdown
print("=== Countdown from 5 to 1 ===")
for i in range(5, 0, -1):
    print(f"{i}...")
print("🚀 Launch!")

# Exercise 2: Find even numbers
print("\n=== Even numbers up to 20 ===")
for number in range(0, 21, 2):
    print(number, end=" ")
print()

# Exercise 3: Multiples of 3
print("\n=== Multiples of 3 up to 30 ===")
for i in range(3, 31, 3):
    print(f"3 × {i//3} = {i}")
````

## Spell 2: The while Loop – Guardian of the Condition

**What it is:** A while loop repeats code as long as a condition is true.

**Structure of a while loop:**
```python
while condition:
    # Code that repeats as long as condition is true
    # IMPORTANT: The condition must become false at some point!
```

**⚠️ Warning:** Infinite loops occur when the condition never becomes false!

````python
# Example 1: Simple while loop
print("=== Example 1: Counting to 5 ===")
counter = 0
while counter <= 5:
    print(f"Counter: {counter}")
    counter += 1  # IMPORTANT: Increment, otherwise infinite loop!

print("\n=== What happens here? ===")
print("1. counter starts at 0")
print("2. As long as counter <= 5, the code runs")
print("3. counter += 1 increases the value each run")
print("4. At counter = 6 the condition is false → loop ends")
````

````python
# Example 2: Game – Health and Energy
print("=== Example 2: Game Loop ===")
health = 100
round_ = 1

while health > 0:
    damage = 20
    health -= damage
    print(f"Round {round_}: -{damage} health, remaining: {health}")
    round_ += 1

print(f"\n💀 Game over after {round_-1} rounds!")
````

````python
# Example 3: Validate user input
print("=== Example 3: Password check ===")
password = ""
attempts = 0
max_attempts = 3

while password != "Secret" and attempts < max_attempts:
    password = input("Enter the password: ")
    attempts += 1
    
    if password == "Secret":
        print("✅ Access granted!")
    elif attempts >= max_attempts:
        print("❌ Too many attempts!")
    else:
        print(f"Wrong! {max_attempts - attempts} attempts remaining")
````

## Spell 3: List Basics – Your First Collection

A list stores multiple values under one name. This is what makes `for` loops truly useful.

```python
inventory = ["Sword", "Shield", "Potion"]  # Create list
inventory.append("Key")                    # Add element
print(inventory[0])                        # First element: "Sword"
print(len(inventory))                      # Count: 4
```

**The essentials:**
- Square brackets `[]` – create a list
- `.append(value)` – add an element to the end
- `list_[0]` – first element (counting starts at 0)
- `len(list_)` – number of elements

> 💡 All further list methods will come in Week 6.

````python
# Create and fill lists
inventory = ["Sword", "Shield", "Potion"]
print(f"Inventory: {inventory}")
print(f"Count: {len(inventory)}")

# Add element
inventory.append("Key")
print(f"After append: {inventory}")

# Access by index (counting starts at 0!)
print(f"First element:  {inventory[0]}")
print(f"Second element: {inventory[1]}")
print(f"Last element:   {inventory[-1]}")

# for loop over a list
print("\nYour inventory:")
for item in inventory:
    print(f"  - {item}")
````
