## 📚 Kapitel 1: Der Zauber-Import – Fremde Magie leihen

**Was es ist:** Ein Modul ist eine Sammlung von vordefinierten Zaubern (Funktionen), die andere Magier bereits geschrieben haben. Mit `import` leihst du sie dir aus.

**Der Aufbau eines Imports:**
```python
# Ganzes Modul importieren
import random

# Mit Spitzname importieren
import random as r

# Nur bestimmte Zauber importieren
from string import ascii_letters
```

**Schritt-für-Schritt-Erklärung:**
1. **`import`** - Zauberwort zum Laden
2. **`random`** - Name der Bibliothek
3. **`as`** - Optionaler Spitzname
4. **`from ... import ...`** - Importiert nur einzelne Zauber

**Warum das dem Archivar wichtig ist:**
- 🎯 Zauber nicht neu erfinden müssen
- 🔍 Geprüfte, zuverlässige Sprüche verwenden
- ⚡ Dein Code bleibt übersichtlich
- 🗝️ Zugriff auf spezialisierte Magie, die dir sonst fehlen würde

````python
# Beispiel 1: Module importieren
import random
import time

print(f"Zufallszahl: {random.randint(1, 10)}")

# Mit Spitzname importieren
import random as r
print(f"Nochmal (über Spitzname r): {r.randint(1, 10)}")

# Nur bestimmte Zauber importieren
from string import ascii_letters
print(f"Erste 5 Buchstaben des Runen-Alphabets: {ascii_letters[:5]}")
````

````python
# Beispiel 2: Verschiedene Import-Methoden vergleichen
import random
print("Methode 1 - import random:")
print(f"  random.randint(1, 6): {random.randint(1, 6)}")

import random as r
print("\nMethode 2 - import random as r:")
print(f"  r.randint(1, 6): {r.randint(1, 6)}")

from random import randint
print("\nMethode 3 - from random import randint:")
print(f"  randint(1, 6): {randint(1, 6)}")
````

````python
# Beispiel 3: Mehrere Module gemeinsam nutzen
import random
import time

print("=== Der Archivar prüft deine erste Beschwörung ===")
wartezeit = random.uniform(0.5, 1.5)
print(f"Ein Siegel-Zauber beginnt zu wirken ({wartezeit:.1f} Sekunden)...")
time.sleep(wartezeit)
print("Das erste Siegel leuchtet auf!")
````

## 🎲 Kapitel 2: Das random-Modul – Orakel der Wahrscheinlichkeit

**Die wichtigsten Orakel-Zauber:**

**Zahlen erzeugen:**
- `random.randint(a, b)` - Zufällige Ganzzahl von a bis b
- `random.uniform(a, b)` - Zufällige Kommazahl von a bis b
- `random.random()` - Zufallszahl zwischen 0 und 1

**Aus Sammlungen wählen:**
- `random.choice(liste)` - Ein zufälliges Element
- `random.sample(liste, k)` - k zufällige, unterschiedliche Elemente
- `random.shuffle(liste)` - Liste mischen (verändert die Liste selbst!)

**Reproduzierbare Magie:**
- `random.seed(zahl)` - Macht "zufällige" Ergebnisse wiederholbar (praktisch zum Testen)

````python
# Beispiel 1: Das Würfel-Orakel befragen
import random

print("=== Das Würfel-Orakel ===")
print(f"Wurf 1: {random.randint(1, 6)}")
print(f"Wurf 2: {random.randint(1, 6)}")
print(f"Kommazahl 0-1: {random.random():.3f}")
print(f"Kommazahl 10-20: {random.uniform(10, 20):.2f}")

# Mit Seed reproduzierbar
random.seed(42)
print(f"\nMit Seed 42: {random.randint(1, 100)}")
random.seed(42)
print(f"Mit Seed 42 (nochmal): {random.randint(1, 100)} (gleiches Ergebnis!)")
````

````python
# Beispiel 2: Regale, Schriftrollen und Mischen
import random

regale = ["Regal der Elemente", "Regal der Tiere", "Regal der Sterne", "Regal der Schatten", "Regal der Helden"]
print(f"Alle Regale: {regale}")

# Ein zufälliges Regal
gewaehltes_regal = random.choice(regale)
print(f"Dein Regal: {gewaehltes_regal}")

# Drei verschiedene Schriftrollen ziehen (ohne Wiederholung!)
schriftrollen = random.sample(regale, 3)
print(f"Gezogene Schriftrollen: {schriftrollen}")

# Reihenfolge mischen
reihenfolge = regale.copy()
random.shuffle(reihenfolge)
print(f"Gemischte Lesereihenfolge: {reihenfolge}")
````

## 🔤 Kapitel 3: Das string-Modul – Runenschmiede

**Was es ist:** Das `string`-Modul liefert dir fertige Zeichensammlungen, mit denen du z.B. Runen-Schlüssel (Passwörter) schmieden kannst.

**Die wichtigsten Zeichensammlungen:**
- `string.ascii_letters` - Alle Buchstaben a-z und A-Z
- `string.ascii_lowercase` - Nur Kleinbuchstaben
- `string.ascii_uppercase` - Nur Großbuchstaben
- `string.digits` - Alle Ziffern 0-9

**Kombiniert mit `random.choice()` kannst du daraus zufällige Runen-Schlüssel schmieden!**

````python
# Beispiel: Einen Runen-Schlüssel schmieden
import random
import string

alphabet = string.ascii_letters + string.digits
print(f"Runen-Alphabet ({len(alphabet)} Zeichen): {alphabet}")

# Einzelne Rune ziehen
print(f"\nEine zufällige Rune: {random.choice(alphabet)}")

# Einen ganzen Schlüssel schmieden (8 Runen)
schluessel = "".join(random.choice(alphabet) for _ in range(8))
print(f"Dein Runen-Schlüssel: {schluessel}")
````

## ⏳ Kapitel 4: Zeitzauber – time und datetime

**Was es ist:** Manche Türen im Archiv öffnen sich nur, wenn du wartest – oder nur zu bestimmten Tageszeiten.

**Die wichtigsten Zeit-Zauber:**
- `time.sleep(sekunden)` - Das Programm für eine Weile pausieren
- `time.time()` - Aktuellen Zeitstempel abrufen
- `from datetime import datetime` und `datetime.now()` - Aktuelles Datum und Uhrzeit
- `datetime.now().hour` - Nur die aktuelle Stunde

````python
# Beispiel: Ein Warte-Ritual und die aktuelle Stunde
import time
from datetime import datetime

print("Ein Ritual beginnt zu wirken...")
time.sleep(1)
print("Das Ritual ist vollendet!")

jetzt = datetime.now()
print(f"\nAktuelle Uhrzeit: {jetzt.strftime('%H:%M:%S')}")

if 6 <= jetzt.hour < 18:
    print("Es ist Tag – die Sonnentore des Archivs stehen offen.")
else:
    print("Es ist Nacht – nur die Mondtore des Archivs sind geöffnet.")
````

## 🧮 Kapitel 5: Das math-Modul – Die Rechenkammer

**Was es ist:** Die **Rechenkammer** hält die mächtigsten Zahlen-Zauber des Archivs bereit – Konstanten und Funktionen für alles, was über einfache Grundrechenarten hinausgeht.

**Die wichtigsten Rechenkammer-Werkzeuge:**

**Konstanten:**
- `math.pi` - Die Zahl Pi (3.14159...)
- `math.e` - Eulersche Zahl (2.71828...)
- `math.inf` - Unendlich

**Grundwerkzeuge:**
- `math.sqrt(x)` - Quadratwurzel
- `math.pow(x, y)` - x hoch y
- `math.ceil(x)` - Aufrunden
- `math.floor(x)` - Abrunden
- `math.fabs(x)` - Absolutwert

````python
# 🔍 Übungen mit math-Konstanten
import math

print("=== Mathematische Konstanten ===")
print(f"Pi (π): {math.pi}")
print(f"Eulersche Zahl (e): {math.e}")
print(f"Unendlich: {math.inf}")
print(f"Tau (2π): {math.tau}")

# Praktische Anwendung: Kreisberechnungen
radius = 10
umfang = 2 * math.pi * radius
flaeche = math.pi * radius ** 2
print(f"\nKreis mit r={radius}:")
print(f"  Umfang: {umfang:.2f}")
print(f"  Fläche: {flaeche:.2f}")
````

````python
# Beispiel 2: Mathematische Grundwerkzeuge
import math

print("=== Mathematische Werkzeuge ===")

# Wurzel und Potenzen
print(f"sqrt(16): {math.sqrt(16)}")
print(f"pow(2, 3): {math.pow(2, 3)}")
print(f"2**3: {2**3}")

# Runden
zahl = 3.7
print(f"\nZahl: {zahl}")
print(f"round({zahl}): {round(zahl)}")
print(f"math.ceil({zahl}): {math.ceil(zahl)}")
print(f"math.floor({zahl}): {math.floor(zahl)}")

# Absolutwert
print(f"\nmath.fabs(-5.5): {math.fabs(-5.5)}")
print(f"abs(-5.5): {abs(-5.5)}")
````

````python
# Beispiel 3: Fortgeschrittene Werkzeuge
import math

print("=== Fortgeschrittene Werkzeuge ===")

# Winkelfunktionen (in Bogenmaß!)
winkel_grad = 45
winkel_rad = math.radians(winkel_grad)
print(f"{winkel_grad}° = {winkel_rad:.3f} rad")
print(f"sin({winkel_grad}°): {math.sin(winkel_rad):.3f}")
print(f"cos({winkel_grad}°): {math.cos(winkel_rad):.3f}")
print(f"tan({winkel_grad}°): {math.tan(winkel_rad):.3f}")

# Logarithmen
print(f"\nlog(10): {math.log(10):.3f}")
print(f"log10(100): {math.log10(100)}")
print(f"log2(8): {math.log2(8)}")

# Spezielle Werkzeuge
print(f"\nfactorial(5): {math.factorial(5)}")
print(f"gcd(24, 36): {math.gcd(24, 36)}")
print(f"lcm(12, 15): {math.lcm(12, 15)}")
````
