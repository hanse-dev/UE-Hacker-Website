### 🐛 Bug #2

What is wrong? Find and fix the error!

```python
class animal:
    def __init__(name):
        self.name = name

class horse(animal):
    def __init__(name, breed):
        super().__init__(name)
        self.breed = breed

stormy = horse("Stormy", "Icelandic Horse")
print(stormy.name)
```

### 🐛 Bug #3

What is wrong? Find and fix the error!

```python
class price:
    def __init__(value):
        self.value = value

    def __add__(other):
        return self.value + other.value

price1 = price(1000)
price2 = price(2000)
total = price1 + price2
print(f"Total price: {total} euros")
```
