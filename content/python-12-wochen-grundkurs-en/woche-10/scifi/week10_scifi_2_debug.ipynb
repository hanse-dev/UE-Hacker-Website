### 🐛 Bug #1

What is wrong? Find and fix the bug!

```python
class spaceship:
    def __init__(self, name):
        self.name = name

ship = spaceship()
print(ship.name)
```

### 🐛 Bug #2

What is wrong? Find and fix the bug!

```python
class Cyborg:
    def __init(name, power_level):
        name = name
        power_level = power_level

terminator = Cyborg("T-800", 9000)
print(terminator.name)
```

### 🐛 Bug #3

What is wrong? Find and fix the bug!
