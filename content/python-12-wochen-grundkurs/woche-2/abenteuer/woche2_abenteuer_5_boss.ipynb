### ⭐⭐⭐⭐☆ Boss-Quest 1: Der Schatzraum

**Belohnung:** 500 XP + Schatzschlüssel

Der Schatzraum hat 8 Truhen. Jede muss katalogisiert werden!

**Schritt 1 – Schätze erfassen:**
Erstelle Variablen für 8 Truhen mit Namen (🔥 Feuer-Element/String) und Goldwert (🪨 Erd-Element/Integer) und gib jeder Truhe eine eindeutige Bezeichnung

**Schritt 2 – Schatzprotokoll erstellen:**
Gib eine Überschrift "=== SCHATZRAUM-PROTOKOLL ===" aus und erstelle eine Beschreibung für jede Truhe mit f-Strings

**Schritt 3 – Durchschnitt berechnen:**
Addiere alle Goldwerte und teile die Summe durch 8

**Schritt 4 – Reichste Truhe finden:**
Vergleiche alle Goldwerte und bestimme die Truhe mit dem höchsten Wert

**Schritt 5 – Gesamtwert ermitteln:**
Berechne den Gesamtwert aller Truhen und addiere alle Goldwerte

**Bonus:** Zähle und gib aus, wie viele Truhen unter dem Durchschnitt liegen!

### ⭐⭐⭐⭐⭐ Boss-Quest 2: Die Zauberwerkstatt

**Belohnung:** 600 XP + Zaubermeister-Titel

In der Zauberwerkstatt experimentierst du mit magischen Formeln!

**Schritt 1 – Zauber erstellen:**
Erstelle 5 Zauber mit Name (🔥 Feuer), Stärke (💧 Wasser/Float 0-100), Komplexität (🪨 Erde/1-5), Zauberdauer (🪨 Erde/Sekunden) und nutze passende Datentypen für jede Eigenschaft

**Schritt 2 – Zauberbeschreibungen ausgeben:**
Gib eine Überschrift "=== ZAUBERWERKSTATT-PROTOKOLL ===" aus und erstelle eine Beschreibung für jeden Zauber mit f-Strings

**Schritt 3 – Komplexität summieren:**
Addiere alle Komplexitätswerte der Zauber und speichere die Gesamtzahl in einer Variable

**Schritt 4 – Stärkste Zauber finden:**
Finde den stärksten Zauber (höchste Stärke) und finde den komplexesten Zauber (höchste Komplexität)

**Schritt 5 – Durchschnittsstärke berechnen:**
Addiere alle Stärken und teile durch 5 und speichere den Durchschnitt

**Schritt 6 – Mächtige filtern:**
Bestimme, welche Zauber über dem Durchschnitt liegen und nutze 💨 Luft-Elemente (Boolean) zur Markierung

**Bonus:** Berechne die Effizienz (Stärke / Komplexität) für jeden Zauber!
