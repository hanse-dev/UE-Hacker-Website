# 📖 Week 1 – Glossary
> You can keep this notebook open all week – here you'll find all terms at a glance.

| Term | Meaning | Example |
|------|---------|--------|
| `print()` | Displays text on the screen | `print("Hello")` |
| **Variable** | A container with a name that stores a value | `horse = "Bobby"` |
| **String** | Text – always in quotation marks | `"Hello World"` |
| **Integer** | Whole number without a decimal | `42` |
| **Boolean** | Yes/No value – either `True` or `False` | `ready = True` |
| **Concatenation** | Joining texts with `+` | `"Hi " + horse` |
| `str()` | Converts a number to text | `str(42)` → `"42"` |
| **Comment** | Explanation in code with `#` – not executed | `# This is a comment` |
| **Syntax** | The spelling rules of Python | Don't forget quotation marks |
| **Output** | What appears on the screen | Result of `print()` |
| **Error / Bug** | When the code doesn't work | `NameError`, `TypeError` |
| **Debugging** | Finding and fixing an error | Find bug → write fix |
| `input()` | Reads input from the user (always returns a String) | `name = input("Horse name? ")` |

## Quick Examples

```python
# String – text in quotation marks
horse = "Bobby"
print(horse)

# Integer – number without quotation marks
age = 5
print(age)

# Boolean – True or False
ready = True
print(ready)

# str() – convert number to text
print("My horse is " + str(age) + " years old.")
```
