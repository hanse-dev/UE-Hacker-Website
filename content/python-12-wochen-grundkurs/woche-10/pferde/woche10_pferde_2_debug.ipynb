### 🐛 Bug #1

Was ist falsch? Finde und behebe den Fehler!

```python
class pferd:
    def __init__(self, name):
        self.name = name

mein_pferd = pferd()
print(mein_pferd.name)
```

### 🐛 Bug #2

Was ist falsch? Finde und behebe den Fehler!

```python
class Reiter:
    def __init(name, level):
        name = name
        level = level

maria = Reiter("Maria", 3)
print(maria.name)
```
