## System Protocol 1: String Command

**What it is:** The f-String is the modern method to fuse text and variables into precise commands.

**Important:** Always put an `f` before the quotation marks!

```python
# Example 1: Simple f-Strings
pilot = "Zara"
rank = 5
print(f"Hello {pilot}!")
print(f"Your rank: {rank}")

# Example 2: With calculations
energy_a = 75
energy_b = 25
print(f"Total energy: {energy_a + energy_b}%")

# Example 3: Multiple variables
ship = "Nebula-Falcon"
captain = "Orion"
mission = "Expedition Alpha"
print(f"{ship} under {captain} launches {mission}!")
```

## System Protocol 2: Types

**What it is:** Every piece of information on Nebula-7 has a quantum type – like the four states of matter.

**Important:** `type()` shows you the quantum type of any information!

```python
# Example 1: Strings – text data
ship_name = "Starlight"
ship_class = "Cruiser"
print(f"Ship: {ship_name}")
print(f"Class: {ship_class}")
print(f"Type: {type(ship_name)}")

# Example 2: Integers – whole numbers
crew_count = 250
deck_count = 15
print(f"Crew: {crew_count}, Decks: {deck_count}")
print(f"Type: {type(crew_count)}")

# Example 3: Floats – precise values
speed = 2.5
coordinate_x = 127.35
print(f"Speed: {speed} light years/h")
print(f"Position X: {coordinate_x}")
print(f"Type: {type(speed)}")

# Example 4: Booleans – system states
system_online = True
alarm_active = False
print(f"System online: {system_online}")
print(f"Alarm active: {alarm_active}")
print(f"Type: {type(system_online)}")
```

## System Protocol 3: Operations and Transformations

**What it is:** How you combine, modify, and transform data.

**Important:** Not all data types can be directly combined!

```python
# Example 1: Mathematical operations
planets_a = 8
planets_b = 4
print(f"Total planets: {planets_a + planets_b}")
print(f"Difference: {planets_a - planets_b}")
print(f"Doubled: {planets_a * 2}")
print(f"Per sector: {planets_a / 2}")

# Example 2: String operations
prefix = "Neo"
suffix = "tron"
system_name = prefix + suffix
print(f"System name: {system_name}")
print(f"Signal: {prefix * 3}")

# Example 3: Data transformation
temperature_str = "273"
temperature_int = int(temperature_str)
print(f"Temperature as text: {temperature_str}K (type: {type(temperature_str)})")
print(f"Temperature as number: {temperature_int}K (type: {type(temperature_int)})")

# Example 4: Process sensor data
# Sensor input simulated (later with input())
sensor_value = "42.5"
value_float = float(sensor_value)
print(f"Sensor value {value_float} doubled: {value_float * 2}")
```
