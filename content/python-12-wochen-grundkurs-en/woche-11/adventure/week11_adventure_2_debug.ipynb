### 🐛 Bug #2

What's wrong? Find and fix the bug!

```python
class Animal:
    def __init__(name):
        self.name = name

class Dog(Animal):
    def __init__(name, breed):
        super().__init__(name)
        self.breed = breed

dog = Dog("Bello", "Dachshund")
print(dog.name)
```

### 🐛 Bug #3

What's wrong? Find and fix the bug!

```python
class GoldCoin:
    def __init__(value):
        self.value = value
    
    def __add__(other):
        return self.value + other.value

gold1 = GoldCoin(10)
gold2 = GoldCoin(20)
total = gold1 + gold2
print(total)
```
