### 🐛 Bug #1

**Was passiert:** Python erwartet einen eingerückten Block. Alles was zu einem `if`, `for`, `while` oder `def` gehört, muss mit 4 Leerzeichen eingerückt werden.

```python
# 🐛 Bug #1

level = 10
if level >= 10:
print("Zugriff gewährt!")
print("Willkommen Kommandant!")
```

### 🐛 Bug #2

**Was passiert:** Mit `=` weist man einer Variable einen Wert zu. Zum Vergleichen braucht man `==` (zwei Gleichheitszeichen) – das ist einer der häufigsten Anfängerfehler!

```python
# 🐛 Bug #2

user = "Admin"
if user = "Admin":
    print("Hallo Admin!")
```

### 🐛 Bug #3

**Was passiert:** `elif` braucht immer eine Bedingung: `elif bedingung:`. Ohne Bedingung weiß Python nicht, wann dieser Zweig gelten soll.

```python
# 🐛 Bug #3

status = 25
if status >= 18:
    print("System aktiv")
elif:
    print("System inaktiv")
```
