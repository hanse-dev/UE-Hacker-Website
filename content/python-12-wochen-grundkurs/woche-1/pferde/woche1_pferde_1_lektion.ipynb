## Lektion 1: Ruf

**Was es ist:** `print()` ist dein erster Befehl. Er gibt Text auf dem Bildschirm aus.

**Wichtig:** Text muss immer in Anführungszeichen `"` oder `'` stehen!

````python
# Beispiel 1: Einfache Textausgabe
print("Hallo auf dem Reiterhof!")
print("Willkommen im Sonnental!")

# Beispiel 2: Mehrere Zeilen
print("Dies ist dein erster Tag.")
print("Du bist jetzt eine Pferdefreund:in!")

# Beispiel 3: Leere Zeile für bessere Übersicht
print()
print("=== Dein Reitbeginn ===")
````

## Lektion 2: Futtertröge

Eine Variable ist wie ein **beschrifteter Futtertrog**: Du gibst ihr einen Namen und kannst darin etwas aufbewahren – und später wieder nachschauen.

```python
pferd_name = "Bobby"   # der Trog heißt 'pferd_name', drin liegt der Text "Bobby"
alter = 5              # der Trog heißt 'alter', drin liegt die Zahl 5
```

**Die drei häufigsten Arten von Werten:**

| Art | Fachbegriff | Beispiel | Erkennungszeichen |
|-----|------------|---------|------------------|
| Text | **String** | `"Bobby"` | Anführungszeichen |
| Ganze Zahl | **Integer** | `5` | keine Anführungszeichen |
| Ja/Nein | **Boolean** | `True` / `False` | kein Anführungszeichen, großes T oder F |

````python
# Beispiel 1: Verschiedene Variablentypen
pferd_name = "Bobby"       # Text (String)
pferd_alter = 5            # Zahl (Integer)
pferd_groesse = 150        # Zahl (Integer)
pferd_freundlich = True    # Wahrheitswert (Boolean)

# Beispiel 2: Variablen ausgeben
print("Pferdename:")
print(pferd_name)
print("Alter:")
print(pferd_alter)
print("Größe in cm:")
print(pferd_groesse)

# Beispiel 3: Variablen können sich ändern
print("Futter vor dem Training: " + str(2) + " kg")
futter_menge = 3  # Nach Training mehr Futter
print("Futter nach dem Training: " + str(futter_menge) + " kg")
````

## Lektion 3

Mit `+` kannst du Texte zusammenkleben – wie Bausteine aneinanderlegen.

```python
print("Pferd: " + pferd_name)
```

**Warum brauche ich `str()`?**

Python behandelt Text und Zahlen als verschiedene *Typen*. Das ist wie Äpfel und Orangen – man kann sie nicht einfach zusammenzählen. `str()` verwandelt eine Zahl in Text, sodass sie sich verbinden lässt:

```python
alter = 5
print("Alter: " + alter)       # ❌ Fehler – Zahl und Text passen nicht zusammen
print("Alter: " + str(alter))  # ✅ klappt – str() macht aus 5 den Text "5"
```

````python
# Beispiel 1: Einfache Kombination
name = "Stormy"
rasse = "Haflinger"
print("Pferd: " + name)
print("Rasse: " + rasse)

# Beispiel 2: Mit Zahlen (str() nötig!)
alter = 8
print("Alter: " + str(alter) + " Jahre")

# Beispiel 3: Komplexe Sätze
print(name + " ist ein " + rasse + " und " + str(alter) + " Jahre alt.")

# Beispiel 4: Mehrere Informationen
print()
print("=== PFERDEPROFIL ===")
print("Name: " + name)
print("Rasse: " + rasse)
print("Alter: " + str(alter) + " Jahre")
print("==================")
````

## Lektion 4: Den Benutzer fragen

`input()` hält das Programm an und wartet auf eine Eingabe. Die Antwort wird als **Text (String)** gespeichert.

```python
name = input("Wie heißt dein Pferd? ")
print("Willkommen, " + name + "!")
```

> ⚠️ `input()` funktioniert nur in einer laufenden Python-Umgebung (lokal oder JupyterHub). In statischen Vorschauen bleibt es hängen.

````python
# Beispiel 1: Einfache Eingabe
pferd = input("Wie heißt dein Pferd? ")
print("Willkommen auf dem Hof, " + pferd + "!")

# Beispiel 2: Mehrere Eingaben
reiter = input("Dein Name: ")
pferd = input("Dein Pferd: ")
print(reiter + " reitet auf " + pferd)
````

## Lektion 5: Stallnotizen

Mit `#` schreibst du einen **Kommentar**. Python überspringt alles, was danach in der Zeile steht – Kommentare sind nur für dich (und andere Pferdefreund:innen), damit ihr den Code später versteht.

```python
# Das ist ein Kommentar - er wird nicht ausgeführt
print("Hallo!")  # Kommentare können auch hinter Code stehen
```

````python
# Beispiel 1: Kommentar über dem Code
# Diese Zeile gibt den Namen des Pferdes aus
print("Bobby")

# Beispiel 2: Kommentar am Ende der Zeile
print("5 Jahre alt")  # zeigt das Alter des Pferdes an

# Beispiel 3: Kommentare zum Abschalten von Code
print("Dies wird ausgeführt")
# print("Dies wird NICHT ausgeführt")
````
