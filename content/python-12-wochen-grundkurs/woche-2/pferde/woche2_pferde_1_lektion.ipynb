## Lektion 1: String-Galopp

**Was es ist:** Der f-String ist die moderne Art, Text und Variablen zu schönen Nachrichten zu verbinden.

**Wichtig:** Immer ein `f` vor die Anführungszeichen setzen!

```python
# Beispiel 1: Einfache f-Strings
pferd_name = "Thunder"
alter = 8
print(f"Hallo {pferd_name}!")
print(f"Du bist {alter} Jahre alt!")

# Beispiel 2: Mit Berechnungen
futter_am_morgen = 3
futter_am_abend = 2
print(f"Tägliches Futter: {futter_am_morgen + futter_am_abend} kg")

# Beispiel 3: Mehrere Variablen
reiter = "Sarah"
pferd = "Luna"
disziplin = "Dressur"
print(f"{reiter} und {pferd} trainieren {disziplin}!")
```

## Lektion 2: Typen

**Was es ist:** Jede Information im Reiterhof hat einen Typ - wie die verschiedenen Gangarten eines Pferdes.

**Wichtig:** `type()` zeigt dir den Typ jeder Information!

```python
# Beispiel 1: Strings - Textinformationen
pferd_name = "Windsturm"
rasse = "Hannoveraner"
print(f"Pferd: {pferd_name}")
print(f"Rasse: {rasse}")
print(f"Typ: {type(pferd_name)}")

# Beispiel 2: Integers - Ganze Zahlen
alter = 7
stockmass = 168  # cm
print(f"Alter: {alter} Jahre, Stockmaß: {stockmass} cm")
print(f"Typ: {type(alter)}")

# Beispiel 3: Floats - Kommazahlen
gewicht = 550.5
futter_menge = 4.25
print(f"Gewicht: {gewicht} kg, Futter: {futter_menge} kg")
print(f"Typ: {type(gewicht)}")

# Beispiel 4: Booleans - Wahrheitswerte
hat_sattel = True
ist_besucht = False
print(f"Hat Sattel: {hat_sattel}")
print(f"Ist besucht: {ist_besucht}")
print(f"Typ: {type(hat_sattel)}")
```

## Lektion 3: Operationen und Umwandlungen

**Was es ist:** Wie du Informationen kombinierst, veränderst und umwandelst.

**Wichtig:** Nicht alle Typen können direkt kombiniert werden!

```python
# Beispiel 1: Mathematische Operationen
pferde_a = 12
pferde_b = 8
print(f"Gesamte Pferde: {pferde_a + pferde_b}")
print(f"Unterschied: {pferde_a - pferde_b}")
print(f"Doppelte Ställe: {pferde_a * 2}")
print(f"Pferde pro Box: {pferde_a / 4}")

# Beispiel 2: String-Operationen
name = "Sternen"
zusatz = "prinz"
voll_name = name + zusatz
print(f"Voller Name: {voll_name}")
print(f"Wiehern: {name * 3}")

# Beispiel 3: Typumwandlung
alter_text = "5"
alter_zahl = int(alter_text)
print(f"Alter als Text: {alter_text} (Typ: {type(alter_text)})")
print(f"Alter als Zahl: {alter_zahl} (Typ: {type(alter_zahl)})")

# Beispiel 4: Eingaben verarbeiten
# Eingabe simuliert (später mit input())
eingabe = "15"
zahl = int(eingabe)
print(f"{zahl} kg Heu für {zahl * 2} Tage reichen.")
```
