## 📚 Zauberformel 1: Der `print()`-Zauber

**Was es ist:** `print()` ist dein erster magischer Befehl. Er gibt Text auf dem Bildschirm aus.

**Warum nützlich:**
- 🔮 Lässt magische Schrift erscheinen
- 📜 Schreibt Prophezeiungen und Boten
- 🗺️ Erstellt Karten und Beschreibungen
- 📖 Hält Abenteuer für die Nachwelt fest

**Wichtig:** Text muss immer in Anführungszeichen `"` oder `'` stehen!

````python
# Beispiel 1: Einfache Textausgabe
print("Hallo Welt!")
print("Willkommen in Pythonia!")

# Beispiel 2: Mehrere Zeilen
print("Dies ist dein erstes Programm.")
print("Du bist jetzt ein Programmierer!")

# Beispiel 3: Leere Zeile für bessere Übersicht
print()
print("=== Dein Programm läuft erfolgreich! ===")
````

## 📦 Zauberformel 2: Variablen - Magische Behälter

**Was es ist:** Eine Variable ist wie eine Schatztruhe mit Namen. Du kannst Informationen darin speichern und später wieder abrufen.

**Warum nützlich:**
- 💰 Bewahrt Schätze und Gold auf
- ⚔️ Merkt sich die Namen berühmter Waffen
- 🧪 Speichert Tränke und ihre Wirkung
- 🔄 Kann jederzeit mit neuen Schätzen gefüllt werden

**Wichtig:**
- Text: in Anführungszeichen (`"Gandalf"`)
- Zahlen: ohne Anführungszeichen (`42`)

````python
# Beispiel 1: Verschiedene Variablentypen
held_name = "Aragorn"      # Text (String)
held_level = 1             # Zahl (Integer)
held_gold = 50             # Zahl (Integer)
held_kaempfer = True       # Wahrheitswert (Boolean)

# Beispiel 2: Variablen ausgeben
print("Helden-Name:")
print(held_name)
print("Level:")
print(held_level)
print("Gold:")
print(held_gold)

# Beispiel 3: Variablen können sich ändern
print("Gold vor dem Kampf: " + str(held_gold))
held_gold = 150  # Neuer Wert
print("Gold nach dem Sieg: " + str(held_gold))
````

## ✨ Zauberformel 3: Texte kombinieren

**Was es ist:** Mit `+` kannst du Texte und Variablen zu einer Nachricht verbinden.

**Warum nützlich:**
- 📜 Schreibt epische Heldengeschichten
- 🏰 Erstellt Beschreibungen von Burgen
- ⚔️ Kombiniert Waffennamen mit ihrer Stärke
- 📊 Macht Ausgaben informativer und magischer

**Wichtig:** Zahlen müssen mit `str()` in Text umgewandelt werden!

````python
# Beispiel 1: Einfache Kombination
name = "Legolas"
waffe = "Bogen"
print("Held: " + name)
print("Waffe: " + waffe)

# Beispiel 2: Mit Zahlen (str() nötig!)
pfeile = 20
print("Pfeile: " + str(pfeile))

# Beispiel 3: Komplexe Sätze
print(name + " hat einen " + waffe + " mit " + str(pfeile) + " Pfeilen.")

# Beispiel 4: Mehrere Informationen
print()
print("=== CHARAKTERBOGEN ===")
print("Name: " + name)
print("Waffe: " + waffe)
print("Munition: " + str(pfeile) + " Pfeile")
print("=====================")
````
